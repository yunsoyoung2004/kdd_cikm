```python
import os
import glob
import shutil
import zipfile
import pandas as pd
from IPython.display import FileLink, display

# ============================================================
# 1. 저장 폴더 생성
# ============================================================

EXPORT_DIR = "./POUR_PAPER_RESULTS_EXPORT"
os.makedirs(EXPORT_DIR, exist_ok=True)

# ============================================================
# 2. 핵심 결과 파일 목록
# ============================================================

core_files = {
    # Main results
    "main_pour_result.csv": "./outputs_pour_main_100_200/test_result.csv",
    "main_pour_train_history.csv": "./outputs_pour_main_100_200/train_history.csv",

    "baseline_gru_100_result.csv": "./baseline_main_gru_100/test_result.csv",
    "baseline_stgc_100_result.csv": "./baseline_main_stgc_100/test_result.csv",
    "baseline_transformer_100_result.csv": "./baseline_main_transformer_100/test_result.csv",

    # RQ results
    "rq1_station_observation_bias.csv": "./final_rq_pour/rq_outputs/rq1_station_observation_bias.csv",
    "rq4_station_uncertainty.csv": "./final_rq_pour/rq_outputs/test_station_uncertainty.csv",
    "rq4_hidden_demand_uncertainty_merged.csv": "./final_rq_pour/rq_outputs/rq4_hidden_demand_uncertainty_merged.csv",
    "rq4_top50_hidden_demand.csv": "./final_rq_pour/rq_outputs/rq4_top50_hidden_demand.csv",
    "rq4_top50_observation_bias.csv": "./final_rq_pour/rq_outputs/rq4_top50_observation_bias.csv",

    # Bigger-scale diagnosis if available
    "full300_rq1_station_observation_bias.csv": "./full_scale_300/rq_outputs/rq1_station_observation_bias.csv",
    "full300_top50_hidden_demand.csv": "./full_scale_300/rq_outputs/rq4_top50_hidden_demand.csv",
    "full300_top50_observation_bias.csv": "./full_scale_300/rq_outputs/rq4_top50_observation_bias.csv",
}

for out_name, src in core_files.items():
    if os.path.exists(src):
        shutil.copy2(src, os.path.join(EXPORT_DIR, out_name))
        print("[COPIED]", src)
    else:
        print("[MISSING]", src)

# ============================================================
# 3. 라벨링 고친 메인 테이블 생성
# ============================================================

rows = []

files = [
    ("POUR", "./outputs_pour_main_100_200/test_result.csv"),
    ("STGC", "./baseline_main_stgc_100/test_result.csv"),
    ("GRU", "./baseline_main_gru_100/test_result.csv"),
    ("Transformer", "./baseline_main_transformer_100/test_result.csv"),
]

for_model_order = {
    "POUR": 0,
    "STGC": 1,
    "GRU": 2,
    "Transformer": 3,
}

for model, path in files:
    if os.path.exists(path):
        df = pd.read_csv(path)
        row = df.iloc[0].to_dict()
        row["model"] = model
        row["source_file"] = path
        rows.append(row)

main_table = pd.DataFrame(rows)

cols = ["model", "mae", "rmse", "uncertainty_error_corr", "source_file"]
main_table = main_table[[c for c in cols if c in main_table.columns]]
main_table["order"] = main_table["model"].map(for_model_order)
main_table = main_table.sort_values("order").drop(columns=["order"])

main_table.to_csv(
    os.path.join(EXPORT_DIR, "PAPER_MAIN_TABLE_FIXED.csv"),
    index=False,
    encoding="cp949"
)

display(main_table)

# ============================================================
# 4. 전체 test_result 모아서 라벨링 고친 summary 생성
# ============================================================

result_files = glob.glob("./**/test_result.csv", recursive=True)
summary_rows = []

def infer_model(exp):
    e = exp.lower()
    if "transformer" in e:
        return "Transformer"
    if "stgc" in e:
        return "STGC"
    if "gru" in e:
        return "GRU"
    if "vanilla" in e:
        return "Vanilla"
    if "no_recon" in e:
        return "POUR w/o Recon"
    if "no_uncertainty" in e:
        return "POUR w/o Uncertainty"
    return "POUR"

for f in result_files:
    try:
        df = pd.read_csv(f)
        row = df.iloc[0].to_dict()
        exp = os.path.basename(os.path.dirname(f))
        row["exp"] = exp
        row["model"] = infer_model(exp)
        row["source_file"] = f
        summary_rows.append(row)
    except Exception as e:
        print("[SKIP]", f, e)

summary = pd.DataFrame(summary_rows)

if len(summary) > 0:
    cols = ["model", "exp", "mae", "rmse", "uncertainty_error_corr", "source_file"]
    summary = summary[[c for c in cols if c in summary.columns]]
    summary = summary.sort_values("mae")
    summary.to_csv(
        os.path.join(EXPORT_DIR, "ALL_TEST_RESULT_SUMMARY_FIXED.csv"),
        index=False,
        encoding="utf-8-sig"
    )
    display(summary)

# ============================================================
# 5. README 생성
# ============================================================

readme = """# POUR Paper Result Export

This folder contains core CSV files for the POUR paper experiments.

Main files:
- PAPER_MAIN_TABLE_FIXED.csv
- ALL_TEST_RESULT_SUMMARY_FIXED.csv
- main_pour_result.csv
- main_pour_train_history.csv
- baseline_gru_100_result.csv
- baseline_stgc_100_result.csv
- baseline_transformer_100_result.csv
- rq1_station_observation_bias.csv
- rq4_station_uncertainty.csv
- rq4_hidden_demand_uncertainty_merged.csv
- rq4_top50_hidden_demand.csv
- rq4_top50_observation_bias.csv

Best current result:
POUR main result is from outputs_pour_main_100_200.
"""

with open(os.path.join(EXPORT_DIR, "README.txt"), "w", encoding="utf-8") as f:
    f.write(readme)

# ============================================================
# 6. ZIP 압축
# ============================================================

ZIP_PATH = "./POUR_PAPER_RESULTS_EXPORT.zip"

if os.path.exists(ZIP_PATH):
    os.remove(ZIP_PATH)

with zipfile.ZipFile(ZIP_PATH, "w", zipfile.ZIP_DEFLATED) as z:
    for root, dirs, files in os.walk(EXPORT_DIR):
        for file in files:
            full_path = os.path.join(root, file)
            arcname = os.path.relpath(full_path, EXPORT_DIR)
            z.write(full_path, arcname)

print("\n[DONE] ZIP created:", ZIP_PATH)

display(FileLink(ZIP_PATH))
```

```python
import os
import glob
import subprocess
import pandas as pd
import matplotlib.pyplot as plt

SCRIPT = "bong.py"
LOCAL_DIR = "./hf_data"
OUT_ROOT = "./missing_experiments_best"

os.makedirs(OUT_ROOT, exist_ok=True)

# 현재 best setting 기준
BASE = {
    "max_subway_files": 100,
    "max_cell_files": 200,
    "window": 24,
    "horizon": 1,
    "mask_ratio": 0.1,
    "graph_type": "topk",
    "graph_topk": 10,
    "hidden_dim": 256,
    "latent_dim": 256,
    "num_layers": 2,
    "lambda_recon": 0.5,
    "lambda_unc": 0.05,
    "batch_size": 4,
    "epochs": 30,
    "num_workers": 0,
}


def run(cmd):
    print("\n" + "=" * 100)
    print(cmd)
    print("=" * 100)
    subprocess.run(cmd, shell=True, check=False)


# ============================================================
# 1. Mask robustness
# ============================================================

def run_mask_robustness():
    for mr in [0.1, 0.3, 0.5, 0.7]:
        out = os.path.join(OUT_ROOT, f"mask_{mr}")

        cmd = f"""
python {SCRIPT} \
  --skip_download \
  --local_dir "{LOCAL_DIR}" \
  --output_dir "{out}" \
  --max_subway_files {BASE["max_subway_files"]} \
  --max_cell_files {BASE["max_cell_files"]} \
  --window {BASE["window"]} \
  --horizon {BASE["horizon"]} \
  --mask_ratio {mr} \
  --graph_type {BASE["graph_type"]} \
  --graph_topk {BASE["graph_topk"]} \
  --hidden_dim {BASE["hidden_dim"]} \
  --latent_dim {BASE["latent_dim"]} \
  --num_layers {BASE["num_layers"]} \
  --lambda_recon {BASE["lambda_recon"]} \
  --lambda_unc {BASE["lambda_unc"]} \
  --batch_size {BASE["batch_size"]} \
  --epochs {BASE["epochs"]} \
  --num_workers {BASE["num_workers"]}
"""
        run(cmd)


# ============================================================
# 2. Ablation
# ============================================================

def run_ablation():
    settings = {
        "FULL": {
            "lambda_recon": 0.5,
            "lambda_unc": 0.05,
            "graph_type": "topk",
        },
        "NO_RECON": {
            "lambda_recon": 0.0,
            "lambda_unc": 0.05,
            "graph_type": "topk",
        },
        "NO_UNCERTAINTY": {
            "lambda_recon": 0.5,
            "lambda_unc": 0.0,
            "graph_type": "topk",
        },
        "VANILLA": {
            "lambda_recon": 0.0,
            "lambda_unc": 0.0,
            "graph_type": "topk",
        },
        "NO_GRAPH": {
            "lambda_recon": 0.5,
            "lambda_unc": 0.05,
            "graph_type": "identity",
        },
        "FULL_GRAPH": {
            "lambda_recon": 0.5,
            "lambda_unc": 0.05,
            "graph_type": "full",
        },
    }

    for name, cfg in settings.items():
        out = os.path.join(OUT_ROOT, f"ablation_{name}")

        cmd = f"""
python {SCRIPT} \
  --skip_download \
  --local_dir "{LOCAL_DIR}" \
  --output_dir "{out}" \
  --max_subway_files {BASE["max_subway_files"]} \
  --max_cell_files {BASE["max_cell_files"]} \
  --window {BASE["window"]} \
  --horizon {BASE["horizon"]} \
  --mask_ratio {BASE["mask_ratio"]} \
  --graph_type {cfg["graph_type"]} \
  --graph_topk {BASE["graph_topk"]} \
  --hidden_dim {BASE["hidden_dim"]} \
  --latent_dim {BASE["latent_dim"]} \
  --num_layers {BASE["num_layers"]} \
  --lambda_recon {cfg["lambda_recon"]} \
  --lambda_unc {cfg["lambda_unc"]} \
  --batch_size {BASE["batch_size"]} \
  --epochs {BASE["epochs"]} \
  --num_workers {BASE["num_workers"]}
"""
        run(cmd)


# ============================================================
# 3. Collect results
# ============================================================

def collect_results():
    rows = []

    for f in glob.glob(os.path.join(OUT_ROOT, "**", "test_result.csv"), recursive=True):
        try:
            df = pd.read_csv(f)
            row = df.iloc[0].to_dict()
            exp = os.path.basename(os.path.dirname(f))

            row["exp"] = exp

            if exp.startswith("mask_"):
                row["group"] = "mask_robustness"
                row["mask_ratio"] = float(exp.replace("mask_", ""))
                row["variant"] = "FULL"
            elif exp.startswith("ablation_"):
                row["group"] = "ablation"
                row["mask_ratio"] = BASE["mask_ratio"]
                row["variant"] = exp.replace("ablation_", "")
            else:
                row["group"] = "other"
                row["mask_ratio"] = None
                row["variant"] = exp

            rows.append(row)

        except Exception as e:
            print("[SKIP]", f, e)

    summary = pd.DataFrame(rows)

    if len(summary) == 0:
        print("[NO RESULTS]")
        return

    summary_path = os.path.join(OUT_ROOT, "missing_experiment_summary.csv")
    summary.to_csv(summary_path, index=False, encoding="utf-8-sig")

    print("\n[SUMMARY]")
    print(summary.sort_values(["group", "mae"]))

    print("\n[SAVED]", summary_path)

    # --------------------------------------------------------
    # Mask robustness figure
    # --------------------------------------------------------

    mask_df = summary[summary["group"] == "mask_robustness"].copy()

    if len(mask_df) > 0:
        mask_df = mask_df.sort_values("mask_ratio")

        plt.figure(figsize=(7, 5))
        plt.plot(mask_df["mask_ratio"], mask_df["mae"], marker="o", label="MAE")
        plt.plot(mask_df["mask_ratio"], mask_df["rmse"], marker="s", label="RMSE")
        plt.xlabel("Mask Ratio")
        plt.ylabel("Error")
        plt.title("Robustness under Partial Observation")
        plt.legend()
        plt.tight_layout()
        plt.savefig(os.path.join(OUT_ROOT, "fig_mask_robustness.png"), dpi=200)
        plt.close()

        mask_df.to_csv(
            os.path.join(OUT_ROOT, "PAPER_MASK_ROBUSTNESS_TABLE.csv"),
            index=False,
            encoding="utf-8-sig",
        )

    # --------------------------------------------------------
    # Ablation figure
    # --------------------------------------------------------

    abl_df = summary[summary["group"] == "ablation"].copy()

    if len(abl_df) > 0:
        abl_df = abl_df.sort_values("mae")

        plt.figure(figsize=(9, 5))
        plt.bar(abl_df["variant"], abl_df["mae"])
        plt.ylabel("MAE")
        plt.title("Ablation Study")
        plt.xticks(rotation=25)
        plt.tight_layout()
        plt.savefig(os.path.join(OUT_ROOT, "fig_ablation_mae.png"), dpi=200)
        plt.close()

        if "uncertainty_error_corr" in abl_df.columns:
            plt.figure(figsize=(9, 5))
            plt.bar(abl_df["variant"], abl_df["uncertainty_error_corr"])
            plt.ylabel("Corr(|error|, uncertainty)")
            plt.title("Uncertainty Calibration Ablation")
            plt.xticks(rotation=25)
            plt.tight_layout()
            plt.savefig(os.path.join(OUT_ROOT, "fig_ablation_uncertainty_corr.png"), dpi=200)
            plt.close()

        abl_df.to_csv(
            os.path.join(OUT_ROOT, "PAPER_ABLATION_TABLE.csv"),
            index=False,
            encoding="utf-8-sig",
        )


if __name__ == "__main__":
    run_mask_robustness()
    run_ablation()
    collect_results()

    print("\nDONE.")
```
