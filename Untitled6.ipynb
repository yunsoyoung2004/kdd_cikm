```python
# ============================================================
# Urban MLLM Audit Experiment
# Auto-build station_evidence.csv from existing result resources
# + Run MLLM audit
# ============================================================


import os
import re
import json
import time
import base64
import mimetypes
from pathlib import Path
from typing import Any, Dict, List, Optional, Tuple

import numpy as np
import pandas as pd
from scipy.stats import spearmanr, pearsonr
from tqdm import tqdm
from openai import OpenAI

# ============================================================
# 0. USER SETTINGS
# ============================================================

os.environ["OPENAI_API_KEY"] = "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"

MODEL = "gpt-5.5"   # 안 되면 사용 가능한 vision model명으로 교체
ROOT_DIR = Path(".") # 네 프로젝트 루트. 예: Path("/content/drive/MyDrive/seoul_station_multimodal")

OUT_DIR = ROOT_DIR / "results_mllm_audit"
OUT_DIR.mkdir(parents=True, exist_ok=True)

AUTO_EVIDENCE_CSV = OUT_DIR / "station_evidence_auto.csv"
OUTPUT_JSONL = OUT_DIR / "mllm_audit_outputs.jsonl"

IMAGE_DETAIL = "low"
LIMIT = 20          # 파일럿. 전체 돌릴 땐 None
RESUME = True

CONDITIONS = [
    "image_only",
    "tabular_only",
    "naive_multimodal",
    "structured_multimodal",
]

TARGET_COL = "mobility_minus_subway_gap"
REFERENCE_COL = "hidden_demand_reconstruction_score_norm"

client = OpenAI(api_key=os.environ.get("OPENAI_API_KEY"))

# ============================================================
# 1. AUTO FIND EXISTING CSV FILES
# ============================================================

def find_csv_files(root: Path) -> List[Path]:
    return sorted(root.rglob("*.csv"))

csv_files = find_csv_files(ROOT_DIR)
print(f"Found {len(csv_files)} CSV files.")
for p in csv_files[:30]:
    print("-", p)

# 우선순위 높은 기존 결과 파일명 후보
PRIORITY_PATTERNS = [
    "quad_reconstructed_hidden_demand_result",
    "quad_latent_decomposition",
    "advanced_station_analysis",
    "station_mobility",
    "subway_total",
    "compare",
    "FINAL_PAPER_RESULT_SUMMARY",
    "PAPER_MAIN_TABLE",
]

def score_csv_path(path: Path) -> int:
    name = path.name.lower()
    score = 0
    for i, pat in enumerate(PRIORITY_PATTERNS):
        if pat.lower() in name:
            score += 100 - i * 5
    return score

candidate_csvs = sorted(csv_files, key=score_csv_path, reverse=True)
print("\nTop candidate CSVs:")
for p in candidate_csvs[:15]:
    print(score_csv_path(p), p)

def try_read_csv(path: Path) -> Optional[pd.DataFrame]:
    encodings = ["utf-8-sig", "utf-8", "cp949", "euc-kr"]
    for enc in encodings:
        try:
            return pd.read_csv(path, encoding=enc)
        except Exception:
            pass
    return None

# 가장 그럴듯한 station-level CSV를 고름
def choose_base_station_csv(paths: List[Path]) -> Tuple[Path, pd.DataFrame]:
    best = None
    best_score = -1
    best_df = None

    station_keys = [
        "station", "nearest_station", "MASTER_STATION_ID", "STATION_ID",
        "station_name", "역", "역명"
    ]
    metric_keys = [
        "mobility", "subway", "hidden", "gap", "reconstruction",
        "temporal", "visual", "score"
    ]

    for p in paths:
        df = try_read_csv(p)
        if df is None or len(df) == 0:
            continue

        cols = [str(c) for c in df.columns]
        cols_lower = [c.lower() for c in cols]
        joined = " ".join(cols_lower)

        s = score_csv_path(p)
        s += sum(10 for k in station_keys if k.lower() in joined)
        s += sum(8 for k in metric_keys if k.lower() in joined)

        # station-level이면 대략 100~1000 rows 근처일 가능성 높음
        if 50 <= len(df) <= 1000:
            s += 30

        # 327개 역이면 가산
        if 250 <= len(df) <= 400:
            s += 50

        if s > best_score:
            best = p
            best_df = df
            best_score = s

    if best is None:
        raise RuntimeError("No usable station-level CSV found. Check ROOT_DIR or upload result CSV files.")

    return best, best_df

base_path, base_df = choose_base_station_csv(candidate_csvs)
print("\nSelected base CSV:", base_path)
print("Shape:", base_df.shape)
display(base_df.head())
print(base_df.columns.tolist())

# ============================================================
# 2. AUTO COLUMN MAPPING
# ============================================================

def find_col(df: pd.DataFrame, candidates: List[str]) -> Optional[str]:
    cols = list(df.columns)
    lower_map = {str(c).lower(): c for c in cols}

    # exact lower match
    for cand in candidates:
        if cand.lower() in lower_map:
            return lower_map[cand.lower()]

    # contains match
    for cand in candidates:
        cand_l = cand.lower()
        for c in cols:
            if cand_l in str(c).lower():
                return c

    return None

def minmax(s: pd.Series) -> pd.Series:
    s = pd.to_numeric(s, errors="coerce")
    mn, mx = s.min(), s.max()
    if pd.isna(mn) or pd.isna(mx) or mx == mn:
        return pd.Series(np.nan, index=s.index)
    return (s - mn) / (mx - mn)

def first_available_numeric(df: pd.DataFrame, candidates: List[str]) -> Optional[pd.Series]:
    col = find_col(df, candidates)
    if col is None:
        return None
    return pd.to_numeric(df[col], errors="coerce")

def build_station_evidence(base_df: pd.DataFrame) -> pd.DataFrame:
    df = base_df.copy()

    station_id_col = find_col(df, [
        "station_id", "MASTER_STATION_ID", "STATION_ID", "id", "역id", "역ID"
    ])
    station_name_col = find_col(df, [
        "station_name", "nearest_station", "station", "역명", "역", "name"
    ])

    if station_id_col is None:
        df["station_id"] = np.arange(len(df)).astype(str)
    else:
        df["station_id"] = df[station_id_col].astype(str)

    if station_name_col is None:
        df["station_name"] = df["station_id"].astype(str)
    else:
        df["station_name"] = df[station_name_col].astype(str)

    # Mobility score
    mobility_raw = first_available_numeric(df, [
        "mobility_score", "mobility_norm", "M_i", "M",
        "surrounding_total_flow", "actual_total_flow",
        "mobility_out_sum", "mobility_in_sum",
        "mean_cell_flow", "station_mobility_score"
    ])
    df["mobility_score"] = minmax(mobility_raw) if mobility_raw is not None else np.nan

    # Observed subway score
    subway_raw = first_available_numeric(df, [
        "observed_subway_score", "subway_score", "S_i", "S",
        "subway_total", "subway_total_flow", "total_subway_activity",
        "boarding", "alighting", "승차", "하차",
        "subway_activity_score"
    ])
    df["observed_subway_score"] = minmax(subway_raw) if subway_raw is not None else np.nan

    # Hidden gap score
    hidden_raw = first_available_numeric(df, [
        "hidden_gap_score", "hidden_demand_score", "H_i", "H",
        "distance_weighted_hidden_demand",
        "subway_under_representation_score",
        "hidden_demand_reconstruction_score",
        "hidden_demand_reconstruction_score_norm"
    ])

    if hidden_raw is not None:
        df["hidden_gap_score"] = minmax(hidden_raw)
    else:
        if df["mobility_score"].notna().any() and df["observed_subway_score"].notna().any():
            df["hidden_gap_score"] = minmax(df["mobility_score"] - df["observed_subway_score"])
        else:
            df["hidden_gap_score"] = np.nan

    # Temporal score
    temporal_raw = first_available_numeric(df, [
        "temporal_score", "T_i", "T",
        "mean_temporal_entropy", "period_entropy",
        "temporal_entropy", "peak_period_score",
        "commute_flow", "late_night_flow"
    ])
    df["temporal_score"] = minmax(temporal_raw) if temporal_raw is not None else np.nan

    # Visual morphology score
    visual_raw = first_available_numeric(df, [
        "visual_morphology_score", "visual_score", "V_i", "V",
        "clip_visual_score", "visual_axis", "morphology_score",
        "semantic_score"
    ])
    df["visual_morphology_score"] = minmax(visual_raw) if visual_raw is not None else np.nan

    # Main target: mobility_minus_subway_gap
    gap_raw = first_available_numeric(df, [
        "mobility_minus_subway_gap", "mobility_minus_subway",
        "gap", "G_i", "G"
    ])

    if gap_raw is not None:
        df["mobility_minus_subway_gap"] = pd.to_numeric(gap_raw, errors="coerce")
    else:
        df["mobility_minus_subway_gap"] = df["mobility_score"] - df["observed_subway_score"]

    # Reference reconstruction score
    recon_raw = first_available_numeric(df, [
        "hidden_demand_reconstruction_score_norm",
        "reconstruction_score_norm",
        "quad_reconstruction_norm",
        "quad_reconstruction",
        "R_i", "R",
        "hidden_demand_reconstruction_score"
    ])

    if recon_raw is not None:
        df["hidden_demand_reconstruction_score_norm"] = minmax(recon_raw)
    else:
        # fallback QuadHDR-like score
        df["hidden_demand_reconstruction_score_norm"] = minmax(
            0.45 * df["mobility_score"].fillna(0)
            + 0.25 * df["hidden_gap_score"].fillna(0)
            + 0.20 * df["temporal_score"].fillna(0)
            + 0.10 * df["visual_morphology_score"].fillna(0)
            - df["observed_subway_score"].fillna(0)
        )

    # Station type
    station_type_col = find_col(df, [
        "station_type", "type", "cluster_label",
        "quad_mobility_subway_label", "reconstruction_label"
    ])
    if station_type_col is not None:
        df["station_type"] = df[station_type_col].astype(str)
    else:
        df["station_type"] = "unknown"

    return df

evidence_df = build_station_evidence(base_df)

# ============================================================
# 3. AUTO IMAGE MAPPING
# ============================================================

image_exts = {".png", ".jpg", ".jpeg", ".webp"}
image_files = [p for p in ROOT_DIR.rglob("*") if p.suffix.lower() in image_exts]

print(f"\nFound {len(image_files)} image files.")
for p in image_files[:30]:
    print("-", p)

def normalize_name(x: str) -> str:
    x = str(x).lower()
    x = re.sub(r"\s+", "", x)
    x = re.sub(r"[^0-9a-z가-힣_()-]", "", x)
    return x

def image_kind(path: Path) -> str:
    name = path.name.lower()
    if any(k in name for k in ["sat", "satellite", "위성"]):
        return "satellite"
    if any(k in name for k in ["map", "osm", "road", "지도"]):
        return "map"
    return "unknown"

def find_image_for_station(station_id: str, station_name: str, kind: str) -> Optional[str]:
    sid = normalize_name(station_id)
    sname = normalize_name(station_name)

    candidates = []
    for p in image_files:
        n = normalize_name(p.name)
        k = image_kind(p)
        score = 0

        if kind == k:
            score += 20
        elif k == "unknown":
            score += 2

        if sid and sid in n:
            score += 50
        if sname and sname in n:
            score += 80

        # station name partially, e.g. English/Korean mismatch may fail
        if score > 0:
            candidates.append((score, p))

    if not candidates:
        return None

    candidates = sorted(candidates, key=lambda x: x[0], reverse=True)
    if candidates[0][0] < 40:
        return None

    return str(candidates[0][1])

map_paths = []
sat_paths = []

for _, row in evidence_df.iterrows():
    sid = row["station_id"]
    sname = row["station_name"]
    map_paths.append(find_image_for_station(sid, sname, "map"))
    sat_paths.append(find_image_for_station(sid, sname, "satellite"))

evidence_df["map_image_path"] = map_paths
evidence_df["satellite_image_path"] = sat_paths

print("\nImage mapping status:")
print("map found:", evidence_df["map_image_path"].notna().sum(), "/", len(evidence_df))
print("sat found:", evidence_df["satellite_image_path"].notna().sum(), "/", len(evidence_df))

display(evidence_df[[
    "station_id", "station_name", "map_image_path", "satellite_image_path",
    "mobility_score", "observed_subway_score", "hidden_gap_score",
    "temporal_score", "visual_morphology_score",
    "mobility_minus_subway_gap",
    "hidden_demand_reconstruction_score_norm",
    "station_type"
]].head(20))

# 저장
evidence_cols = [
    "station_id",
    "station_name",
    "map_image_path",
    "satellite_image_path",
    "mobility_score",
    "observed_subway_score",
    "hidden_gap_score",
    "temporal_score",
    "visual_morphology_score",
    "mobility_minus_subway_gap",
    "hidden_demand_reconstruction_score_norm",
    "station_type",
]

station_evidence = evidence_df[evidence_cols].copy()
station_evidence.to_csv(AUTO_EVIDENCE_CSV, index=False, encoding="utf-8-sig")
print("\nSaved auto evidence CSV:", AUTO_EVIDENCE_CSV)

# 이미지가 거의 없으면 image condition은 자동 제외
image_available = (
    station_evidence["map_image_path"].notna().sum() > 0 or
    station_evidence["satellite_image_path"].notna().sum() > 0
)

if not image_available:
    print("\nWARNING: No station-level images were mapped.")
    print("Image-based conditions will be skipped.")
    CONDITIONS = ["tabular_only", "structured_multimodal"]  # structured는 이미지 없이 evidence만으로 실행 가능
else:
    print("\nImage-based conditions enabled.")

# ============================================================
# 4. MLLM HELPER FUNCTIONS
# ============================================================

EVIDENCE_COLUMNS = {
    "mobility_score": "Mobility demand score",
    "observed_subway_score": "Observed subway activity score",
    "hidden_gap_score": "Hidden gap score",
    "temporal_score": "Temporal activity score",
    "visual_morphology_score": "Visual morphology score",
    "mobility_minus_subway_gap": "Mobility-minus-subway gap",
    "hidden_demand_reconstruction_score_norm": "QuadHDR reconstruction score",
    "station_type": "Station type",
}

def image_to_data_url(path: str) -> str:
    path_obj = Path(path)
    if not path_obj.exists():
        raise FileNotFoundError(f"Image file not found: {path}")

    mime_type, _ = mimetypes.guess_type(str(path_obj))
    if mime_type is None:
        mime_type = "image/png"

    with open(path_obj, "rb") as f:
        encoded = base64.b64encode(f.read()).decode("utf-8")

    return f"data:{mime_type};base64,{encoded}"

def build_evidence_text(row: pd.Series, structured: bool) -> str:
    station_name = row.get("station_name", row.get("station_id", "Unknown Station"))

    if structured:
        lines = [
            f"Station: {station_name}",
            "",
            "Structured evidence:",
            f"- Mobility evidence: mobility_score = {row.get('mobility_score')}",
            f"- Observed transit evidence: observed_subway_score = {row.get('observed_subway_score')}",
            f"- Hidden-gap evidence: hidden_gap_score = {row.get('hidden_gap_score')}, "
            f"mobility_minus_subway_gap = {row.get('mobility_minus_subway_gap')}",
            f"- Temporal evidence: temporal_score = {row.get('temporal_score')}",
            f"- Visual morphology evidence: visual_morphology_score = {row.get('visual_morphology_score')}",
            f"- Reference reconstruction score: hidden_demand_reconstruction_score_norm = "
            f"{row.get('hidden_demand_reconstruction_score_norm')}",
            f"- Station type: {row.get('station_type', 'unknown')}",
        ]
    else:
        lines = [f"Station: {station_name}", "", "Station statistics:"]
        for col, desc in EVIDENCE_COLUMNS.items():
            if col in row and not pd.isna(row[col]):
                lines.append(f"- {desc}: {row[col]}")
    return "\n".join(lines)

def build_prompt(row: pd.Series, condition: str) -> str:
    station_name = row.get("station_name", row.get("station_id", "Unknown Station"))

    base = f"""
You are auditing MLLM-assisted urban mobility discovery.

The task is to estimate whether the station area has hidden public-transit demand.
Hidden demand means that actual neighborhood mobility appears high while observed subway usage is relatively low.

Return ONLY valid JSON with these keys:
{{
  "station_name": "{station_name}",
  "hidden_demand_score": integer from 1 to 5,
  "hidden_demand_label": "low" | "medium" | "high",
  "confidence": integer from 1 to 5,
  "primary_evidence": "visual" | "mobility_subway" | "temporal" | "mixed" | "insufficient",
  "uses_visual_reasoning": true or false,
  "uses_mobility_subway_reasoning": true or false,
  "uses_temporal_reasoning": true or false,
  "explanation": "one concise paragraph",
  "risk_note": "one sentence about possible uncertainty, bias, or unsupported inference"
}}

Rules:
- Do not assume that visually dense or urban-looking areas necessarily have hidden transit demand.
- Hidden demand should be supported mainly by mismatch between mobility demand and observed subway use.
- Visual morphology can support interpretation, but it should not override contradictory mobility/subway evidence.
- If evidence is insufficient, lower the confidence and say why.
""".strip()

    if condition == "image_only":
        return base + """

Condition: IMAGE-ONLY.
You are given only map/satellite images.
Use visual morphology cautiously. Do not invent mobility statistics.
""".strip()

    if condition == "tabular_only":
        return base + f"""

Condition: TABULAR-ONLY.
You are given only station-level numerical evidence.

{build_evidence_text(row, structured=False)}
""".strip()

    if condition == "naive_multimodal":
        return base + f"""

Condition: NAIVE MULTIMODAL.
You are given images and raw station-level statistics.

{build_evidence_text(row, structured=False)}
""".strip()

    if condition == "structured_multimodal":
        return base + f"""

Condition: STRUCTURED MULTIMODAL.
You are given images when available and evidence decomposed into mobility, observed subway, hidden-gap, temporal, and visual axes.
Base your judgment primarily on evidence consistency, not visual appearance alone.

{build_evidence_text(row, structured=True)}
""".strip()

    raise ValueError(f"Unknown condition: {condition}")

def extract_json(text: str) -> Dict[str, Any]:
    text = text.strip()
    try:
        return json.loads(text)
    except json.JSONDecodeError:
        pass

    match = re.search(r"\{.*\}", text, flags=re.DOTALL)
    if not match:
        raise ValueError(f"No JSON object found in model output: {text[:500]}")
    return json.loads(match.group(0))

def make_content(row: pd.Series, condition: str) -> List[Dict[str, Any]]:
    prompt = build_prompt(row, condition)
    content = [{"type": "input_text", "text": prompt}]

    use_images = condition in {"image_only", "naive_multimodal", "structured_multimodal"}

    if use_images:
        for col in ["map_image_path", "satellite_image_path"]:
            path = row.get(col, None)
            if isinstance(path, str) and path.strip() and path != "nan" and Path(path).exists():
                content.append({
                    "type": "input_image",
                    "image_url": image_to_data_url(path),
                    "detail": IMAGE_DETAIL,
                })

    return content

def call_mllm(row: pd.Series, condition: str, max_retries: int = 3) -> Dict[str, Any]:
    content = make_content(row, condition)

    for attempt in range(max_retries):
        try:
            response = client.responses.create(
                model=MODEL,
                input=[{"role": "user", "content": content}],
            )
            raw_text = response.output_text
            parsed = extract_json(raw_text)

            parsed["_raw_output"] = raw_text
            parsed["_condition"] = condition
            parsed["_model"] = MODEL
            parsed["_station_id"] = row.get("station_id")
            parsed["_station_name"] = row.get("station_name")
            return parsed

        except Exception as e:
            if attempt == max_retries - 1:
                return {
                    "_condition": condition,
                    "_model": MODEL,
                    "_station_id": row.get("station_id"),
                    "_station_name": row.get("station_name"),
                    "_error": str(e),
                }
            time.sleep(2 * (attempt + 1))

# ============================================================
# 5. RUN MLLM EXPERIMENT
# ============================================================

run_df = station_evidence.head(LIMIT).copy() if LIMIT is not None else station_evidence.copy()

done_keys = set()
if RESUME and OUTPUT_JSONL.exists():
    with open(OUTPUT_JSONL, "r", encoding="utf-8") as f:
        for line in f:
            try:
                item = json.loads(line)
                done_keys.add((str(item.get("_station_id")), item.get("_condition")))
            except Exception:
                pass

print("\nConditions:", CONDITIONS)
print("Run rows:", len(run_df))
print("Already completed:", len(done_keys))

with open(OUTPUT_JSONL, "a", encoding="utf-8") as out:
    for row_idx, row in tqdm(run_df.iterrows(), total=len(run_df), desc="Running MLLM audit"):
        station_id = str(row.get("station_id", row_idx))

        for condition in CONDITIONS:
            key = (station_id, condition)
            if key in done_keys:
                continue

            result = call_mllm(row, condition)

            record = row.to_dict()
            record.update(result)

            out.write(json.dumps(record, ensure_ascii=False) + "\n")
            out.flush()

            done_keys.add(key)

print("\nSaved JSONL:", OUTPUT_JSONL)

# ============================================================
# 6. LOAD RESULTS + EVALUATE
# ============================================================

def read_jsonl(path: Path) -> pd.DataFrame:
    rows = []
    with open(path, "r", encoding="utf-8") as f:
        for line in f:
            if line.strip():
                rows.append(json.loads(line))
    return pd.DataFrame(rows)

res_df = read_jsonl(OUTPUT_JSONL)
flat_csv = OUT_DIR / "mllm_audit_outputs_flat.csv"
res_df.to_csv(flat_csv, index=False, encoding="utf-8-sig")

print("Result shape:", res_df.shape)
print("Saved flat CSV:", flat_csv)

if "_error" in res_df.columns:
    err_df = res_df[res_df["_error"].notna()] if res_df["_error"].notna().any() else pd.DataFrame()
    print("Errors:", len(err_df))
    if len(err_df) > 0:
        display(err_df[["_station_id", "_station_name", "_condition", "_error"]].head())

def safe_numeric(series: pd.Series) -> pd.Series:
    return pd.to_numeric(series, errors="coerce")

def corr_pair(data: pd.DataFrame, x_col: str, y_col: str):
    if x_col not in data.columns or y_col not in data.columns:
        return np.nan, np.nan, 0

    x = safe_numeric(data[x_col])
    y = safe_numeric(data[y_col])
    mask = x.notna() & y.notna()

    if int(mask.sum()) < 3:
        return np.nan, np.nan, int(mask.sum())

    return float(spearmanr(x[mask], y[mask]).correlation), float(pearsonr(x[mask], y[mask])[0]), int(mask.sum())

def topk_overlap(data: pd.DataFrame, score_col: str, target_col: str, k: int = 10) -> float:
    needed = [score_col, target_col, "_station_id"]
    if any(c not in data.columns for c in needed):
        return np.nan

    temp = data[needed].copy()
    temp[score_col] = safe_numeric(temp[score_col])
    temp[target_col] = safe_numeric(temp[target_col])
    temp = temp.dropna(subset=[score_col, target_col, "_station_id"])

    if len(temp) < k:
        return np.nan

    pred_top = set(temp.sort_values(score_col, ascending=False).head(k)["_station_id"].astype(str))
    true_top = set(temp.sort_values(target_col, ascending=False).head(k)["_station_id"].astype(str))

    return len(pred_top & true_top) / k

def bool_series(series: pd.Series) -> pd.Series:
    return series.astype(str).str.lower().isin(["true", "1", "yes", "y"])

def label_accuracy_top_quantile(data: pd.DataFrame, target_col: str, q: float = 0.80) -> float:
    if "hidden_demand_label" not in data.columns or target_col not in data.columns:
        return np.nan

    temp = data[["hidden_demand_label", target_col]].copy()
    temp[target_col] = safe_numeric(temp[target_col])
    temp = temp.dropna()

    if len(temp) == 0:
        return np.nan

    threshold = temp[target_col].quantile(q)
    y_true = temp[target_col] >= threshold
    y_pred = temp["hidden_demand_label"].astype(str).str.lower().eq("high")
    return float((y_true == y_pred).mean())

def unsupported_visual_reasoning_rate(data: pd.DataFrame, target_col: str, q: float = 0.50) -> float:
    if "uses_visual_reasoning" not in data.columns or target_col not in data.columns:
        return np.nan

    temp = data[["uses_visual_reasoning", target_col]].copy()
    temp[target_col] = safe_numeric(temp[target_col])
    temp = temp.dropna()

    if len(temp) == 0:
        return np.nan

    threshold = temp[target_col].quantile(q)
    visual_reason = bool_series(temp["uses_visual_reasoning"])
    unsupported = visual_reason & (temp[target_col] < threshold)
    return float(unsupported.mean())

def rate_bool(data: pd.DataFrame, col: str) -> float:
    if col not in data.columns:
        return np.nan
    return float(bool_series(data[col]).mean())

def primary_visual_rate(data: pd.DataFrame) -> float:
    if "primary_evidence" not in data.columns:
        return np.nan
    return float(data["primary_evidence"].astype(str).str.lower().eq("visual").mean())

summary_rows = []
for condition in CONDITIONS:
    sub = res_df[res_df["_condition"] == condition].copy()
    if len(sub) == 0:
        continue

    sub["hidden_demand_score"] = safe_numeric(sub.get("hidden_demand_score", pd.Series(dtype=float)))

    sp_gap, pe_gap, n_gap = corr_pair(sub, "hidden_demand_score", TARGET_COL)
    sp_ref, pe_ref, n_ref = corr_pair(sub, "hidden_demand_score", REFERENCE_COL)

    summary_rows.append({
        "condition": condition,
        "n": len(sub),
        f"spearman_with_{TARGET_COL}": sp_gap,
        f"pearson_with_{TARGET_COL}": pe_gap,
        f"spearman_with_{REFERENCE_COL}": sp_ref,
        f"pearson_with_{REFERENCE_COL}": pe_ref,
        "top10_overlap_gap": topk_overlap(sub, "hidden_demand_score", TARGET_COL, 10),
        "top10_overlap_reference": topk_overlap(sub, "hidden_demand_score", REFERENCE_COL, 10),
        "label_accuracy_top20pct_gap": label_accuracy_top_quantile(sub, TARGET_COL, 0.80),
        "unsupported_visual_reasoning_rate": unsupported_visual_reasoning_rate(sub, TARGET_COL, 0.50),
        "visual_primary_rate": primary_visual_rate(sub),
        "uses_visual_reasoning_rate": rate_bool(sub, "uses_visual_reasoning"),
        "uses_mobility_subway_reasoning_rate": rate_bool(sub, "uses_mobility_subway_reasoning"),
        "uses_temporal_reasoning_rate": rate_bool(sub, "uses_temporal_reasoning"),
        "valid_corr_n_gap": n_gap,
        "valid_corr_n_reference": n_ref,
    })

summary_df = pd.DataFrame(summary_rows)
summary_path = OUT_DIR / "condition_summary.csv"
summary_df.to_csv(summary_path, index=False, encoding="utf-8-sig")

print("\n=== Condition Summary ===")
display(summary_df)

# station-level robustness
needed = ["_station_id", "_station_name", "_condition", "hidden_demand_score"]
if all(c in res_df.columns for c in needed):
    tmp = res_df[needed].copy()
    tmp["hidden_demand_score"] = safe_numeric(tmp["hidden_demand_score"])
    tmp = tmp.dropna(subset=["_station_id", "_condition", "hidden_demand_score"])

    robust_df = tmp.pivot_table(
        index=["_station_id", "_station_name"],
        columns="_condition",
        values="hidden_demand_score",
        aggfunc="mean",
    ).reset_index()

    condition_cols = [c for c in CONDITIONS if c in robust_df.columns]
    robust_df["score_mean_across_conditions"] = robust_df[condition_cols].mean(axis=1)
    robust_df["score_std_across_conditions"] = robust_df[condition_cols].std(axis=1)
    robust_df["score_range_across_conditions"] = robust_df[condition_cols].max(axis=1) - robust_df[condition_cols].min(axis=1)

    robust_df = robust_df.sort_values("score_std_across_conditions", ascending=False)
else:
    robust_df = pd.DataFrame()

robust_path = OUT_DIR / "condition_robustness_by_station.csv"
robust_df.to_csv(robust_path, index=False, encoding="utf-8-sig")

print("\n=== Robustness by Station ===")
display(robust_df.head(20))

# error cases
if "hidden_demand_score" in res_df.columns and TARGET_COL in res_df.columns:
    temp = res_df.copy()
    temp["hidden_demand_score"] = safe_numeric(temp["hidden_demand_score"])
    temp[TARGET_COL] = safe_numeric(temp[TARGET_COL])
    temp = temp.dropna(subset=["hidden_demand_score", TARGET_COL])

    high_pred = temp["hidden_demand_score"] >= 4
    low_true = temp[TARGET_COL] < temp[TARGET_COL].quantile(0.50)

    visual = bool_series(temp["uses_visual_reasoning"]) if "uses_visual_reasoning" in temp.columns else pd.Series(False, index=temp.index)
    spurious_visual = temp[high_pred & low_true & visual].copy()
    spurious_path = OUT_DIR / "spurious_visual_cases.csv"
    spurious_visual.to_csv(spurious_path, index=False, encoding="utf-8-sig")

    low_pred = temp["hidden_demand_score"] <= 2
    high_true = temp[TARGET_COL] >= temp[TARGET_COL].quantile(0.80)
    missed_hidden = temp[low_pred & high_true].copy()
    missed_path = OUT_DIR / "missed_hidden_demand_cases.csv"
    missed_hidden.to_csv(missed_path, index=False, encoding="utf-8-sig")

    print("\nSpurious visual cases:", len(spurious_visual))
    display(spurious_visual[["_station_id", "_station_name", "_condition", "hidden_demand_score", TARGET_COL, "primary_evidence", "explanation"]].head(10))

    print("\nMissed hidden-demand cases:", len(missed_hidden))
    display(missed_hidden[["_station_id", "_station_name", "_condition", "hidden_demand_score", TARGET_COL, "primary_evidence", "explanation"]].head(10))

# LaTeX table
paper_cols = [
    "condition",
    "n",
    f"spearman_with_{TARGET_COL}",
    f"pearson_with_{TARGET_COL}",
    "top10_overlap_gap",
    "label_accuracy_top20pct_gap",
    "unsupported_visual_reasoning_rate",
    "visual_primary_rate",
    "uses_mobility_subway_reasoning_rate",
    "uses_temporal_reasoning_rate",
]

existing_paper_cols = [c for c in paper_cols if c in summary_df.columns]
latex_table = summary_df[existing_paper_cols].to_latex(index=False, float_format="%.3f", escape=True)

latex_path = OUT_DIR / "condition_summary_latex_table.tex"
with open(latex_path, "w", encoding="utf-8") as f:
    f.write(latex_table)

print("\nSaved files:")
print("- Auto evidence CSV:", AUTO_EVIDENCE_CSV)
print("- JSONL:", OUTPUT_JSONL)
print("- Flat CSV:", flat_csv)
print("- Summary CSV:", summary_path)
print("- Robustness CSV:", robust_path)
print("- LaTeX table:", latex_path)
```

```python
# ============================================================
# FAST FULL Urban MLLM Audit Experiment
# - Auto-build station_evidence from existing CSV resources
# - Run ALL stations with fast parallel API calls
# - Default: all stations × 2 conditions
# - Conditions: tabular_only, structured_multimodal
# ============================================================

import os
import re
import json
import time
import base64
import mimetypes
import threading
from pathlib import Path
from typing import Any, Dict, List, Optional, Tuple
from concurrent.futures import ThreadPoolExecutor, as_completed

import numpy as np
import pandas as pd
from scipy.stats import spearmanr, pearsonr
from tqdm import tqdm
from openai import OpenAI

# ============================================================
# 0. USER SETTINGS
# ============================================================

# 절대 노트북에 API key 직접 쓰지 말기.
# 터미널에서 export OPENAI_API_KEY="..." 해두거나,
# 아래 줄에서 임시 입력만 하되 공유 금지.
# os.environ["OPENAI_API_KEY"] = "YOUR_NEW_KEY_HERE"

if not os.environ.get("OPENAI_API_KEY"):
    raise ValueError("OPENAI_API_KEY가 없습니다. 새 key를 환경변수로 설정하세요.")

client = OpenAI(api_key=os.environ.get("OPENAI_API_KEY"))

# 네 계정에서 사용 가능한 vision-capable 모델명으로 설정
# 기존에 쓰던 모델명이 안 되면 여기만 바꾸면 됨.
MODEL = "gpt-4o-mini"

ROOT_DIR = Path(".")
OUT_DIR = ROOT_DIR / "results_mllm_audit_fast_full"
OUT_DIR.mkdir(parents=True, exist_ok=True)

AUTO_EVIDENCE_CSV = OUT_DIR / "station_evidence_auto.csv"
OUTPUT_JSONL = OUT_DIR / "mllm_audit_outputs_FAST_FULL.jsonl"

# 전체 검사: None
LIMIT = None

# 속도 옵션
MAX_WORKERS = 8          # 너무 높이면 rate limit 가능. 6~10 추천.
IMAGE_DETAIL = "low"     # low가 빠름
USE_ONLY_ONE_IMAGE = True # True면 satellite 우선 1장만 넣음. 훨씬 빠름.

# 기본 빠른 전체실험 조건
# 전체 327개 × 2조건 = 654 calls
CONDITIONS = [
    "tabular_only",
    "structured_multimodal",
]

# 정말 4조건 전부 필요하면 아래로 바꾸면 됨. 대신 느림.
# CONDITIONS = ["image_only", "tabular_only", "naive_multimodal", "structured_multimodal"]

TARGET_COL = "mobility_minus_subway_gap"
REFERENCE_COL = "hidden_demand_reconstruction_score_norm"

RESUME = True
WRITE_LOCK = threading.Lock()

print("MODEL:", MODEL)
print("OUT_DIR:", OUT_DIR)
print("CONDITIONS:", CONDITIONS)
print("MAX_WORKERS:", MAX_WORKERS)

# ============================================================
# 1. CSV DISCOVERY AND BASE TABLE SELECTION
# ============================================================

def read_csv_safe(path: Path) -> pd.DataFrame:
    for enc in ["utf-8-sig", "utf-8", "cp949", "euc-kr"]:
        try:
            return pd.read_csv(path, encoding=enc)
        except Exception:
            pass
    raise RuntimeError(f"Cannot read CSV: {path}")

def find_csv_files(root: Path) -> List[Path]:
    return sorted(root.rglob("*.csv"))

csv_files = find_csv_files(ROOT_DIR)
print(f"Found {len(csv_files)} CSV files.")

PRIORITY_PATTERNS = [
    "08_compare_subway_total_filtered",
    "quad_reconstructed_hidden_demand_result",
    "quad_latent_decomposition",
    "advanced_station_analysis",
    "6_advanced",
    "PAPER_MAIN_TABLE_FIXED",
    "PAPER_MAIN_TABLE",
    "FINAL_PAPER_RESULT_SUMMARY",
    "full300_top50_hidden_demand",
    "rq4_hidden_demand_uncertainty_merged",
]

def score_csv_path(path: Path) -> int:
    name = path.name.lower()
    score = 0
    for i, pat in enumerate(PRIORITY_PATTERNS):
        if pat.lower() in name:
            score += 200 - i * 10
    if ".ipynb_checkpoints" in str(path):
        score -= 200
    return score

def try_read_csv(path: Path) -> Optional[pd.DataFrame]:
    try:
        return read_csv_safe(path)
    except Exception:
        return None

def choose_base_station_csv(paths: List[Path]) -> Tuple[Path, pd.DataFrame]:
    best_path = None
    best_df = None
    best_score = -10**9

    station_keys = [
        "master_station_id", "station_id", "nearest_station", "station_name",
        "nearest_line", "station_type"
    ]
    metric_keys = [
        "mobility_norm", "subway_norm", "mobility_minus_subway_gap",
        "hidden_demand_score", "surrounding_total_flow",
        "subway_total_activity", "mean_temporal_entropy",
        "period_entropy", "directional_mismatch_score"
    ]

    for p in paths:
        df = try_read_csv(p)
        if df is None or len(df) == 0:
            continue

        cols_lower = " ".join([str(c).lower() for c in df.columns])
        score = score_csv_path(p)

        score += sum(20 for k in station_keys if k in cols_lower)
        score += sum(20 for k in metric_keys if k in cols_lower)

        if 250 <= len(df) <= 400:
            score += 100
        elif 50 <= len(df) <= 1000:
            score += 30

        if score > best_score:
            best_path = p
            best_df = df
            best_score = score

    if best_path is None:
        raise RuntimeError("Station-level CSV를 찾지 못했습니다.")

    return best_path, best_df

candidate_csvs = sorted(csv_files, key=score_csv_path, reverse=True)
print("\nTop candidate CSVs:")
for p in candidate_csvs[:20]:
    print(score_csv_path(p), p)

base_path, base_df = choose_base_station_csv(candidate_csvs)
print("\nSelected base CSV:", base_path)
print("Base shape:", base_df.shape)
display(base_df.head())
print("Columns:", base_df.columns.tolist())

# ============================================================
# 2. COLUMN MAPPING
# ============================================================

def find_col(df: pd.DataFrame, candidates: List[str]) -> Optional[str]:
    cols = list(df.columns)
    lower_map = {str(c).lower(): c for c in cols}

    for cand in candidates:
        if cand.lower() in lower_map:
            return lower_map[cand.lower()]

    for cand in candidates:
        cand_l = cand.lower()
        for c in cols:
            if cand_l in str(c).lower():
                return c
    return None

def minmax(s: pd.Series) -> pd.Series:
    s = pd.to_numeric(s, errors="coerce")
    mn, mx = s.min(), s.max()
    if pd.isna(mn) or pd.isna(mx) or mx == mn:
        return pd.Series(np.nan, index=s.index)
    return (s - mn) / (mx - mn)

def numeric_col(df: pd.DataFrame, candidates: List[str]) -> Tuple[Optional[str], Optional[pd.Series]]:
    col = find_col(df, candidates)
    if col is None:
        return None, None
    return col, pd.to_numeric(df[col], errors="coerce")

def build_station_evidence(base_df: pd.DataFrame) -> Tuple[pd.DataFrame, Dict[str, Optional[str]]]:
    df = base_df.copy()
    mapping = {}

    station_id_col = find_col(df, ["MASTER_STATION_ID", "station_id", "STATION_ID", "id"])
    station_name_col = find_col(df, ["nearest_station", "station_name", "station", "역명", "역"])
    line_col = find_col(df, ["nearest_line", "line", "호선"])

    mapping["station_id_col"] = station_id_col
    mapping["station_name_col"] = station_name_col
    mapping["line_col"] = line_col

    df["station_id"] = df[station_id_col].astype(str) if station_id_col else np.arange(len(df)).astype(str)
    df["station_name"] = df[station_name_col].astype(str) if station_name_col else df["station_id"].astype(str)
    df["station_line"] = df[line_col].astype(str) if line_col else ""

    # 이미 norm 컬럼이 있으면 우선 사용
    mobility_col, mobility_raw = numeric_col(df, [
        "mobility_norm",
        "mobility_score",
        "surrounding_total_flow",
        "actual_total_flow",
        "mean_cell_flow",
        "surrounding_outflow",
        "surrounding_inflow",
    ])
    if mobility_col == "mobility_norm":
        df["mobility_score"] = mobility_raw
    else:
        df["mobility_score"] = minmax(mobility_raw) if mobility_raw is not None else np.nan
    mapping["mobility_col"] = mobility_col

    subway_col, subway_raw = numeric_col(df, [
        "subway_norm",
        "observed_subway_score",
        "subway_score",
        "subway_total_activity",
        "subway_out_sum",
        "subway_in_sum",
        "subway_file_count",
    ])
    if subway_col == "subway_norm":
        df["observed_subway_score"] = subway_raw
    else:
        df["observed_subway_score"] = minmax(subway_raw) if subway_raw is not None else np.nan
    mapping["subway_col"] = subway_col

    hidden_col, hidden_raw = numeric_col(df, [
        "hidden_demand_score",
        "hidden_gap_score",
        "distance_weighted_hidden_demand",
        "outer_hidden_demand_score",
        "diverse_living_area_score",
        "mobility_minus_subway_gap",
    ])
    df["hidden_gap_score"] = minmax(hidden_raw) if hidden_raw is not None else minmax(df["mobility_score"] - df["observed_subway_score"])
    mapping["hidden_col"] = hidden_col

    temporal_col, temporal_raw = numeric_col(df, [
        "mean_temporal_entropy",
        "period_entropy",
        "temporal_score",
        "distance_band_entropy",
        "morning_peak_flow",
        "evening_peak_flow",
        "night_flow",
    ])
    df["temporal_score"] = minmax(temporal_raw) if temporal_raw is not None else np.nan
    mapping["temporal_col"] = temporal_col

    # visual score가 없으면 morphology 관련 score fallback
    visual_col, visual_raw = numeric_col(df, [
        "visual_morphology_score",
        "visual_score",
        "morphology_score",
        "diverse_living_area_score",
        "outer_hidden_demand_score",
        "distance_norm",
    ])
    df["visual_morphology_score"] = minmax(visual_raw) if visual_raw is not None else np.nan
    mapping["visual_col"] = visual_col

    gap_col, gap_raw = numeric_col(df, [
        "mobility_minus_subway_gap",
        "mobility_minus_subway",
        "gap",
    ])
    df["mobility_minus_subway_gap"] = gap_raw if gap_raw is not None else df["mobility_score"] - df["observed_subway_score"]
    mapping["gap_col"] = gap_col

    recon_col, recon_raw = numeric_col(df, [
        "hidden_demand_reconstruction_score_norm",
        "hidden_demand_reconstruction_score",
        "quad_reconstruction",
        "reconstruction_score",
    ])
    if recon_raw is not None:
        df["hidden_demand_reconstruction_score_norm"] = minmax(recon_raw)
    else:
        raw = (
            0.45 * df["mobility_score"].fillna(0)
            + 0.25 * df["hidden_gap_score"].fillna(0)
            + 0.20 * df["temporal_score"].fillna(0)
            + 0.10 * df["visual_morphology_score"].fillna(0)
            - df["observed_subway_score"].fillna(0)
        )
        df["hidden_demand_reconstruction_score_norm"] = minmax(raw)
    mapping["recon_col"] = recon_col

    type_col = find_col(df, ["station_type", "reconstruction_label", "quad_mobility_subway_label", "cluster"])
    df["station_type"] = df[type_col].astype(str) if type_col else "unknown"
    mapping["type_col"] = type_col

    return df, mapping

evidence_df, col_mapping = build_station_evidence(base_df)

print("\nColumn mapping:")
for k, v in col_mapping.items():
    print(k, "=>", v)

# ============================================================
# 3. IMAGE MAPPING FIXED
# ============================================================

image_exts = {".png", ".jpg", ".jpeg", ".webp"}
image_files_all = [p for p in ROOT_DIR.rglob("*") if p.suffix.lower() in image_exts]

# 논문 실험용 station tile만 우선
station_image_files = [
    p for p in image_files_all
    if "station_tiles" in str(p).lower() or "seoul_station_multimodal" in str(p).lower()
]
if len(station_image_files) == 0:
    station_image_files = image_files_all

print(f"\nFound station-related image files: {len(station_image_files)}")
for p in station_image_files[:20]:
    print("-", p)

def normalize_name(x: str) -> str:
    x = str(x).lower()
    x = re.sub(r"\s+", "", x)
    x = re.sub(r"[^0-9a-z가-힣_()-]", "", x)
    return x

def image_kind(path: Path) -> str:
    full = str(path).lower()
    name = path.name.lower()

    # 경로까지 봐야 satellite 폴더 / map 폴더 구분 가능
    if "/satellite/" in full or "\\satellite\\" in full or "satellite" in full or "위성" in full:
        return "satellite"
    if "/map/" in full or "\\map\\" in full or "map" in full or "osm" in full or "지도" in full:
        return "map"
    if "sat" in name:
        return "satellite"
    return "unknown"

def find_image_for_station(station_id: str, station_name: str, target_kind: str) -> Optional[str]:
    sid = normalize_name(station_id)
    sname = normalize_name(station_name)

    best_score = -1
    best_path = None

    for p in station_image_files:
        n = normalize_name(p.name)
        fulln = normalize_name(str(p))
        k = image_kind(p)

        score = 0

        if k == target_kind:
            score += 100
        elif k == "unknown":
            score += 5
        else:
            score -= 50

        if sid and sid in n:
            score += 100
        elif sid and sid in fulln:
            score += 60

        if sname and sname in n:
            score += 120
        elif sname and sname in fulln:
            score += 70

        if score > best_score:
            best_score = score
            best_path = p

    if best_score < 80:
        return None
    return str(best_path)

evidence_df["map_image_path"] = [
    find_image_for_station(sid, sname, "map")
    for sid, sname in zip(evidence_df["station_id"], evidence_df["station_name"])
]
evidence_df["satellite_image_path"] = [
    find_image_for_station(sid, sname, "satellite")
    for sid, sname in zip(evidence_df["station_id"], evidence_df["station_name"])
]

evidence_cols = [
    "station_id",
    "station_name",
    "station_line",
    "map_image_path",
    "satellite_image_path",
    "mobility_score",
    "observed_subway_score",
    "hidden_gap_score",
    "temporal_score",
    "visual_morphology_score",
    "mobility_minus_subway_gap",
    "hidden_demand_reconstruction_score_norm",
    "station_type",
]

station_evidence = evidence_df[evidence_cols].copy()
station_evidence.to_csv(AUTO_EVIDENCE_CSV, index=False, encoding="utf-8-sig")

print("\nImage mapping status:")
print("map found:", station_evidence["map_image_path"].notna().sum(), "/", len(station_evidence))
print("sat found:", station_evidence["satellite_image_path"].notna().sum(), "/", len(station_evidence))
display(station_evidence.head(20))
print("Saved evidence:", AUTO_EVIDENCE_CSV)

# ============================================================
# 4. PROMPT + API
# ============================================================

EVIDENCE_COLUMNS = {
    "mobility_score": "Mobility demand score",
    "observed_subway_score": "Observed subway activity score",
    "hidden_gap_score": "Hidden gap score",
    "temporal_score": "Temporal activity score",
    "visual_morphology_score": "Visual morphology score",
    "mobility_minus_subway_gap": "Mobility-minus-subway gap",
    "hidden_demand_reconstruction_score_norm": "Reference reconstruction score",
    "station_type": "Station type",
}

def image_to_data_url(path: str) -> str:
    path_obj = Path(path)
    if not path_obj.exists():
        raise FileNotFoundError(f"Image file not found: {path}")
    mime_type, _ = mimetypes.guess_type(str(path_obj))
    if mime_type is None:
        mime_type = "image/png"
    with open(path_obj, "rb") as f:
        encoded = base64.b64encode(f.read()).decode("utf-8")
    return f"data:{mime_type};base64,{encoded}"

def clean_float(x):
    try:
        if pd.isna(x):
            return None
        return round(float(x), 6)
    except Exception:
        return None

def build_evidence_text(row: pd.Series, structured: bool) -> str:
    station_name = row.get("station_name", row.get("station_id", "Unknown Station"))
    station_line = row.get("station_line", "")

    if structured:
        return f"""
Station: {station_name}
Line: {station_line}

Structured evidence:
- Mobility evidence: mobility_score = {clean_float(row.get('mobility_score'))}
- Observed transit evidence: observed_subway_score = {clean_float(row.get('observed_subway_score'))}
- Hidden-gap evidence: hidden_gap_score = {clean_float(row.get('hidden_gap_score'))}, mobility_minus_subway_gap = {clean_float(row.get('mobility_minus_subway_gap'))}
- Temporal evidence: temporal_score = {clean_float(row.get('temporal_score'))}
- Visual morphology evidence: visual_morphology_score = {clean_float(row.get('visual_morphology_score'))}
- Reference reconstruction score: hidden_demand_reconstruction_score_norm = {clean_float(row.get('hidden_demand_reconstruction_score_norm'))}
- Station type: {row.get('station_type', 'unknown')}
""".strip()

    lines = [f"Station: {station_name}", f"Line: {station_line}", "", "Station statistics:"]
    for col, desc in EVIDENCE_COLUMNS.items():
        if col in row and not pd.isna(row[col]):
            lines.append(f"- {desc}: {row[col]}")
    return "\n".join(lines)

def build_prompt(row: pd.Series, condition: str) -> str:
    station_name = row.get("station_name", row.get("station_id", "Unknown Station"))

    base = f"""
You are auditing MLLM-assisted urban mobility discovery.

Task:
Estimate whether the station area has hidden public-transit demand.
Hidden demand means that actual neighborhood mobility appears high while observed subway usage is relatively low.

Return ONLY valid JSON:
{{
  "station_name": "{station_name}",
  "hidden_demand_score": 1,
  "hidden_demand_label": "low",
  "confidence": 1,
  "primary_evidence": "visual",
  "uses_visual_reasoning": true,
  "uses_mobility_subway_reasoning": true,
  "uses_temporal_reasoning": true,
  "explanation": "one concise paragraph",
  "risk_note": "one sentence"
}}

Allowed values:
- hidden_demand_score: integer 1 to 5
- hidden_demand_label: "low", "medium", or "high"
- confidence: integer 1 to 5
- primary_evidence: "visual", "mobility_subway", "temporal", "mixed", or "insufficient"

Rules:
- Do not assume that visually dense or urban-looking areas necessarily have hidden transit demand.
- Hidden demand should be supported mainly by mismatch between mobility demand and observed subway use.
- Visual morphology can support interpretation, but it should not override contradictory mobility/subway evidence.
- If evidence is insufficient, lower the confidence.
""".strip()

    if condition == "tabular_only":
        return base + f"""

Condition: TABULAR-ONLY.
You are given only station-level numerical evidence.

{build_evidence_text(row, structured=False)}
""".strip()

    if condition == "structured_multimodal":
        return base + f"""

Condition: STRUCTURED MULTIMODAL.
You are given station-level evidence decomposed into mobility, observed subway, hidden-gap, temporal, and visual axes.
If an image is provided, use it only as auxiliary visual context.

{build_evidence_text(row, structured=True)}
""".strip()

    if condition == "image_only":
        return base + """

Condition: IMAGE-ONLY.
You are given only map/satellite images. Use visual morphology cautiously and do not invent mobility statistics.
""".strip()

    if condition == "naive_multimodal":
        return base + f"""

Condition: NAIVE MULTIMODAL.
You are given images and raw station-level statistics.

{build_evidence_text(row, structured=False)}
""".strip()

    raise ValueError(condition)

def extract_json(text: str) -> Dict[str, Any]:
    text = text.strip()
    try:
        return json.loads(text)
    except Exception:
        pass
    match = re.search(r"\{.*\}", text, flags=re.DOTALL)
    if not match:
        raise ValueError(f"No JSON found: {text[:300]}")
    return json.loads(match.group(0))

def make_content(row: pd.Series, condition: str) -> List[Dict[str, Any]]:
    prompt = build_prompt(row, condition)
    content = [{"type": "input_text", "text": prompt}]

    use_images = condition in {"image_only", "naive_multimodal", "structured_multimodal"}

    if use_images:
        paths = []

        # 속도 위해 satellite 1장 우선
        sat = row.get("satellite_image_path")
        mp = row.get("map_image_path")

        if isinstance(sat, str) and sat.strip() and Path(sat).exists():
            paths.append(sat)
        if not USE_ONLY_ONE_IMAGE:
            if isinstance(mp, str) and mp.strip() and Path(mp).exists():
                paths.append(mp)

        # satellite 없으면 map 사용
        if len(paths) == 0 and isinstance(mp, str) and mp.strip() and Path(mp).exists():
            paths.append(mp)

        for p in paths:
            content.append({
                "type": "input_image",
                "image_url": image_to_data_url(p),
                "detail": IMAGE_DETAIL,
            })

    return content

def call_mllm_single(record: Dict[str, Any], condition: str, max_retries: int = 3) -> Dict[str, Any]:
    row = pd.Series(record)

    for attempt in range(max_retries):
        try:
            content = make_content(row, condition)
            response = client.responses.create(
                model=MODEL,
                input=[{"role": "user", "content": content}],
            )
            raw_text = response.output_text
            parsed = extract_json(raw_text)

            parsed["_raw_output"] = raw_text
            parsed["_condition"] = condition
            parsed["_model"] = MODEL
            parsed["_station_id"] = row.get("station_id")
            parsed["_station_name"] = row.get("station_name")

            out = dict(record)
            out.update(parsed)
            return out

        except Exception as e:
            if attempt == max_retries - 1:
                out = dict(record)
                out.update({
                    "_condition": condition,
                    "_model": MODEL,
                    "_station_id": row.get("station_id"),
                    "_station_name": row.get("station_name"),
                    "_error": str(e),
                })
                return out
            time.sleep(1.5 * (attempt + 1))

# ============================================================
# 5. FAST FULL PARALLEL RUN
# ============================================================

run_df = station_evidence.head(LIMIT).copy() if LIMIT is not None else station_evidence.copy()

done_keys = set()
if RESUME and OUTPUT_JSONL.exists():
    with open(OUTPUT_JSONL, "r", encoding="utf-8") as f:
        for line in f:
            try:
                item = json.loads(line)
                done_keys.add((str(item.get("_station_id")), item.get("_condition")))
            except Exception:
                pass

tasks = []
for _, row in run_df.iterrows():
    record = row.to_dict()
    sid = str(record.get("station_id"))
    for condition in CONDITIONS:
        key = (sid, condition)
        if key not in done_keys:
            tasks.append((record, condition))

print("\nRUN CONFIG")
print("Stations:", len(run_df))
print("Conditions:", CONDITIONS)
print("Total expected calls:", len(run_df) * len(CONDITIONS))
print("Remaining calls:", len(tasks))
print("Workers:", MAX_WORKERS)
print("Output:", OUTPUT_JSONL)

start_time = time.time()

with ThreadPoolExecutor(max_workers=MAX_WORKERS) as executor:
    futures = [executor.submit(call_mllm_single, record, condition) for record, condition in tasks]

    with open(OUTPUT_JSONL, "a", encoding="utf-8") as out:
        for fut in tqdm(as_completed(futures), total=len(futures), desc="Parallel MLLM audit"):
            result = fut.result()
            with WRITE_LOCK:
                out.write(json.dumps(result, ensure_ascii=False) + "\n")
                out.flush()

elapsed = time.time() - start_time
print(f"\nFinished. Elapsed: {elapsed/60:.2f} min")
print("Saved:", OUTPUT_JSONL)

# ============================================================
# 6. EVALUATION
# ============================================================

def read_jsonl(path: Path) -> pd.DataFrame:
    rows = []
    with open(path, "r", encoding="utf-8") as f:
        for line in f:
            if line.strip():
                rows.append(json.loads(line))
    return pd.DataFrame(rows)

res_df = read_jsonl(OUTPUT_JSONL)
flat_csv = OUT_DIR / "mllm_audit_outputs_flat.csv"
res_df.to_csv(flat_csv, index=False, encoding="utf-8-sig")

print("Result shape:", res_df.shape)
print("Saved flat CSV:", flat_csv)

if "_error" in res_df.columns:
    err_df = res_df[res_df["_error"].notna()] if res_df["_error"].notna().any() else pd.DataFrame()
    print("Errors:", len(err_df))
    if len(err_df) > 0:
        err_path = OUT_DIR / "api_or_parse_errors.csv"
        err_df.to_csv(err_path, index=False, encoding="utf-8-sig")
        display(err_df[["_station_id", "_station_name", "_condition", "_error"]].head(20))

def safe_numeric(series: pd.Series) -> pd.Series:
    return pd.to_numeric(series, errors="coerce")

def corr_pair(data: pd.DataFrame, x_col: str, y_col: str):
    if x_col not in data.columns or y_col not in data.columns:
        return np.nan, np.nan, 0
    x = safe_numeric(data[x_col])
    y = safe_numeric(data[y_col])
    mask = x.notna() & y.notna()
    if int(mask.sum()) < 3:
        return np.nan, np.nan, int(mask.sum())
    return float(spearmanr(x[mask], y[mask]).correlation), float(pearsonr(x[mask], y[mask])[0]), int(mask.sum())

def topk_overlap(data: pd.DataFrame, score_col: str, target_col: str, k: int = 10) -> float:
    needed = [score_col, target_col, "_station_id"]
    if any(c not in data.columns for c in needed):
        return np.nan
    temp = data[needed].copy()
    temp[score_col] = safe_numeric(temp[score_col])
    temp[target_col] = safe_numeric(temp[target_col])
    temp = temp.dropna(subset=[score_col, target_col, "_station_id"])
    if len(temp) < k:
        return np.nan
    pred_top = set(temp.sort_values(score_col, ascending=False).head(k)["_station_id"].astype(str))
    true_top = set(temp.sort_values(target_col, ascending=False).head(k)["_station_id"].astype(str))
    return len(pred_top & true_top) / k

def bool_series(series: pd.Series) -> pd.Series:
    return series.astype(str).str.lower().isin(["true", "1", "yes", "y"])

def label_accuracy_top_quantile(data: pd.DataFrame, target_col: str, q: float = 0.80) -> float:
    if "hidden_demand_label" not in data.columns or target_col not in data.columns:
        return np.nan
    temp = data[["hidden_demand_label", target_col]].copy()
    temp[target_col] = safe_numeric(temp[target_col])
    temp = temp.dropna()
    if len(temp) == 0:
        return np.nan
    threshold = temp[target_col].quantile(q)
    y_true = temp[target_col] >= threshold
    y_pred = temp["hidden_demand_label"].astype(str).str.lower().eq("high")
    return float((y_true == y_pred).mean())

def unsupported_visual_reasoning_rate(data: pd.DataFrame, target_col: str, q: float = 0.50) -> float:
    if "uses_visual_reasoning" not in data.columns or target_col not in data.columns:
        return np.nan
    temp = data[["uses_visual_reasoning", target_col]].copy()
    temp[target_col] = safe_numeric(temp[target_col])
    temp = temp.dropna()
    if len(temp) == 0:
        return np.nan
    threshold = temp[target_col].quantile(q)
    visual_reason = bool_series(temp["uses_visual_reasoning"])
    unsupported = visual_reason & (temp[target_col] < threshold)
    return float(unsupported.mean())

def rate_bool(data: pd.DataFrame, col: str) -> float:
    if col not in data.columns:
        return np.nan
    return float(bool_series(data[col]).mean())

def primary_visual_rate(data: pd.DataFrame) -> float:
    if "primary_evidence" not in data.columns:
        return np.nan
    return float(data["primary_evidence"].astype(str).str.lower().eq("visual").mean())

summary_rows = []
for condition in CONDITIONS:
    sub = res_df[res_df["_condition"] == condition].copy()
    if len(sub) == 0:
        continue

    sub["hidden_demand_score"] = safe_numeric(sub.get("hidden_demand_score", pd.Series(dtype=float)))

    sp_gap, pe_gap, n_gap = corr_pair(sub, "hidden_demand_score", TARGET_COL)
    sp_ref, pe_ref, n_ref = corr_pair(sub, "hidden_demand_score", REFERENCE_COL)

    summary_rows.append({
        "condition": condition,
        "n": len(sub),
        f"spearman_with_{TARGET_COL}": sp_gap,
        f"pearson_with_{TARGET_COL}": pe_gap,
        f"spearman_with_{REFERENCE_COL}": sp_ref,
        f"pearson_with_{REFERENCE_COL}": pe_ref,
        "top10_overlap_gap": topk_overlap(sub, "hidden_demand_score", TARGET_COL, 10),
        "top10_overlap_reference": topk_overlap(sub, "hidden_demand_score", REFERENCE_COL, 10),
        "top20_overlap_gap": topk_overlap(sub, "hidden_demand_score", TARGET_COL, 20),
        "top20_overlap_reference": topk_overlap(sub, "hidden_demand_score", REFERENCE_COL, 20),
        "label_accuracy_top20pct_gap": label_accuracy_top_quantile(sub, TARGET_COL, 0.80),
        "unsupported_visual_reasoning_rate": unsupported_visual_reasoning_rate(sub, TARGET_COL, 0.50),
        "visual_primary_rate": primary_visual_rate(sub),
        "uses_visual_reasoning_rate": rate_bool(sub, "uses_visual_reasoning"),
        "uses_mobility_subway_reasoning_rate": rate_bool(sub, "uses_mobility_subway_reasoning"),
        "uses_temporal_reasoning_rate": rate_bool(sub, "uses_temporal_reasoning"),
        "valid_corr_n_gap": n_gap,
        "valid_corr_n_reference": n_ref,
    })

summary_df = pd.DataFrame(summary_rows)
summary_path = OUT_DIR / "condition_summary.csv"
summary_df.to_csv(summary_path, index=False, encoding="utf-8-sig")

print("\n=== Condition Summary ===")
display(summary_df)

# station robustness
needed = ["_station_id", "_station_name", "_condition", "hidden_demand_score"]
if all(c in res_df.columns for c in needed):
    tmp = res_df[needed].copy()
    tmp["hidden_demand_score"] = safe_numeric(tmp["hidden_demand_score"])
    tmp = tmp.dropna(subset=["_station_id", "_condition", "hidden_demand_score"])

    robust_df = tmp.pivot_table(
        index=["_station_id", "_station_name"],
        columns="_condition",
        values="hidden_demand_score",
        aggfunc="mean",
    ).reset_index()

    condition_cols = [c for c in CONDITIONS if c in robust_df.columns]
    robust_df["score_mean_across_conditions"] = robust_df[condition_cols].mean(axis=1)
    robust_df["score_std_across_conditions"] = robust_df[condition_cols].std(axis=1)
    robust_df["score_range_across_conditions"] = robust_df[condition_cols].max(axis=1) - robust_df[condition_cols].min(axis=1)
    robust_df = robust_df.sort_values("score_std_across_conditions", ascending=False)
else:
    robust_df = pd.DataFrame()

robust_path = OUT_DIR / "condition_robustness_by_station.csv"
robust_df.to_csv(robust_path, index=False, encoding="utf-8-sig")

print("\n=== Robustness by Station ===")
display(robust_df.head(20))

# error cases
if "hidden_demand_score" in res_df.columns and TARGET_COL in res_df.columns:
    temp = res_df.copy()
    temp["hidden_demand_score"] = safe_numeric(temp["hidden_demand_score"])
    temp[TARGET_COL] = safe_numeric(temp[TARGET_COL])
    temp = temp.dropna(subset=["hidden_demand_score", TARGET_COL])

    high_pred = temp["hidden_demand_score"] >= 4
    low_true = temp[TARGET_COL] < temp[TARGET_COL].quantile(0.50)

    visual = bool_series(temp["uses_visual_reasoning"]) if "uses_visual_reasoning" in temp.columns else pd.Series(False, index=temp.index)
    spurious_visual = temp[high_pred & low_true & visual].copy()
    spurious_path = OUT_DIR / "spurious_visual_cases.csv"
    spurious_visual.to_csv(spurious_path, index=False, encoding="utf-8-sig")

    low_pred = temp["hidden_demand_score"] <= 2
    high_true = temp[TARGET_COL] >= temp[TARGET_COL].quantile(0.80)
    missed_hidden = temp[low_pred & high_true].copy()
    missed_path = OUT_DIR / "missed_hidden_demand_cases.csv"
    missed_hidden.to_csv(missed_path, index=False, encoding="utf-8-sig")

    print("\nSpurious visual cases:", len(spurious_visual))
    if len(spurious_visual) > 0:
        display(spurious_visual[["_station_id", "_station_name", "_condition", "hidden_demand_score", TARGET_COL, "primary_evidence", "explanation"]].head(10))

    print("\nMissed hidden-demand cases:", len(missed_hidden))
    if len(missed_hidden) > 0:
        display(missed_hidden[["_station_id", "_station_name", "_condition", "hidden_demand_score", TARGET_COL, "primary_evidence", "explanation"]].head(10))

# LaTeX paper table
paper_cols = [
    "condition",
    "n",
    f"spearman_with_{TARGET_COL}",
    f"pearson_with_{TARGET_COL}",
    "top10_overlap_gap",
    "top20_overlap_gap",
    "label_accuracy_top20pct_gap",
    "unsupported_visual_reasoning_rate",
    "visual_primary_rate",
    "uses_mobility_subway_reasoning_rate",
    "uses_temporal_reasoning_rate",
]
existing_paper_cols = [c for c in paper_cols if c in summary_df.columns]
latex_table = summary_df[existing_paper_cols].to_latex(index=False, float_format="%.3f", escape=True)
latex_path = OUT_DIR / "condition_summary_latex_table.tex"
with open(latex_path, "w", encoding="utf-8") as f:
    f.write(latex_table)

print("\nSaved files:")
print("- Evidence CSV:", AUTO_EVIDENCE_CSV)
print("- JSONL:", OUTPUT_JSONL)
print("- Flat CSV:", flat_csv)
print("- Summary CSV:", summary_path)
print("- Robustness CSV:", robust_path)
print("- LaTeX table:", latex_path)
```
