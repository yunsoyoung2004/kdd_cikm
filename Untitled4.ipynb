```python
import os
import pandas as pd
import matplotlib.pyplot as plt

OUT_DIR = "seoul_station_multimodal/results"
os.makedirs(OUT_DIR, exist_ok=True)

placement = pd.DataFrame([
    ["Main result table",
     "quad_reconstructed_hidden_demand_result.csv",
     "Early Results section",
     "Source table for final reconstruction results and top hidden-demand ranking"],
    ["Axis decomposition table",
     "quad_latent_decomposition.csv",
     "After Method or Appendix",
     "Quad axis definitions and station-level axis scores"],
    ["Case study figure",
     "case_study_top_hidden_demand.png",
     "Middle of Results / Case Study section",
     "Visualization of top five hidden-demand cases"],
    ["Case study table",
     "case_study_top_hidden_demand_table.csv",
     "Immediately after case study figure",
     "Numerical summary of stations shown in the figure"],
    ["Baseline figure/table",
     "baseline_reconstruction_comparison.png / .csv",
     "Quantitative Results section",
     "Baseline comparison"],
    ["Weight sensitivity",
     "weight_sensitivity_results.csv / weight_sensitivity_distribution.png",
     "Appendix or later Quantitative Results",
     "Weight sensitivity analysis and distribution"],
    ["Robustness figure/table",
     "top_station_robustness.png / .csv",
     "Later Quantitative Results section",
     "Top-station robustness"],
    ["Semantic profile",
     "hidden_demand_semantic_profile.png / .csv",
     "After Case Study or Appendix",
     "Interpretation of hidden-demand morphology"],
    ["Hidden vs non-hidden",
     "hidden_vs_nonhidden_semantic_difference.png / .csv",
     "Appendix",
     "Limitations of semantic difference analysis"],
], columns=["Manuscript Item", "File Path", "Placement", "Recommended Caption / Purpose"])

csv_path = f"{OUT_DIR}/table5_manuscript_figure_table_placement_en.csv"
placement.to_csv(csv_path, index=False)

# Figure-style table image
fig, ax = plt.subplots(figsize=(18, 8))
ax.axis("off")

table = ax.table(
    cellText=placement.values,
    colLabels=placement.columns,
    loc="center",
    cellLoc="left",
    colLoc="center",
)

table.auto_set_font_size(False)
table.set_fontsize(9)
table.scale(1, 2.2)

for (row, col), cell in table.get_celld().items():
    cell.set_edgecolor("black")
    if row == 0:
        cell.set_text_props(weight="bold")
        cell.set_height(0.08)
    else:
        cell.set_height(0.095)

# Column widths
widths = {
    0: 0.18,
    1: 0.30,
    2: 0.22,
    3: 0.30,
}
for col, width in widths.items():
    for row in range(len(placement) + 1):
        table[(row, col)].set_width(width)

plt.title(
    "Table 5. Recommended Placement of Existing Result Files in the Manuscript",
    fontsize=15,
    weight="bold",
    pad=18,
)

png_path = f"{OUT_DIR}/table5_manuscript_figure_table_placement_en.png"
pdf_path = f"{OUT_DIR}/table5_manuscript_figure_table_placement_en.pdf"

plt.savefig(png_path, dpi=300, bbox_inches="tight")
plt.savefig(pdf_path, bbox_inches="tight")
plt.show()

print("Saved CSV:", csv_path)
print("Saved PNG:", png_path)
print("Saved PDF:", pdf_path)
```

```python
# ============================================================
# CASE STUDY PLATE FIGURE
# 6 stations x map/satellite/compact score strip
# English-only, paper-ready
# ============================================================

import os
import glob
import pandas as pd
import matplotlib.pyplot as plt
from PIL import Image

RESULT_DIR = "seoul_station_multimodal/results"
FIG_DIR = f"{RESULT_DIR}/paper_figures_en"
os.makedirs(FIG_DIR, exist_ok=True)

RECON_PATH = f"{RESULT_DIR}/quad_reconstructed_hidden_demand_result.csv"

IMAGE_ROOT = "/workspace/yunsoyoung_cikm_kdd../images/station_tiles"
MAP_DIR = f"{IMAGE_ROOT}/map"
SAT_DIR = f"{IMAGE_ROOT}/satellite"

plt.rcParams["font.family"] = "DejaVu Sans"
plt.rcParams["axes.unicode_minus"] = False

station_name_map = {
    "마천": "Macheon",
    "단대오거리": "Dandaeogeori",
    "신정네거리": "Sinjeongnegeori",
    "까치산": "Kkachisan",
    "부천시청": "Bucheon City Hall",
    "남한산성입구(성남법원.검찰청)": "Namhansanseong",
    "거여": "Geoyeo",
    "북한산보국문": "Bukhansan Bogungmun",
    "가양": "Gayang",
    "양천구청": "Yangcheon-gu Office",
    "홍제": "Hongje",
    "광명사거리": "Gwangmyeongsageori",
}

def station_en(x):
    x = str(x)
    return station_name_map.get(x, x.encode("ascii", "ignore").decode() or "Station")

def find_station_image(img_dir, station_id):
    patterns = [
        os.path.join(img_dir, f"{station_id}_*.png"),
        os.path.join(img_dir, f"{station_id}_*.jpg"),
        os.path.join(img_dir, f"{station_id}_*.jpeg"),
    ]
    matches = []
    for p in patterns:
        matches.extend(glob.glob(p))
    return matches[0] if matches else None

def short_label(row):
    if row["axis_observed_subway_score"] < 0.15 and row["axis_hidden_gap_score"] > 0.6:
        return "High mobility, weak subway capture"
    if row["axis_visual_morphology_score"] > 0.6:
        return "Morphology-supported hidden demand"
    if row["axis_temporal_dynamics_score"] > 0.6:
        return "Dynamic activity rhythm"
    return "Reconstructed hidden demand"

# ============================================================
# LOAD AND SELECT TOP 6
# ============================================================

df = pd.read_csv(RECON_PATH)
name_col = "nearest_station" if "nearest_station" in df.columns else "station_name"
df["station_name_en"] = df[name_col].apply(station_en)

top = (
    df.sort_values("hidden_demand_reconstruction_score_norm", ascending=False)
    .head(6)
    .copy()
)

# ============================================================
# FIGURE LAYOUT
# 2 rows x 3 stations
# Each station card: map, satellite, compact score strip
# ============================================================

fig = plt.figure(figsize=(18, 11))
outer = fig.add_gridspec(2, 3, wspace=0.18, hspace=0.28)

score_cols = [
    "axis_mobility_demand_score",
    "axis_hidden_gap_score",
    "axis_temporal_dynamics_score",
    "axis_visual_morphology_score",
    "axis_observed_subway_score",
]

score_labels = ["M", "H", "T", "V", "S"]

for idx, (_, row) in enumerate(top.iterrows()):
    r = idx // 3
    c = idx % 3

    inner = outer[r, c].subgridspec(
        3, 2,
        height_ratios=[0.18, 1.0, 0.34],
        width_ratios=[1, 1],
        wspace=0.05,
        hspace=0.08,
    )

    # title
    ax_title = fig.add_subplot(inner[0, :])
    ax_title.axis("off")
    title = (
        f"#{int(row['priority_rank'])} {row['station_name_en']} "
        f"({row.get('nearest_line', '')})"
    )
    subtitle = (
        f"Recon={row['hidden_demand_reconstruction_score_norm']:.3f} | "
        f"{short_label(row)}"
    )
    ax_title.text(
        0.5, 0.75, title,
        ha="center", va="center",
        fontsize=12, fontweight="bold"
    )
    ax_title.text(
        0.5, 0.20, subtitle,
        ha="center", va="center",
        fontsize=9
    )

    # map
    ax_map = fig.add_subplot(inner[1, 0])
    map_path = find_station_image(MAP_DIR, row["station_id"])
    if map_path and os.path.exists(map_path):
        ax_map.imshow(Image.open(map_path).convert("RGB"))
    ax_map.set_title("Map", fontsize=9)
    ax_map.axis("off")

    # satellite
    ax_sat = fig.add_subplot(inner[1, 1])
    sat_path = find_station_image(SAT_DIR, row["station_id"])
    if sat_path and os.path.exists(sat_path):
        ax_sat.imshow(Image.open(sat_path).convert("RGB"))
    ax_sat.set_title("Satellite", fontsize=9)
    ax_sat.axis("off")

    # compact score strip
    ax_score = fig.add_subplot(inner[2, :])
    vals = [float(row[col]) for col in score_cols]
    ax_score.bar(score_labels, vals)
    ax_score.set_ylim(0, 1)
    ax_score.set_ylabel("Score", fontsize=8)
    ax_score.tick_params(axis="both", labelsize=8)
    ax_score.grid(axis="y", alpha=0.25)

    # small legend text inside
    ax_score.text(
        0.02, 0.92,
        "M=Mobility, H=Hidden Gap, T=Temporal, V=Visual, S=Observed Subway",
        transform=ax_score.transAxes,
        fontsize=7,
        va="top"
    )

fig.suptitle(
    "Representative Case Studies of Reconstructed Hidden Transit Demand",
    fontsize=16,
    fontweight="bold",
    y=0.985
)

out_path = f"{FIG_DIR}/fig_case_study_plate_top6.png"
plt.savefig(out_path, dpi=300, bbox_inches="tight")
plt.show()

print("Saved:", out_path)

# save selected station table
table_path = f"{FIG_DIR}/fig_case_study_plate_top6_table.csv"
top[
    [
        "priority_rank",
        "station_id",
        "station_name_en",
        "nearest_line",
        "reconstruction_label",
        "hidden_demand_reconstruction_score_norm",
        "axis_mobility_demand_score",
        "axis_hidden_gap_score",
        "axis_temporal_dynamics_score",
        "axis_visual_morphology_score",
        "axis_observed_subway_score",
    ]
].to_csv(table_path, index=False)

print("Saved:", table_path)
```

```python
# ============================================================
# CLEAN PAPER-READY CASE STUDY FIGURE
# - English only
# - No Korean glyph warning
# - Fixed overlap/layout
# - 6 representative stations
# ============================================================

import os
import glob
import textwrap
import warnings
import pandas as pd
import matplotlib.pyplot as plt

from PIL import Image

# ============================================================
# CONFIG
# ============================================================

RESULT_DIR = "seoul_station_multimodal/results"
FIG_DIR = f"{RESULT_DIR}/paper_figures_en"

os.makedirs(FIG_DIR, exist_ok=True)

RECON_PATH = f"{RESULT_DIR}/quad_reconstructed_hidden_demand_result.csv"

IMAGE_ROOT = "/workspace/yunsoyoung_cikm_kdd../images/station_tiles"

MAP_DIR = f"{IMAGE_ROOT}/map"
SAT_DIR = f"{IMAGE_ROOT}/satellite"

plt.rcParams["font.family"] = "DejaVu Sans"
plt.rcParams["axes.unicode_minus"] = False

warnings.filterwarnings("ignore", category=UserWarning)

# ============================================================
# ENGLISH MAPPING
# ============================================================

station_name_map = {
    "마천": "Macheon",
    "단대오거리": "Dandaeogeori",
    "신정네거리": "Sinjeongnegeori",
    "까치산": "Kkachisan",
    "부천시청": "Bucheon City Hall",
    "남한산성입구(성남법원.검찰청)": "Namhansanseong",
    "거여": "Geoyeo",
    "북한산보국문": "Bukhansan Bogungmun",
    "가양": "Gayang",
    "양천구청": "Yangcheon-gu Office",
    "홍제": "Hongje",
    "광명사거리": "Gwangmyeongsageori",
    "중앙보훈병원": "VHS Medical Center",
    "신중동": "Sinjung-dong",
    "개화": "Gaehwa",
    "신목동": "Sinmokdong",
    "장지": "Jangji",
}

line_map = {
    "1호선": "Line 1",
    "2호선": "Line 2",
    "3호선": "Line 3",
    "4호선": "Line 4",
    "5호선": "Line 5",
    "6호선": "Line 6",
    "7호선": "Line 7",
    "8호선": "Line 8",
    "9호선": "Line 9",
    "9호선(연장)": "Line 9 Ext.",
    "우이신설선": "Ui-Sinseol",
}

def station_en(x):
    x = str(x)
    return station_name_map.get(
        x,
        x.encode("ascii", "ignore").decode() or "Station"
    )

def line_en(x):
    x = str(x)
    return line_map.get(x, x)

# ============================================================
# INTERPRETATION
# ============================================================

def short_label(row):

    if (
        row["axis_observed_subway_score"] < 0.15
        and row["axis_hidden_gap_score"] > 0.6
    ):
        return "High mobility with weak subway capture"

    if row["axis_visual_morphology_score"] > 0.65:
        return "Morphology-supported latent demand"

    if row["axis_temporal_dynamics_score"] > 0.65:
        return "Dynamic temporal activity pattern"

    return "Reconstructed hidden transit demand"

# ============================================================
# IMAGE FINDER
# ============================================================

def find_station_image(img_dir, station_id):

    patterns = [
        os.path.join(img_dir, f"{station_id}_*.png"),
        os.path.join(img_dir, f"{station_id}_*.jpg"),
        os.path.join(img_dir, f"{station_id}_*.jpeg"),
    ]

    matches = []

    for p in patterns:
        matches.extend(glob.glob(p))

    return matches[0] if matches else None

# ============================================================
# LOAD DATA
# ============================================================

df = pd.read_csv(RECON_PATH)

name_col = (
    "nearest_station"
    if "nearest_station" in df.columns
    else "station_name"
)

df["station_name_en"] = df[name_col].apply(station_en)

df["nearest_line_en"] = (
    df["nearest_line"]
    .astype(str)
    .apply(line_en)
)

# ============================================================
# SELECT TOP 6
# ============================================================

top = (
    df.sort_values(
        "hidden_demand_reconstruction_score_norm",
        ascending=False
    )
    .head(6)
    .copy()
)

# ============================================================
# FIGURE LAYOUT
# ============================================================

fig = plt.figure(figsize=(20, 13))

outer = fig.add_gridspec(
    2,
    3,
    wspace=0.18,
    hspace=0.30,
)

score_cols = [
    "axis_mobility_demand_score",
    "axis_hidden_gap_score",
    "axis_temporal_dynamics_score",
    "axis_visual_morphology_score",
    "axis_observed_subway_score",
]

score_labels = [
    "Mobility",
    "Hidden",
    "Temporal",
    "Visual",
    "Subway",
]

# ============================================================
# BUILD PANELS
# ============================================================

for idx, (_, row) in enumerate(top.iterrows()):

    r = idx // 3
    c = idx % 3

    inner = outer[r, c].subgridspec(
        3,
        2,
        height_ratios=[0.18, 1.0, 0.38],
        width_ratios=[1, 1],
        wspace=0.04,
        hspace=0.10,
    )

    # ========================================================
    # TITLE
    # ========================================================

    ax_title = fig.add_subplot(inner[0, :])
    ax_title.axis("off")

    title = (
        f"#{int(row['priority_rank'])} "
        f"{row['station_name_en']} "
        f"({row['nearest_line_en']})"
    )

    subtitle = textwrap.fill(
        (
            f"Recon={row['hidden_demand_reconstruction_score_norm']:.3f} | "
            f"{short_label(row)}"
        ),
        width=42
    )

    ax_title.text(
        0.5,
        0.72,
        title,
        ha="center",
        va="center",
        fontsize=10,
        fontweight="bold",
    )

    ax_title.text(
        0.5,
        0.18,
        subtitle,
        ha="center",
        va="center",
        fontsize=8,
    )

    # ========================================================
    # MAP IMAGE
    # ========================================================

    ax_map = fig.add_subplot(inner[1, 0])

    map_path = find_station_image(
        MAP_DIR,
        row["station_id"]
    )

    if map_path and os.path.exists(map_path):
        img = Image.open(map_path).convert("RGB")
        ax_map.imshow(img)

    ax_map.set_title(
        "Map View",
        fontsize=8
    )

    ax_map.axis("off")

    # ========================================================
    # SATELLITE IMAGE
    # ========================================================

    ax_sat = fig.add_subplot(inner[1, 1])

    sat_path = find_station_image(
        SAT_DIR,
        row["station_id"]
    )

    if sat_path and os.path.exists(sat_path):
        img = Image.open(sat_path).convert("RGB")
        ax_sat.imshow(img)

    ax_sat.set_title(
        "Satellite View",
        fontsize=8
    )

    ax_sat.axis("off")

    # ========================================================
    # SCORE STRIP
    # ========================================================

    ax_score = fig.add_subplot(inner[2, :])

    vals = [
        float(row[col])
        for col in score_cols
    ]

    bars = ax_score.bar(
        score_labels,
        vals,
    )

    ax_score.set_ylim(0, 1)

    ax_score.tick_params(
        axis="x",
        labelsize=7,
        rotation=0,
    )

    ax_score.tick_params(
        axis="y",
        labelsize=7,
    )

    ax_score.grid(
        axis="y",
        alpha=0.25,
    )

    # annotate values

    for b, v in zip(bars, vals):

        ax_score.text(
            b.get_x() + b.get_width() / 2,
            v + 0.02,
            f"{v:.2f}",
            ha="center",
            fontsize=6,
        )

# ============================================================
# GLOBAL TITLE
# ============================================================

fig.suptitle(
    "Representative Case Studies of Reconstructed Hidden Transit Demand",
    fontsize=16,
    fontweight="bold",
    y=0.985,
)

# ============================================================
# SAVE
# ============================================================

plt.tight_layout(
    rect=[0, 0, 1, 0.97]
)

out_path = (
    f"{FIG_DIR}/"
    f"fig_case_study_plate_top6_clean.png"
)

plt.savefig(
    out_path,
    dpi=300,
    bbox_inches="tight",
)

plt.close()

print("=" * 60)
print("DONE.")
print("Saved:")
print(out_path)

# ============================================================
# SAVE TABLE
# ============================================================

table_path = (
    f"{FIG_DIR}/"
    f"case_study_top6_clean_table.csv"
)

top[
    [
        "priority_rank",
        "station_id",
        "station_name_en",
        "nearest_line_en",
        "reconstruction_label",
        "hidden_demand_reconstruction_score_norm",
        "axis_mobility_demand_score",
        "axis_hidden_gap_score",
        "axis_temporal_dynamics_score",
        "axis_visual_morphology_score",
        "axis_observed_subway_score",
    ]
].to_csv(
    table_path,
    index=False,
)

print(table_path)
```
