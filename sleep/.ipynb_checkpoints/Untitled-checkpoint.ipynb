```python
# ============================================================
# improved_sleep_policy_full_jupyter_safe.py
# ============================================================

from __future__ import annotations

import argparse
import random
import numpy as np
import torch
import torch.nn as nn
import torch.nn.functional as F

# ============================================================
# JUPYTER SAFE ARGPARSE
# ============================================================

def parse_args():
    parser = argparse.ArgumentParser()

    parser.add_argument("--epochs", type=int, default=30)
    parser.add_argument("--batch_size", type=int, default=64)
    parser.add_argument("--lr", type=float, default=2e-4)
    parser.add_argument("--device", type=str, default="cuda")
    parser.add_argument("--seed", type=int, default=42)
    parser.add_argument("--save_path", type=str, default="./sleep_model.pt")

    # NOTE: Jupyter kernel argument safe
    args, _ = parser.parse_known_args()

    return args


# ============================================================
# SEED
# ============================================================

def set_seed(seed):

    random.seed(seed)
    np.random.seed(seed)

    torch.manual_seed(seed)

    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)


# ============================================================
# STATE / ACTION
# ============================================================

STATE_NAMES = [
    "rested_focused",
    "sleep_deprived_slow",
    "high_noise_overloaded",
    "emotionally_reactive",
]

ACTION_NAMES = [
    "normal_guidance",
    "low_complexity_checkin",
    "step_by_step_support",
    "emotional_validation",
    "noise_reduction_prompt",
]


# ============================================================
# SYNTHETIC DATA
# ============================================================

def generate_sample():

    state = np.random.randint(0, 4)

    obs = np.random.rand(8)

    action = np.random.randint(0, len(ACTION_NAMES))

    reward = np.random.uniform(0.4, 1.0)

    return obs, action, reward, state


# ============================================================
# DATASET
# ============================================================

class SleepDataset(torch.utils.data.Dataset):

    def __init__(self, size):

        self.data = [generate_sample() for _ in range(size)]

    def __len__(self):

        return len(self.data)

    def __getitem__(self, idx):

        obs, action, reward, state = self.data[idx]

        return (
            torch.tensor(obs).float(),
            torch.tensor(action).long(),
            torch.tensor(reward).float(),
            torch.tensor(state).long(),
        )


# ============================================================
# MODEL
# ============================================================

class SleepModel(nn.Module):

    def __init__(self):

        super().__init__()

        self.encoder = nn.Sequential(
            nn.Linear(8,128),
            nn.LayerNorm(128),
            nn.GELU(),
            nn.Linear(128,128)
        )

        self.state_head = nn.Linear(128,4)

        self.action_head = nn.Linear(128,5)

        self.reward_head = nn.Linear(128,1)

    def forward(self,x):

        h = self.encoder(x)

        state_logits = self.state_head(h)

        action_logits = self.action_head(h)

        reward = self.reward_head(h)

        return state_logits, action_logits, reward


# ============================================================
# TRAIN
# ============================================================

def train_epoch(model,loader,optim,device):

    model.train()

    total_loss = 0

    for obs,action,reward,state in loader:

        obs = obs.to(device)
        action = action.to(device)
        reward = reward.to(device)
        state = state.to(device)

        state_logits, action_logits, reward_pred = model(obs)

        state_loss = F.cross_entropy(state_logits,state)

        action_loss = F.cross_entropy(action_logits,action)

        reward_loss = F.mse_loss(reward_pred.squeeze(),reward)

        loss = state_loss + 0.5*action_loss + 0.3*reward_loss

        optim.zero_grad()

        loss.backward()

        optim.step()

        total_loss += loss.item()

    return total_loss/len(loader)


# ============================================================
# EVAL
# ============================================================

def evaluate(model,loader,device):

    model.eval()

    correct = 0
    total = 0

    with torch.no_grad():

        for obs,action,reward,state in loader:

            obs = obs.to(device)

            state_logits,_,_ = model(obs)

            pred = state_logits.argmax(dim=1)

            correct += (pred.cpu()==state).sum().item()

            total += len(state)

    return correct/total


# ============================================================
# MAIN
# ============================================================

def main():

    args = parse_args()

    device = torch.device(args.device if torch.cuda.is_available() else "cpu")

    print("device =",device)

    set_seed(args.seed)

    train_ds = SleepDataset(10000)

    val_ds = SleepDataset(2000)

    train_loader = torch.utils.data.DataLoader(
        train_ds,
        batch_size=args.batch_size,
        shuffle=True
    )

    val_loader = torch.utils.data.DataLoader(
        val_ds,
        batch_size=args.batch_size
    )

    model = SleepModel().to(device)

    optim = torch.optim.Adam(
        model.parameters(),
        lr=args.lr
    )

    best = 0

    for epoch in range(args.epochs):

        loss = train_epoch(
            model,
            train_loader,
            optim,
            device
        )

        acc = evaluate(
            model,
            val_loader,
            device
        )

        print(
            f"epoch {epoch+1} | loss={loss:.4f} | acc={acc:.4f}"
        )

        if acc > best:

            best = acc

            torch.save(
                model.state_dict(),
                args.save_path
            )

    print("best acc =",best)


# ============================================================
# ENTRY
# ============================================================

if __name__ == "__main__":

    main()
```

```python
# ============================================================
# improved_sleep_research_jupyter_safe_full.py
# ============================================================

from __future__ import annotations

import argparse
import random
from dataclasses import dataclass

import numpy as np
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader, WeightedRandomSampler


# ============================================================
# Jupyter-safe argparse
# ============================================================

def parse_args():
    parser = argparse.ArgumentParser()
    parser.add_argument("--epochs", type=int, default=25)
    parser.add_argument("--batch_size", type=int, default=128)
    parser.add_argument("--lr", type=float, default=2e-4)
    parser.add_argument("--device", type=str, default="cuda")
    parser.add_argument("--seed", type=int, default=42)
    parser.add_argument("--save_path", type=str, default="./best_sleep_research_model.pt")
    args, _ = parser.parse_known_args()
    return args


# ============================================================
# Reproducibility
# ============================================================

def set_seed(seed: int = 42):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)


# ============================================================
# Definitions
# ============================================================

STATE_NAMES = [
    "rested_focused",
    "sleep_deprived_slow",
    "high_noise_overloaded",
    "emotionally_reactive",
]
STATE2ID = {s: i for i, s in enumerate(STATE_NAMES)}

ACTION_NAMES = [
    "normal_guidance",
    "low_complexity_checkin",
    "step_by_step_support",
    "emotional_validation",
    "noise_reduction_prompt",
    "sleep_hygiene_prompt",
]
ACTION2ID = {a: i for i, a in enumerate(ACTION_NAMES)}


@dataclass
class Config:
    seq_len: int = 12
    obs_dim: int = 8
    hidden_dim: int = 128
    num_layers: int = 1
    dropout: float = 0.10

    train_samples: int = 12000
    val_samples: int = 2000
    test_samples: int = 2000

    epochs: int = 25
    batch_size: int = 128
    lr: float = 2e-4
    weight_decay: float = 1e-4

    action_coef: float = 0.4
    reward_coef: float = 0.25
    transition_coef: float = 0.35
    entropy_coef: float = 0.003

    device: str = "cuda"
    seed: int = 42


# ============================================================
# Synthetic environment with learnable structure
# ============================================================

def clamp01(x: float) -> float:
    return max(0.0, min(1.0, x))


def state_proto(state_id: int):
    if state_id == STATE2ID["rested_focused"]:
        return {
            "response_time": 0.25,
            "typing_var": 0.28,
            "sleep_score": 0.86,
            "noise_load": 0.22,
            "affect_intensity": 0.18,
            "engagement": 0.86,
            "adherence": 0.88,
        }
    if state_id == STATE2ID["sleep_deprived_slow"]:
        return {
            "response_time": 0.82,
            "typing_var": 0.40,
            "sleep_score": 0.18,
            "noise_load": 0.28,
            "affect_intensity": 0.20,
            "engagement": 0.72,
            "adherence": 0.76,
        }
    if state_id == STATE2ID["high_noise_overloaded"]:
        return {
            "response_time": 0.60,
            "typing_var": 0.62,
            "sleep_score": 0.48,
            "noise_load": 0.88,
            "affect_intensity": 0.30,
            "engagement": 0.66,
            "adherence": 0.64,
        }
    if state_id == STATE2ID["emotionally_reactive"]:
        return {
            "response_time": 0.46,
            "typing_var": 0.68,
            "sleep_score": 0.50,
            "noise_load": 0.30,
            "affect_intensity": 0.90,
            "engagement": 0.60,
            "adherence": 0.58,
        }
    raise ValueError("invalid state")


def best_action_for_state(state_id: int) -> int:
    if state_id == STATE2ID["rested_focused"]:
        return ACTION2ID["normal_guidance"]
    if state_id == STATE2ID["sleep_deprived_slow"]:
        return ACTION2ID["low_complexity_checkin"]
    if state_id == STATE2ID["high_noise_overloaded"]:
        return ACTION2ID["noise_reduction_prompt"]
    if state_id == STATE2ID["emotionally_reactive"]:
        return ACTION2ID["emotional_validation"]
    raise ValueError("invalid state")


def reward_from_state_action(state_id: int, action_id: int) -> float:
    best = best_action_for_state(state_id)
    if action_id == best:
        return np.random.uniform(0.90, 0.98)

    acceptable = {
        STATE2ID["rested_focused"]: {ACTION2ID["step_by_step_support"]},
        STATE2ID["sleep_deprived_slow"]: {
            ACTION2ID["step_by_step_support"],
            ACTION2ID["sleep_hygiene_prompt"],
        },
        STATE2ID["high_noise_overloaded"]: {
            ACTION2ID["low_complexity_checkin"],
            ACTION2ID["step_by_step_support"],
        },
        STATE2ID["emotionally_reactive"]: {
            ACTION2ID["low_complexity_checkin"],
            ACTION2ID["step_by_step_support"],
        },
    }

    if action_id in acceptable[state_id]:
        return np.random.uniform(0.72, 0.86)

    return np.random.uniform(0.35, 0.65)


def transition_next_state(curr_state: int, action_id: int) -> int:
    p = np.random.rand()
    best = best_action_for_state(curr_state)

    if action_id == best:
        if curr_state == STATE2ID["sleep_deprived_slow"] and p < 0.18:
            return STATE2ID["rested_focused"]
        if curr_state == STATE2ID["high_noise_overloaded"] and p < 0.20:
            return STATE2ID["rested_focused"]
        if curr_state == STATE2ID["emotionally_reactive"] and p < 0.22:
            return STATE2ID["rested_focused"]
        return curr_state

    if curr_state == STATE2ID["sleep_deprived_slow"] and action_id == ACTION2ID["normal_guidance"] and p < 0.30:
        return STATE2ID["high_noise_overloaded"]

    if curr_state == STATE2ID["high_noise_overloaded"] and action_id == ACTION2ID["normal_guidance"] and p < 0.30:
        return STATE2ID["emotionally_reactive"]

    if curr_state == STATE2ID["emotionally_reactive"] and action_id == ACTION2ID["normal_guidance"] and p < 0.35:
        return STATE2ID["high_noise_overloaded"]

    return curr_state


def generate_observation(state_id: int, hour: int) -> np.ndarray:
    proto = state_proto(state_id)

    hour_norm = hour / 23.0

    circadian = 0.0
    if 0 <= hour <= 5:
        circadian = 0.10
    elif 6 <= hour <= 9:
        circadian = -0.08
    elif 22 <= hour <= 23:
        circadian = 0.06

    response_time = clamp01(np.random.normal(proto["response_time"], 0.05))
    typing_var = clamp01(np.random.normal(proto["typing_var"], 0.06))
    sleep_score = clamp01(np.random.normal(proto["sleep_score"] - circadian, 0.08))
    noise_load = clamp01(np.random.normal(proto["noise_load"], 0.07))
    affect_intensity = clamp01(np.random.normal(proto["affect_intensity"], 0.07))
    engagement = clamp01(np.random.normal(proto["engagement"], 0.06))
    adherence = clamp01(np.random.normal(proto["adherence"], 0.06))

    # state separation 강화
    if state_id == STATE2ID["sleep_deprived_slow"]:
        response_time = clamp01(response_time + 0.05)
        sleep_score = clamp01(sleep_score - 0.05)

    elif state_id == STATE2ID["high_noise_overloaded"]:
        noise_load = clamp01(noise_load + 0.05)
        typing_var = clamp01(typing_var + 0.04)

    elif state_id == STATE2ID["emotionally_reactive"]:
        affect_intensity = clamp01(affect_intensity + 0.06)
        typing_var = clamp01(typing_var + 0.03)

    obs = np.array(
        [
            hour_norm,
            response_time,
            typing_var,
            sleep_score,
            noise_load,
            affect_intensity,
            engagement,
            adherence,
        ],
        dtype=np.float32,
    )
    return obs


def generate_sequence(seq_len: int):
    state = np.random.choice(4, p=[0.25, 0.30, 0.22, 0.23])
    start_hour = np.random.randint(0, 24)

    obs_seq = []
    action_seq = []
    reward_seq = []
    state_seq = []
    next_state_seq = []

    for t in range(seq_len):
        hour = (start_hour + t * np.random.randint(1, 4)) % 24
        obs = generate_observation(state, hour)

        if np.random.rand() < 0.72:
            action = best_action_for_state(state)
        else:
            action = np.random.randint(0, len(ACTION_NAMES))

        reward = reward_from_state_action(state, action)
        next_state = transition_next_state(state, action)

        obs_seq.append(obs)
        action_seq.append(action)
        reward_seq.append(reward)
        state_seq.append(state)
        next_state_seq.append(next_state)

        state = next_state

    return {
        "obs": np.stack(obs_seq),
        "actions": np.array(action_seq, dtype=np.int64),
        "rewards": np.array(reward_seq, dtype=np.float32),
        "states": np.array(state_seq, dtype=np.int64),
        "next_states": np.array(next_state_seq, dtype=np.int64),
    }


# ============================================================
# Dataset
# ============================================================

class SleepSequenceDataset(Dataset):
    def __init__(self, n_samples: int, seq_len: int):
        self.samples = [generate_sequence(seq_len) for _ in range(n_samples)]
        self.labels = [int(x["states"][-1]) for x in self.samples]

        counts = np.bincount(self.labels, minlength=len(STATE_NAMES))
        inv = 1.0 / np.maximum(counts, 1)
        self.sample_weights = [inv[y] for y in self.labels]

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):
        x = self.samples[idx]
        return {
            "obs": torch.tensor(x["obs"], dtype=torch.float32),
            "actions": torch.tensor(x["actions"], dtype=torch.long),
            "rewards": torch.tensor(x["rewards"], dtype=torch.float32),
            "states": torch.tensor(x["states"], dtype=torch.long),
            "next_states": torch.tensor(x["next_states"], dtype=torch.long),
        }


def build_loader(dataset: SleepSequenceDataset, batch_size: int, balanced: bool):
    if balanced:
        sampler = WeightedRandomSampler(
            weights=dataset.sample_weights,
            num_samples=len(dataset),
            replacement=True,
        )
        return DataLoader(dataset, batch_size=batch_size, sampler=sampler)
    return DataLoader(dataset, batch_size=batch_size, shuffle=False)


# ============================================================
# Model
# ============================================================

class SleepBeliefPolicyModel(nn.Module):
    def __init__(self, cfg: Config):
        super().__init__()
        self.obs_encoder = nn.Sequential(
            nn.Linear(cfg.obs_dim, cfg.hidden_dim),
            nn.LayerNorm(cfg.hidden_dim),
            nn.GELU(),
            nn.Dropout(cfg.dropout),
            nn.Linear(cfg.hidden_dim, cfg.hidden_dim),
            nn.LayerNorm(cfg.hidden_dim),
            nn.GELU(),
        )

        self.action_emb = nn.Embedding(len(ACTION_NAMES), cfg.hidden_dim // 4)
        self.reward_proj = nn.Linear(1, cfg.hidden_dim // 8)

        rnn_in_dim = cfg.hidden_dim + cfg.hidden_dim // 4 + cfg.hidden_dim // 8

        self.gru = nn.GRU(
            input_size=rnn_in_dim,
            hidden_size=cfg.hidden_dim,
            num_layers=cfg.num_layers,
            batch_first=True,
            dropout=cfg.dropout if cfg.num_layers > 1 else 0.0,
        )

        self.shared = nn.Sequential(
            nn.Linear(cfg.hidden_dim, cfg.hidden_dim),
            nn.LayerNorm(cfg.hidden_dim),
            nn.GELU(),
            nn.Dropout(cfg.dropout),
        )

        self.state_head = nn.Linear(cfg.hidden_dim, len(STATE_NAMES))
        self.action_head = nn.Linear(cfg.hidden_dim, len(ACTION_NAMES))
        self.reward_head = nn.Linear(cfg.hidden_dim, 1)
        self.next_state_head = nn.Linear(cfg.hidden_dim, len(STATE_NAMES))

    def forward(self, obs, prev_actions, prev_rewards):
        obs_feat = self.obs_encoder(obs)
        act_feat = self.action_emb(prev_actions)
        rew_feat = self.reward_proj(prev_rewards.unsqueeze(-1))

        x = torch.cat([obs_feat, act_feat, rew_feat], dim=-1)
        h, _ = self.gru(x)
        z = self.shared(h)

        return {
            "state_logits": self.state_head(z),
            "action_logits": self.action_head(z),
            "reward_pred": self.reward_head(z).squeeze(-1),
            "next_state_logits": self.next_state_head(z),
        }


# ============================================================
# Metrics / utils
# ============================================================

def entropy_from_logits(logits):
    p = F.softmax(logits, dim=-1)
    logp = F.log_softmax(logits, dim=-1)
    return -(p * logp).sum(dim=-1).mean()


def compute_class_weights(dataset, device):
    counts = np.bincount(dataset.labels, minlength=len(STATE_NAMES)).astype(np.float32)
    w = counts.sum() / np.maximum(counts, 1.0)
    w = w / w.mean()
    return torch.tensor(w, dtype=torch.float32, device=device)


def confusion_matrix(y_true, y_pred, n_cls):
    mat = np.zeros((n_cls, n_cls), dtype=np.int64)
    for t, p in zip(y_true, y_pred):
        mat[t, p] += 1
    return mat


def print_confusion_matrix(cm):
    print("\n[Confusion Matrix] rows=true cols=pred")
    print(" " * 24 + " ".join([f"{s[:10]:>12}" for s in STATE_NAMES]))
    for i, row in enumerate(cm):
        print(f"{STATE_NAMES[i]:>24} " + " ".join([f"{int(v):12d}" for v in row]))


# ============================================================
# Train / Eval
# ============================================================

def train_one_epoch(model, loader, optimizer, class_weights, device, cfg):
    model.train()
    total_loss = 0.0

    for batch in loader:
        obs = batch["obs"].to(device)
        actions = batch["actions"].to(device)
        rewards = batch["rewards"].to(device)
        states = batch["states"].to(device)
        next_states = batch["next_states"].to(device)

        prev_actions = torch.roll(actions, shifts=1, dims=1)
        prev_rewards = torch.roll(rewards, shifts=1, dims=1)
        prev_actions[:, 0] = 0
        prev_rewards[:, 0] = 0.0

        out = model(obs, prev_actions, prev_rewards)

        B, T, S = out["state_logits"].shape
        A = out["action_logits"].shape[-1]

        state_logits = out["state_logits"].reshape(B * T, S)
        action_logits = out["action_logits"].reshape(B * T, A)
        next_state_logits = out["next_state_logits"].reshape(B * T, S)
        reward_pred = out["reward_pred"].reshape(B * T)

        state_targets = states.reshape(B * T)
        action_targets = actions.reshape(B * T)
        next_state_targets = next_states.reshape(B * T)
        reward_targets = rewards.reshape(B * T)

        state_loss = F.cross_entropy(state_logits, state_targets, weight=class_weights)
        action_loss = F.cross_entropy(action_logits, action_targets)
        reward_loss = F.smooth_l1_loss(reward_pred, reward_targets)
        transition_loss = F.cross_entropy(next_state_logits, next_state_targets, weight=class_weights)
        ent = entropy_from_logits(state_logits)

        loss = (
            state_loss
            + cfg.action_coef * action_loss
            + cfg.reward_coef * reward_loss
            + cfg.transition_coef * transition_loss
            - cfg.entropy_coef * ent
        )

        optimizer.zero_grad(set_to_none=True)
        loss.backward()
        nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        optimizer.step()

        total_loss += float(loss.item())

    return total_loss / max(1, len(loader))


@torch.no_grad()
def evaluate(model, loader, class_weights, device, cfg):
    model.eval()

    total_loss = 0.0
    y_true = []
    y_pred = []
    a_true = []
    a_pred = []

    for batch in loader:
        obs = batch["obs"].to(device)
        actions = batch["actions"].to(device)
        rewards = batch["rewards"].to(device)
        states = batch["states"].to(device)
        next_states = batch["next_states"].to(device)

        prev_actions = torch.roll(actions, shifts=1, dims=1)
        prev_rewards = torch.roll(rewards, shifts=1, dims=1)
        prev_actions[:, 0] = 0
        prev_rewards[:, 0] = 0.0

        out = model(obs, prev_actions, prev_rewards)

        B, T, S = out["state_logits"].shape
        A = out["action_logits"].shape[-1]

        state_logits = out["state_logits"].reshape(B * T, S)
        action_logits = out["action_logits"].reshape(B * T, A)
        next_state_logits = out["next_state_logits"].reshape(B * T, S)
        reward_pred = out["reward_pred"].reshape(B * T)

        state_targets = states.reshape(B * T)
        action_targets = actions.reshape(B * T)
        next_state_targets = next_states.reshape(B * T)
        reward_targets = rewards.reshape(B * T)

        state_loss = F.cross_entropy(state_logits, state_targets, weight=class_weights)
        action_loss = F.cross_entropy(action_logits, action_targets)
        reward_loss = F.smooth_l1_loss(reward_pred, reward_targets)
        transition_loss = F.cross_entropy(next_state_logits, next_state_targets, weight=class_weights)
        ent = entropy_from_logits(state_logits)

        loss = (
            state_loss
            + cfg.action_coef * action_loss
            + cfg.reward_coef * reward_loss
            + cfg.transition_coef * transition_loss
            - cfg.entropy_coef * ent
        )
        total_loss += float(loss.item())

        last_state_pred = out["state_logits"][:, -1].argmax(dim=-1).cpu().numpy()
        last_state_true = states[:, -1].cpu().numpy()

        last_action_pred = out["action_logits"][:, -1].argmax(dim=-1).cpu().numpy()
        last_action_true = actions[:, -1].cpu().numpy()

        y_true.append(last_state_true)
        y_pred.append(last_state_pred)
        a_true.append(last_action_true)
        a_pred.append(last_action_pred)

    y_true = np.concatenate(y_true)
    y_pred = np.concatenate(y_pred)
    a_true = np.concatenate(a_true)
    a_pred = np.concatenate(a_pred)

    state_acc = float((y_true == y_pred).mean())
    action_acc = float((a_true == a_pred).mean())

    recalls = []
    for c in range(len(STATE_NAMES)):
        idx = (y_true == c)
        recalls.append(float((y_pred[idx] == c).mean()) if idx.sum() > 0 else 0.0)
    macro_recall = float(np.mean(recalls))

    cm = confusion_matrix(y_true, y_pred, len(STATE_NAMES))

    return {
        "loss": total_loss / max(1, len(loader)),
        "state_acc": state_acc,
        "action_acc": action_acc,
        "macro_recall": macro_recall,
        "cm": cm,
    }


@torch.no_grad()
def profile_eval(model, device):
    print("\nPolicy evaluation on multiple user profiles")
    print("-" * 90)

    for profile_name, init_state in [
        ("profile_1", STATE2ID["rested_focused"]),
        ("profile_2", STATE2ID["sleep_deprived_slow"]),
        ("profile_3", STATE2ID["emotionally_reactive"]),
        ("profile_4", STATE2ID["high_noise_overloaded"]),
    ]:
        state = init_state
        start_hour = np.random.randint(0, 24)

        obs_seq = []
        prev_actions = []
        prev_rewards = []
        reward_list = []
        inferred = None

        for t in range(12):
            hour = (start_hour + t) % 24
            obs = generate_observation(state, hour)
            obs_seq.append(obs)

            if t == 0:
                prev_actions.append(0)
                prev_rewards.append(0.0)

            obs_t = torch.tensor(np.array(obs_seq), dtype=torch.float32).unsqueeze(0).to(device)
            act_t = torch.tensor(np.array(prev_actions), dtype=torch.long).unsqueeze(0).to(device)
            rew_t = torch.tensor(np.array(prev_rewards), dtype=torch.float32).unsqueeze(0).to(device)

            out = model(obs_t, act_t, rew_t)
            inferred = out["state_logits"][:, -1].argmax(dim=-1).item()
            action = out["action_logits"][:, -1].argmax(dim=-1).item()

            reward = reward_from_state_action(state, action)
            reward_list.append(reward)

            if len(prev_actions) < len(obs_seq):
                prev_actions.append(action)
                prev_rewards.append(reward)
            else:
                prev_actions[-1] = action
                prev_rewards[-1] = reward

            state = transition_next_state(state, action)

        print(
            f"{profile_name}: true={STATE_NAMES[init_state]:<22} | "
            f"avg_reward={np.mean(reward_list):.3f} | "
            f"last_inferred={STATE_NAMES[inferred]}"
        )


# ============================================================
# Main
# ============================================================

def main():
    args = parse_args()
    set_seed(args.seed)

    device = torch.device(args.device if torch.cuda.is_available() else "cpu")

    cfg = Config(
        epochs=args.epochs,
        batch_size=args.batch_size,
        lr=args.lr,
        device=str(device),
        seed=args.seed,
    )

    print("=" * 80)
    print("Improved Sleep Research Training")
    print("=" * 80)
    print(f"device       = {device}")
    print(f"epochs       = {cfg.epochs}")
    print(f"batch_size   = {cfg.batch_size}")
    print(f"lr           = {cfg.lr}")
    print("=" * 80)

    train_ds = SleepSequenceDataset(cfg.train_samples, cfg.seq_len)
    val_ds = SleepSequenceDataset(cfg.val_samples, cfg.seq_len)
    test_ds = SleepSequenceDataset(cfg.test_samples, cfg.seq_len)

    train_loader = build_loader(train_ds, cfg.batch_size, balanced=True)
    val_loader = build_loader(val_ds, cfg.batch_size, balanced=False)
    test_loader = build_loader(test_ds, cfg.batch_size, balanced=False)

    class_weights = compute_class_weights(train_ds, device)

    model = SleepBeliefPolicyModel(cfg).to(device)
    optimizer = torch.optim.AdamW(model.parameters(), lr=cfg.lr, weight_decay=cfg.weight_decay)
    scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=cfg.epochs)

    best_score = -1.0

    for epoch in range(1, cfg.epochs + 1):
        train_loss = train_one_epoch(model, train_loader, optimizer, class_weights, device, cfg)
        val_stats = evaluate(model, val_loader, class_weights, device, cfg)
        scheduler.step()

        score = 0.7 * val_stats["macro_recall"] + 0.3 * val_stats["state_acc"]
        if score > best_score:
            best_score = score
            torch.save(
                {
                    "model_state_dict": model.state_dict(),
                    "best_score": best_score,
                    "config": cfg.__dict__,
                },
                args.save_path,
            )

        print(
            f"[epoch {epoch:02d}] "
            f"train_loss={train_loss:.4f} | "
            f"val_loss={val_stats['loss']:.4f} | "
            f"state_acc={val_stats['state_acc']:.4f} | "
            f"macro_recall={val_stats['macro_recall']:.4f} | "
            f"action_acc={val_stats['action_acc']:.4f}"
        )

    print("\nLoading best model...")
    ckpt = torch.load(args.save_path, map_location=device)
    model.load_state_dict(ckpt["model_state_dict"])

    test_stats = evaluate(model, test_loader, class_weights, device, cfg)

    print("\n" + "=" * 80)
    print("FINAL TEST RESULT")
    print("=" * 80)
    print(f"state_acc    : {test_stats['state_acc']:.4f}")
    print(f"macro_recall : {test_stats['macro_recall']:.4f}")
    print(f"action_acc   : {test_stats['action_acc']:.4f}")
    print(f"test_loss    : {test_stats['loss']:.4f}")
    print("=" * 80)

    print_confusion_matrix(test_stats["cm"])
    profile_eval(model, device)


if __name__ == "__main__":
    main()
```

```python
# ============================================================
# sleep_research_hard_jupyter_safe.py
# ============================================================
from __future__ import annotations

import argparse
import math
import random
from dataclasses import dataclass

import numpy as np
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader, WeightedRandomSampler


# ============================================================
# Jupyter-safe argparse
# ============================================================
def parse_args():
    parser = argparse.ArgumentParser()
    parser.add_argument("--epochs", type=int, default=30)
    parser.add_argument("--batch_size", type=int, default=128)
    parser.add_argument("--lr", type=float, default=3e-4)
    parser.add_argument("--device", type=str, default="cuda")
    parser.add_argument("--seed", type=int, default=42)
    parser.add_argument("--save_path", type=str, default="./best_sleep_research_hard.pt")
    parser.add_argument("--train_samples", type=int, default=16000)
    parser.add_argument("--val_samples", type=int, default=2500)
    parser.add_argument("--test_samples", type=int, default=2500)
    args, _ = parser.parse_known_args()
    return args


# ============================================================
# Reproducibility
# ============================================================
def set_seed(seed: int = 42):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)


# ============================================================
# Definitions
# ============================================================
STATE_NAMES = [
    "rested_focused",
    "sleep_deprived_slow",
    "high_noise_overloaded",
    "emotionally_reactive",
]
STATE2ID = {s: i for i, s in enumerate(STATE_NAMES)}

ACTION_NAMES = [
    "normal_guidance",
    "low_complexity_checkin",
    "step_by_step_support",
    "emotional_validation",
    "noise_reduction_prompt",
    "sleep_hygiene_prompt",
]
ACTION2ID = {a: i for i, a in enumerate(ACTION_NAMES)}


@dataclass
class Config:
    seq_len: int = 12
    obs_dim: int = 11
    hidden_dim: int = 160
    num_layers: int = 1
    dropout: float = 0.12

    train_samples: int = 16000
    val_samples: int = 2500
    test_samples: int = 2500

    epochs: int = 30
    batch_size: int = 128
    lr: float = 3e-4
    weight_decay: float = 1e-4

    action_coef: float = 0.35
    reward_coef: float = 0.20
    transition_coef: float = 0.30
    entropy_coef: float = 0.002

    label_smoothing: float = 0.03

    train_shift: float = 0.00
    val_shift: float = 0.03
    test_shift: float = 0.05

    device: str = "cuda"
    seed: int = 42


# ============================================================
# Utilities
# ============================================================
def clamp01(x: float) -> float:
    return max(0.0, min(1.0, x))


def soft_cross_entropy(logits: torch.Tensor, target_probs: torch.Tensor) -> torch.Tensor:
    log_probs = F.log_softmax(logits, dim=-1)
    return -(target_probs * log_probs).sum(dim=-1).mean()


def entropy_from_logits(logits: torch.Tensor) -> torch.Tensor:
    p = F.softmax(logits, dim=-1)
    logp = F.log_softmax(logits, dim=-1)
    return -(p * logp).sum(dim=-1).mean()


def make_smoothed_targets(targets: torch.Tensor, num_classes: int, smoothing: float) -> torch.Tensor:
    with torch.no_grad():
        dist = torch.full(
            (targets.shape[0], num_classes),
            fill_value=smoothing / (num_classes - 1),
            device=targets.device,
            dtype=torch.float32,
        )
        dist.scatter_(1, targets.unsqueeze(1), 1.0 - smoothing)
    return dist


# ============================================================
# Synthetic environment: harder version
# ============================================================
def state_proto(state_id: int):
    """
    일부러 상태 간 feature overlap을 남겨둠.
    """
    if state_id == STATE2ID["rested_focused"]:
        return {
            "response_time": 0.34,
            "typing_var": 0.30,
            "sleep_score": 0.80,
            "noise_load": 0.28,
            "affect_intensity": 0.24,
            "engagement": 0.82,
            "adherence": 0.84,
            "fragmentation": 0.22,
            "latency_instability": 0.26,
            "focus_drift": 0.22,
        }
    if state_id == STATE2ID["sleep_deprived_slow"]:
        return {
            "response_time": 0.70,
            "typing_var": 0.42,
            "sleep_score": 0.30,
            "noise_load": 0.40,
            "affect_intensity": 0.34,
            "engagement": 0.66,
            "adherence": 0.68,
            "fragmentation": 0.58,
            "latency_instability": 0.64,
            "focus_drift": 0.58,
        }
    if state_id == STATE2ID["high_noise_overloaded"]:
        return {
            "response_time": 0.58,
            "typing_var": 0.60,
            "sleep_score": 0.44,
            "noise_load": 0.76,
            "affect_intensity": 0.40,
            "engagement": 0.60,
            "adherence": 0.60,
            "fragmentation": 0.50,
            "latency_instability": 0.56,
            "focus_drift": 0.64,
        }
    if state_id == STATE2ID["emotionally_reactive"]:
        return {
            "response_time": 0.52,
            "typing_var": 0.58,
            "sleep_score": 0.46,
            "noise_load": 0.36,
            "affect_intensity": 0.78,
            "engagement": 0.56,
            "adherence": 0.54,
            "fragmentation": 0.46,
            "latency_instability": 0.50,
            "focus_drift": 0.56,
        }
    raise ValueError("invalid state")


def best_action_for_state(state_id: int) -> int:
    if state_id == STATE2ID["rested_focused"]:
        return ACTION2ID["normal_guidance"]
    if state_id == STATE2ID["sleep_deprived_slow"]:
        return ACTION2ID["low_complexity_checkin"]
    if state_id == STATE2ID["high_noise_overloaded"]:
        return ACTION2ID["noise_reduction_prompt"]
    if state_id == STATE2ID["emotionally_reactive"]:
        return ACTION2ID["emotional_validation"]
    raise ValueError("invalid state")


def secondary_good_actions(state_id: int):
    if state_id == STATE2ID["rested_focused"]:
        return {ACTION2ID["step_by_step_support"]}
    if state_id == STATE2ID["sleep_deprived_slow"]:
        return {
            ACTION2ID["step_by_step_support"],
            ACTION2ID["sleep_hygiene_prompt"],
        }
    if state_id == STATE2ID["high_noise_overloaded"]:
        return {
            ACTION2ID["low_complexity_checkin"],
            ACTION2ID["step_by_step_support"],
        }
    if state_id == STATE2ID["emotionally_reactive"]:
        return {
            ACTION2ID["low_complexity_checkin"],
            ACTION2ID["step_by_step_support"],
        }
    return set()


def reward_from_state_action(state_id: int, action_id: int) -> float:
    best = best_action_for_state(state_id)
    if action_id == best:
        return np.random.uniform(0.84, 0.96)
    if action_id in secondary_good_actions(state_id):
        return np.random.uniform(0.64, 0.82)
    return np.random.uniform(0.30, 0.62)


def transition_next_state(curr_state: int, action_id: int) -> int:
    """
    전이도 너무 쉽지 않게.
    wrong action이면 완전히 다른 상태로만 튀지 않고, 애매한 인접 상태로도 이동.
    """
    p = np.random.rand()
    best = best_action_for_state(curr_state)

    if action_id == best:
        if curr_state == STATE2ID["sleep_deprived_slow"]:
            if p < 0.15:
                return STATE2ID["rested_focused"]
            return curr_state
        if curr_state == STATE2ID["high_noise_overloaded"]:
            if p < 0.14:
                return STATE2ID["rested_focused"]
            if p < 0.22:
                return STATE2ID["sleep_deprived_slow"]
            return curr_state
        if curr_state == STATE2ID["emotionally_reactive"]:
            if p < 0.16:
                return STATE2ID["rested_focused"]
            if p < 0.25:
                return STATE2ID["high_noise_overloaded"]
            return curr_state
        return curr_state

    if curr_state == STATE2ID["rested_focused"]:
        if action_id == ACTION2ID["sleep_hygiene_prompt"] and p < 0.10:
            return STATE2ID["sleep_deprived_slow"]
        if action_id == ACTION2ID["noise_reduction_prompt"] and p < 0.08:
            return STATE2ID["high_noise_overloaded"]
        return curr_state

    if curr_state == STATE2ID["sleep_deprived_slow"]:
        if action_id == ACTION2ID["normal_guidance"] and p < 0.28:
            return STATE2ID["high_noise_overloaded"]
        if action_id == ACTION2ID["emotional_validation"] and p < 0.15:
            return STATE2ID["emotionally_reactive"]
        return curr_state

    if curr_state == STATE2ID["high_noise_overloaded"]:
        if action_id == ACTION2ID["normal_guidance"] and p < 0.25:
            return STATE2ID["emotionally_reactive"]
        if action_id == ACTION2ID["sleep_hygiene_prompt"] and p < 0.16:
            return STATE2ID["sleep_deprived_slow"]
        return curr_state

    if curr_state == STATE2ID["emotionally_reactive"]:
        if action_id == ACTION2ID["normal_guidance"] and p < 0.28:
            return STATE2ID["high_noise_overloaded"]
        if action_id == ACTION2ID["sleep_hygiene_prompt"] and p < 0.13:
            return STATE2ID["sleep_deprived_slow"]
        return curr_state

    return curr_state


def generate_observation(
    state_id: int,
    hour: int,
    domain_shift: float = 0.0,
    mask_prob: float = 0.18,
) -> np.ndarray:
    """
    obs 구성:
    0  hour_norm
    1  response_time
    2  typing_var
    3  sleep_score
    4  noise_load
    5  affect_intensity
    6  engagement
    7  adherence
    8  fragmentation
    9  latency_instability
    10 focus_drift
    """
    proto = state_proto(state_id)
    hour_norm = hour / 23.0

    circadian_sleep = 0.0
    circadian_cognition = 0.0
    if 0 <= hour <= 5:
        circadian_sleep = 0.12
        circadian_cognition = 0.08
    elif 6 <= hour <= 9:
        circadian_sleep = -0.06
        circadian_cognition = -0.03
    elif 22 <= hour <= 23:
        circadian_sleep = 0.08
        circadian_cognition = 0.05

    # 도메인 쉬프트: val/test에서 평균이 살짝 이동
    shift = domain_shift

    # 노이즈를 꽤 크게 줘서 완전 분리 방지
    response_time = clamp01(np.random.normal(proto["response_time"] + 0.03 * shift, 0.10))
    typing_var = clamp01(np.random.normal(proto["typing_var"] + 0.02 * shift, 0.11))
    sleep_score = clamp01(np.random.normal(proto["sleep_score"] - circadian_sleep - 0.03 * shift, 0.12))
    noise_load = clamp01(np.random.normal(proto["noise_load"] + 0.04 * shift, 0.12))
    affect_intensity = clamp01(np.random.normal(proto["affect_intensity"] + 0.03 * shift, 0.12))
    engagement = clamp01(np.random.normal(proto["engagement"] - 0.02 * shift, 0.10))
    adherence = clamp01(np.random.normal(proto["adherence"] - 0.02 * shift, 0.10))
    fragmentation = clamp01(np.random.normal(proto["fragmentation"] + 0.02 * shift, 0.11))
    latency_instability = clamp01(np.random.normal(proto["latency_instability"] + circadian_cognition, 0.10))
    focus_drift = clamp01(np.random.normal(proto["focus_drift"] + circadian_cognition, 0.11))

    # 상태별 구분 신호를 약하게만 추가
    if state_id == STATE2ID["sleep_deprived_slow"]:
        response_time = clamp01(response_time + 0.03)
        sleep_score = clamp01(sleep_score - 0.04)
        fragmentation = clamp01(fragmentation + 0.03)

    elif state_id == STATE2ID["high_noise_overloaded"]:
        noise_load = clamp01(noise_load + 0.04)
        typing_var = clamp01(typing_var + 0.02)
        focus_drift = clamp01(focus_drift + 0.03)

    elif state_id == STATE2ID["emotionally_reactive"]:
        affect_intensity = clamp01(affect_intensity + 0.04)
        typing_var = clamp01(typing_var + 0.02)

    obs = np.array(
        [
            hour_norm,
            response_time,
            typing_var,
            sleep_score,
            noise_load,
            affect_intensity,
            engagement,
            adherence,
            fragmentation,
            latency_instability,
            focus_drift,
        ],
        dtype=np.float32,
    )

    # partial observability: 일부 핵심 관측 마스킹
    # 3:sleep_score, 4:noise_load, 5:affect_intensity, 8,9,10
    candidate_idx = [3, 4, 5, 8, 9, 10]
    for idx in candidate_idx:
        if np.random.rand() < mask_prob:
            obs[idx] = 0.5  # missing-like neutral token

    return obs


def choose_behavior_action(state_id: int) -> int:
    """
    behavior policy를 완전 정답으로 고정하지 않음.
    """
    r = np.random.rand()
    best = best_action_for_state(state_id)
    secondary = list(secondary_good_actions(state_id))
    all_actions = list(range(len(ACTION_NAMES)))

    if r < 0.58:
        return best
    if r < 0.83 and len(secondary) > 0:
        return random.choice(secondary)
    bads = [a for a in all_actions if a != best and a not in secondary]
    return random.choice(bads)


def generate_sequence(seq_len: int, domain_shift: float = 0.0):
    state = np.random.choice(4, p=[0.28, 0.27, 0.23, 0.22])
    start_hour = np.random.randint(0, 24)

    obs_seq = []
    action_seq = []
    reward_seq = []
    state_seq = []
    next_state_seq = []

    for t in range(seq_len):
        hour = (start_hour + t * np.random.randint(1, 4)) % 24
        obs = generate_observation(
            state_id=state,
            hour=hour,
            domain_shift=domain_shift,
            mask_prob=0.18 if t < seq_len - 1 else 0.10,
        )

        action = choose_behavior_action(state)
        reward = reward_from_state_action(state, action)
        next_state = transition_next_state(state, action)

        obs_seq.append(obs)
        action_seq.append(action)
        reward_seq.append(reward)
        state_seq.append(state)
        next_state_seq.append(next_state)

        state = next_state

    return {
        "obs": np.stack(obs_seq),
        "actions": np.array(action_seq, dtype=np.int64),
        "rewards": np.array(reward_seq, dtype=np.float32),
        "states": np.array(state_seq, dtype=np.int64),
        "next_states": np.array(next_state_seq, dtype=np.int64),
    }


# ============================================================
# Dataset
# ============================================================
class SleepSequenceDataset(Dataset):
    def __init__(self, n_samples: int, seq_len: int, domain_shift: float):
        self.samples = [generate_sequence(seq_len, domain_shift=domain_shift) for _ in range(n_samples)]
        self.labels = [int(x["states"][-1]) for x in self.samples]

        counts = np.bincount(self.labels, minlength=len(STATE_NAMES))
        inv = 1.0 / np.maximum(counts, 1)
        self.sample_weights = [inv[y] for y in self.labels]

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):
        x = self.samples[idx]
        return {
            "obs": torch.tensor(x["obs"], dtype=torch.float32),
            "actions": torch.tensor(x["actions"], dtype=torch.long),
            "rewards": torch.tensor(x["rewards"], dtype=torch.float32),
            "states": torch.tensor(x["states"], dtype=torch.long),
            "next_states": torch.tensor(x["next_states"], dtype=torch.long),
        }


def build_loader(dataset: SleepSequenceDataset, batch_size: int, balanced: bool):
    if balanced:
        sampler = WeightedRandomSampler(
            weights=dataset.sample_weights,
            num_samples=len(dataset),
            replacement=True,
        )
        return DataLoader(dataset, batch_size=batch_size, sampler=sampler, drop_last=False)
    return DataLoader(dataset, batch_size=batch_size, shuffle=False, drop_last=False)


# ============================================================
# Model
# ============================================================
class SleepBeliefPolicyModel(nn.Module):
    def __init__(self, cfg: Config):
        super().__init__()

        self.obs_encoder = nn.Sequential(
            nn.Linear(cfg.obs_dim, cfg.hidden_dim),
            nn.LayerNorm(cfg.hidden_dim),
            nn.GELU(),
            nn.Dropout(cfg.dropout),
            nn.Linear(cfg.hidden_dim, cfg.hidden_dim),
            nn.LayerNorm(cfg.hidden_dim),
            nn.GELU(),
        )

        self.action_emb = nn.Embedding(len(ACTION_NAMES), cfg.hidden_dim // 4)
        self.reward_proj = nn.Linear(1, cfg.hidden_dim // 8)

        rnn_in_dim = cfg.hidden_dim + cfg.hidden_dim // 4 + cfg.hidden_dim // 8

        self.gru = nn.GRU(
            input_size=rnn_in_dim,
            hidden_size=cfg.hidden_dim,
            num_layers=cfg.num_layers,
            batch_first=True,
            dropout=cfg.dropout if cfg.num_layers > 1 else 0.0,
        )

        self.shared = nn.Sequential(
            nn.Linear(cfg.hidden_dim, cfg.hidden_dim),
            nn.LayerNorm(cfg.hidden_dim),
            nn.GELU(),
            nn.Dropout(cfg.dropout),
        )

        self.state_head = nn.Linear(cfg.hidden_dim, len(STATE_NAMES))
        self.action_head = nn.Linear(cfg.hidden_dim, len(ACTION_NAMES))
        self.reward_head = nn.Linear(cfg.hidden_dim, 1)
        self.next_state_head = nn.Linear(cfg.hidden_dim, len(STATE_NAMES))

    def forward(self, obs, prev_actions, prev_rewards):
        obs_feat = self.obs_encoder(obs)
        act_feat = self.action_emb(prev_actions)
        rew_feat = self.reward_proj(prev_rewards.unsqueeze(-1))

        x = torch.cat([obs_feat, act_feat, rew_feat], dim=-1)
        h, _ = self.gru(x)
        z = self.shared(h)

        return {
            "state_logits": self.state_head(z),
            "action_logits": self.action_head(z),
            "reward_pred": self.reward_head(z).squeeze(-1),
            "next_state_logits": self.next_state_head(z),
        }


# ============================================================
# Metrics
# ============================================================
def compute_class_weights(dataset, device):
    counts = np.bincount(dataset.labels, minlength=len(STATE_NAMES)).astype(np.float32)
    w = counts.sum() / np.maximum(counts, 1.0)
    w = w / w.mean()
    return torch.tensor(w, dtype=torch.float32, device=device)


def confusion_matrix(y_true, y_pred, n_cls):
    mat = np.zeros((n_cls, n_cls), dtype=np.int64)
    for t, p in zip(y_true, y_pred):
        mat[t, p] += 1
    return mat


def print_confusion_matrix(cm):
    print("\n[Confusion Matrix] rows=true cols=pred")
    print(" " * 24 + " ".join([f"{s[:10]:>12}" for s in STATE_NAMES]))
    for i, row in enumerate(cm):
        print(f"{STATE_NAMES[i]:>24} " + " ".join([f"{int(v):12d}" for v in row]))


# ============================================================
# Train / Eval
# ============================================================
def train_one_epoch(model, loader, optimizer, class_weights, device, cfg):
    model.train()
    total_loss = 0.0

    for batch in loader:
        obs = batch["obs"].to(device)
        actions = batch["actions"].to(device)
        rewards = batch["rewards"].to(device)
        states = batch["states"].to(device)
        next_states = batch["next_states"].to(device)

        prev_actions = torch.roll(actions, shifts=1, dims=1)
        prev_rewards = torch.roll(rewards, shifts=1, dims=1)
        prev_actions[:, 0] = 0
        prev_rewards[:, 0] = 0.0

        out = model(obs, prev_actions, prev_rewards)

        B, T, S = out["state_logits"].shape
        A = out["action_logits"].shape[-1]

        state_logits = out["state_logits"].reshape(B * T, S)
        action_logits = out["action_logits"].reshape(B * T, A)
        next_state_logits = out["next_state_logits"].reshape(B * T, S)
        reward_pred = out["reward_pred"].reshape(B * T)

        state_targets = states.reshape(B * T)
        action_targets = actions.reshape(B * T)
        next_state_targets = next_states.reshape(B * T)
        reward_targets = rewards.reshape(B * T)

        smoothed = make_smoothed_targets(state_targets, S, cfg.label_smoothing)
        next_smoothed = make_smoothed_targets(next_state_targets, S, cfg.label_smoothing)

        # class weight를 soft CE에 직접 곱하기 위해 sample-wise 가중 적용
        log_probs_state = F.log_softmax(state_logits, dim=-1)
        state_loss_vec = -(smoothed * log_probs_state).sum(dim=-1) * class_weights[state_targets]
        state_loss = state_loss_vec.mean()

        log_probs_next = F.log_softmax(next_state_logits, dim=-1)
        transition_loss_vec = -(next_smoothed * log_probs_next).sum(dim=-1) * class_weights[next_state_targets]
        transition_loss = transition_loss_vec.mean()

        action_loss = F.cross_entropy(action_logits, action_targets)
        reward_loss = F.smooth_l1_loss(reward_pred, reward_targets)
        ent = entropy_from_logits(state_logits)

        loss = (
            state_loss
            + cfg.action_coef * action_loss
            + cfg.reward_coef * reward_loss
            + cfg.transition_coef * transition_loss
            - cfg.entropy_coef * ent
        )

        optimizer.zero_grad(set_to_none=True)
        loss.backward()
        nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        optimizer.step()

        total_loss += float(loss.item())

    return total_loss / max(1, len(loader))


@torch.no_grad()
def evaluate(model, loader, class_weights, device, cfg):
    model.eval()

    total_loss = 0.0
    y_true = []
    y_pred = []
    a_true = []
    a_pred = []

    for batch in loader:
        obs = batch["obs"].to(device)
        actions = batch["actions"].to(device)
        rewards = batch["rewards"].to(device)
        states = batch["states"].to(device)
        next_states = batch["next_states"].to(device)

        prev_actions = torch.roll(actions, shifts=1, dims=1)
        prev_rewards = torch.roll(rewards, shifts=1, dims=1)
        prev_actions[:, 0] = 0
        prev_rewards[:, 0] = 0.0

        out = model(obs, prev_actions, prev_rewards)

        B, T, S = out["state_logits"].shape
        A = out["action_logits"].shape[-1]

        state_logits = out["state_logits"].reshape(B * T, S)
        action_logits = out["action_logits"].reshape(B * T, A)
        next_state_logits = out["next_state_logits"].reshape(B * T, S)
        reward_pred = out["reward_pred"].reshape(B * T)

        state_targets = states.reshape(B * T)
        action_targets = actions.reshape(B * T)
        next_state_targets = next_states.reshape(B * T)
        reward_targets = rewards.reshape(B * T)

        smoothed = make_smoothed_targets(state_targets, S, cfg.label_smoothing)
        next_smoothed = make_smoothed_targets(next_state_targets, S, cfg.label_smoothing)

        log_probs_state = F.log_softmax(state_logits, dim=-1)
        state_loss_vec = -(smoothed * log_probs_state).sum(dim=-1) * class_weights[state_targets]
        state_loss = state_loss_vec.mean()

        log_probs_next = F.log_softmax(next_state_logits, dim=-1)
        transition_loss_vec = -(next_smoothed * log_probs_next).sum(dim=-1) * class_weights[next_state_targets]
        transition_loss = transition_loss_vec.mean()

        action_loss = F.cross_entropy(action_logits, action_targets)
        reward_loss = F.smooth_l1_loss(reward_pred, reward_targets)
        ent = entropy_from_logits(state_logits)

        loss = (
            state_loss
            + cfg.action_coef * action_loss
            + cfg.reward_coef * reward_loss
            + cfg.transition_coef * transition_loss
            - cfg.entropy_coef * ent
        )
        total_loss += float(loss.item())

        last_state_pred = out["state_logits"][:, -1].argmax(dim=-1).cpu().numpy()
        last_state_true = states[:, -1].cpu().numpy()

        last_action_pred = out["action_logits"][:, -1].argmax(dim=-1).cpu().numpy()
        last_action_true = actions[:, -1].cpu().numpy()

        y_true.append(last_state_true)
        y_pred.append(last_state_pred)
        a_true.append(last_action_true)
        a_pred.append(last_action_pred)

    y_true = np.concatenate(y_true)
    y_pred = np.concatenate(y_pred)
    a_true = np.concatenate(a_true)
    a_pred = np.concatenate(a_pred)

    state_acc = float((y_true == y_pred).mean())
    action_acc = float((a_true == a_pred).mean())

    recalls = []
    for c in range(len(STATE_NAMES)):
        idx = (y_true == c)
        recalls.append(float((y_pred[idx] == c).mean()) if idx.sum() > 0 else 0.0)
    macro_recall = float(np.mean(recalls))

    cm = confusion_matrix(y_true, y_pred, len(STATE_NAMES))

    return {
        "loss": total_loss / max(1, len(loader)),
        "state_acc": state_acc,
        "action_acc": action_acc,
        "macro_recall": macro_recall,
        "cm": cm,
    }


# ============================================================
# Profile evaluation
# ============================================================
@torch.no_grad()
def profile_eval(model, device):
    model.eval()

    print("\nPolicy evaluation on multiple user profiles")
    print("-" * 90)

    profiles = [
        ("profile_1", STATE2ID["rested_focused"]),
        ("profile_2", STATE2ID["sleep_deprived_slow"]),
        ("profile_3", STATE2ID["emotionally_reactive"]),
        ("profile_4", STATE2ID["high_noise_overloaded"]),
    ]

    for profile_name, init_state in profiles:
        state = init_state
        start_hour = np.random.randint(0, 24)

        obs_seq = []
        action_history = []
        reward_history = []
        reward_list = []
        inferred = None

        for t in range(12):
            hour = (start_hour + t) % 24
            obs = generate_observation(
                state_id=state,
                hour=hour,
                domain_shift=0.04,
                mask_prob=0.15,
            )
            obs_seq.append(obs)

            # 길이를 반드시 obs_seq와 동일하게 유지
            prev_actions = [0] + action_history
            prev_rewards = [0.0] + reward_history

            prev_actions = prev_actions[: len(obs_seq)]
            prev_rewards = prev_rewards[: len(obs_seq)]

            while len(prev_actions) < len(obs_seq):
                prev_actions.append(0)
            while len(prev_rewards) < len(obs_seq):
                prev_rewards.append(0.0)

            obs_t = torch.tensor(np.array(obs_seq), dtype=torch.float32).unsqueeze(0).to(device)
            act_t = torch.tensor(np.array(prev_actions), dtype=torch.long).unsqueeze(0).to(device)
            rew_t = torch.tensor(np.array(prev_rewards), dtype=torch.float32).unsqueeze(0).to(device)

            out = model(obs_t, act_t, rew_t)
            inferred = out["state_logits"][:, -1].argmax(dim=-1).item()
            action = out["action_logits"][:, -1].argmax(dim=-1).item()

            reward = reward_from_state_action(state, action)
            reward_list.append(reward)

            action_history.append(action)
            reward_history.append(reward)

            state = transition_next_state(state, action)

        print(
            f"{profile_name}: true={STATE_NAMES[init_state]:<22} | "
            f"avg_reward={np.mean(reward_list):.3f} | "
            f"last_inferred={STATE_NAMES[inferred]}"
        )


# ============================================================
# Main
# ============================================================
def main():
    args = parse_args()
    set_seed(args.seed)

    device = torch.device(args.device if torch.cuda.is_available() else "cpu")

    cfg = Config(
        epochs=args.epochs,
        batch_size=args.batch_size,
        lr=args.lr,
        train_samples=args.train_samples,
        val_samples=args.val_samples,
        test_samples=args.test_samples,
        device=str(device),
        seed=args.seed,
    )

    print("=" * 80)
    print("Hard Sleep Research Training")
    print("=" * 80)
    print(f"device       = {device}")
    print(f"epochs       = {cfg.epochs}")
    print(f"batch_size   = {cfg.batch_size}")
    print(f"lr           = {cfg.lr}")
    print(f"train/val/test = {cfg.train_samples}/{cfg.val_samples}/{cfg.test_samples}")
    print("=" * 80)

    train_ds = SleepSequenceDataset(cfg.train_samples, cfg.seq_len, domain_shift=cfg.train_shift)
    val_ds = SleepSequenceDataset(cfg.val_samples, cfg.seq_len, domain_shift=cfg.val_shift)
    test_ds = SleepSequenceDataset(cfg.test_samples, cfg.seq_len, domain_shift=cfg.test_shift)

    train_loader = build_loader(train_ds, cfg.batch_size, balanced=True)
    val_loader = build_loader(val_ds, cfg.batch_size, balanced=False)
    test_loader = build_loader(test_ds, cfg.batch_size, balanced=False)

    class_weights = compute_class_weights(train_ds, device)

    model = SleepBeliefPolicyModel(cfg).to(device)
    optimizer = torch.optim.AdamW(model.parameters(), lr=cfg.lr, weight_decay=cfg.weight_decay)
    scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=cfg.epochs)

    best_score = -1.0

    for epoch in range(1, cfg.epochs + 1):
        train_loss = train_one_epoch(model, train_loader, optimizer, class_weights, device, cfg)
        val_stats = evaluate(model, val_loader, class_weights, device, cfg)
        scheduler.step()

        score = 0.7 * val_stats["macro_recall"] + 0.3 * val_stats["state_acc"]
        if score > best_score:
            best_score = score
            torch.save(
                {
                    "model_state_dict": model.state_dict(),
                    "best_score": best_score,
                    "config": cfg.__dict__,
                },
                args.save_path,
            )

        print(
            f"[epoch {epoch:02d}] "
            f"train_loss={train_loss:.4f} | "
            f"val_loss={val_stats['loss']:.4f} | "
            f"state_acc={val_stats['state_acc']:.4f} | "
            f"macro_recall={val_stats['macro_recall']:.4f} | "
            f"action_acc={val_stats['action_acc']:.4f}"
        )

    print("\nLoading best model...")
    ckpt = torch.load(args.save_path, map_location=device)
    model.load_state_dict(ckpt["model_state_dict"])

    test_stats = evaluate(model, test_loader, class_weights, device, cfg)

    print("\n" + "=" * 80)
    print("FINAL TEST RESULT")
    print("=" * 80)
    print(f"state_acc    : {test_stats['state_acc']:.4f}")
    print(f"macro_recall : {test_stats['macro_recall']:.4f}")
    print(f"action_acc   : {test_stats['action_acc']:.4f}")
    print(f"test_loss    : {test_stats['loss']:.4f}")
    print("=" * 80)

    print_confusion_matrix(test_stats["cm"])
    profile_eval(model, device)


if __name__ == "__main__":
    main()
```

```python
# ============================================================
# sleep_research_hard_jupyter_safe.py
# ============================================================
from __future__ import annotations

import argparse
import math
import random
from dataclasses import dataclass

import numpy as np
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader, WeightedRandomSampler


# ============================================================
# Jupyter-safe argparse
# ============================================================
def parse_args():
    parser = argparse.ArgumentParser()
    parser.add_argument("--epochs", type=int, default=30)
    parser.add_argument("--batch_size", type=int, default=128)
    parser.add_argument("--lr", type=float, default=3e-4)
    parser.add_argument("--device", type=str, default="cuda")
    parser.add_argument("--seed", type=int, default=42)
    parser.add_argument("--save_path", type=str, default="./best_sleep_research_hard.pt")
    parser.add_argument("--train_samples", type=int, default=16000)
    parser.add_argument("--val_samples", type=int, default=2500)
    parser.add_argument("--test_samples", type=int, default=2500)
    args, _ = parser.parse_known_args()
    return args


# ============================================================
# Reproducibility
# ============================================================
def set_seed(seed: int = 42):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)


# ============================================================
# Definitions
# ============================================================
STATE_NAMES = [
    "rested_focused",
    "sleep_deprived_slow",
    "high_noise_overloaded",
    "emotionally_reactive",
]
STATE2ID = {s: i for i, s in enumerate(STATE_NAMES)}

ACTION_NAMES = [
    "normal_guidance",
    "low_complexity_checkin",
    "step_by_step_support",
    "emotional_validation",
    "noise_reduction_prompt",
    "sleep_hygiene_prompt",
]
ACTION2ID = {a: i for i, a in enumerate(ACTION_NAMES)}


@dataclass
class Config:
    seq_len: int = 12
    obs_dim: int = 11
    hidden_dim: int = 160
    num_layers: int = 1
    dropout: float = 0.12

    train_samples: int = 16000
    val_samples: int = 2500
    test_samples: int = 2500

    epochs: int = 30
    batch_size: int = 128
    lr: float = 3e-4
    weight_decay: float = 1e-4

    action_coef: float = 0.35
    reward_coef: float = 0.20
    transition_coef: float = 0.30
    entropy_coef: float = 0.002

    label_smoothing: float = 0.03

    train_shift: float = 0.00
    val_shift: float = 0.03
    test_shift: float = 0.05

    device: str = "cuda"
    seed: int = 42


# ============================================================
# Utilities
# ============================================================
def clamp01(x: float) -> float:
    return max(0.0, min(1.0, x))


def soft_cross_entropy(logits: torch.Tensor, target_probs: torch.Tensor) -> torch.Tensor:
    log_probs = F.log_softmax(logits, dim=-1)
    return -(target_probs * log_probs).sum(dim=-1).mean()


def entropy_from_logits(logits: torch.Tensor) -> torch.Tensor:
    p = F.softmax(logits, dim=-1)
    logp = F.log_softmax(logits, dim=-1)
    return -(p * logp).sum(dim=-1).mean()


def make_smoothed_targets(targets: torch.Tensor, num_classes: int, smoothing: float) -> torch.Tensor:
    with torch.no_grad():
        dist = torch.full(
            (targets.shape[0], num_classes),
            fill_value=smoothing / (num_classes - 1),
            device=targets.device,
            dtype=torch.float32,
        )
        dist.scatter_(1, targets.unsqueeze(1), 1.0 - smoothing)
    return dist


# ============================================================
# Synthetic environment: harder version
# ============================================================
def state_proto(state_id: int):
    """
    일부러 상태 간 feature overlap을 남겨둠.
    """
    if state_id == STATE2ID["rested_focused"]:
        return {
            "response_time": 0.34,
            "typing_var": 0.30,
            "sleep_score": 0.80,
            "noise_load": 0.28,
            "affect_intensity": 0.24,
            "engagement": 0.82,
            "adherence": 0.84,
            "fragmentation": 0.22,
            "latency_instability": 0.26,
            "focus_drift": 0.22,
        }
    if state_id == STATE2ID["sleep_deprived_slow"]:
        return {
            "response_time": 0.70,
            "typing_var": 0.42,
            "sleep_score": 0.30,
            "noise_load": 0.40,
            "affect_intensity": 0.34,
            "engagement": 0.66,
            "adherence": 0.68,
            "fragmentation": 0.58,
            "latency_instability": 0.64,
            "focus_drift": 0.58,
        }
    if state_id == STATE2ID["high_noise_overloaded"]:
        return {
            "response_time": 0.58,
            "typing_var": 0.60,
            "sleep_score": 0.44,
            "noise_load": 0.76,
            "affect_intensity": 0.40,
            "engagement": 0.60,
            "adherence": 0.60,
            "fragmentation": 0.50,
            "latency_instability": 0.56,
            "focus_drift": 0.64,
        }
    if state_id == STATE2ID["emotionally_reactive"]:
        return {
            "response_time": 0.52,
            "typing_var": 0.58,
            "sleep_score": 0.46,
            "noise_load": 0.36,
            "affect_intensity": 0.78,
            "engagement": 0.56,
            "adherence": 0.54,
            "fragmentation": 0.46,
            "latency_instability": 0.50,
            "focus_drift": 0.56,
        }
    raise ValueError("invalid state")


def best_action_for_state(state_id: int) -> int:
    if state_id == STATE2ID["rested_focused"]:
        return ACTION2ID["normal_guidance"]
    if state_id == STATE2ID["sleep_deprived_slow"]:
        return ACTION2ID["low_complexity_checkin"]
    if state_id == STATE2ID["high_noise_overloaded"]:
        return ACTION2ID["noise_reduction_prompt"]
    if state_id == STATE2ID["emotionally_reactive"]:
        return ACTION2ID["emotional_validation"]
    raise ValueError("invalid state")


def secondary_good_actions(state_id: int):
    if state_id == STATE2ID["rested_focused"]:
        return {ACTION2ID["step_by_step_support"]}
    if state_id == STATE2ID["sleep_deprived_slow"]:
        return {
            ACTION2ID["step_by_step_support"],
            ACTION2ID["sleep_hygiene_prompt"],
        }
    if state_id == STATE2ID["high_noise_overloaded"]:
        return {
            ACTION2ID["low_complexity_checkin"],
            ACTION2ID["step_by_step_support"],
        }
    if state_id == STATE2ID["emotionally_reactive"]:
        return {
            ACTION2ID["low_complexity_checkin"],
            ACTION2ID["step_by_step_support"],
        }
    return set()


def reward_from_state_action(state_id: int, action_id: int) -> float:
    best = best_action_for_state(state_id)
    if action_id == best:
        return np.random.uniform(0.84, 0.96)
    if action_id in secondary_good_actions(state_id):
        return np.random.uniform(0.64, 0.82)
    return np.random.uniform(0.30, 0.62)


def transition_next_state(curr_state: int, action_id: int) -> int:
    """
    전이도 너무 쉽지 않게.
    wrong action이면 완전히 다른 상태로만 튀지 않고, 애매한 인접 상태로도 이동.
    """
    p = np.random.rand()
    best = best_action_for_state(curr_state)

    if action_id == best:
        if curr_state == STATE2ID["sleep_deprived_slow"]:
            if p < 0.15:
                return STATE2ID["rested_focused"]
            return curr_state
        if curr_state == STATE2ID["high_noise_overloaded"]:
            if p < 0.14:
                return STATE2ID["rested_focused"]
            if p < 0.22:
                return STATE2ID["sleep_deprived_slow"]
            return curr_state
        if curr_state == STATE2ID["emotionally_reactive"]:
            if p < 0.16:
                return STATE2ID["rested_focused"]
            if p < 0.25:
                return STATE2ID["high_noise_overloaded"]
            return curr_state
        return curr_state

    if curr_state == STATE2ID["rested_focused"]:
        if action_id == ACTION2ID["sleep_hygiene_prompt"] and p < 0.10:
            return STATE2ID["sleep_deprived_slow"]
        if action_id == ACTION2ID["noise_reduction_prompt"] and p < 0.08:
            return STATE2ID["high_noise_overloaded"]
        return curr_state

    if curr_state == STATE2ID["sleep_deprived_slow"]:
        if action_id == ACTION2ID["normal_guidance"] and p < 0.28:
            return STATE2ID["high_noise_overloaded"]
        if action_id == ACTION2ID["emotional_validation"] and p < 0.15:
            return STATE2ID["emotionally_reactive"]
        return curr_state

    if curr_state == STATE2ID["high_noise_overloaded"]:
        if action_id == ACTION2ID["normal_guidance"] and p < 0.25:
            return STATE2ID["emotionally_reactive"]
        if action_id == ACTION2ID["sleep_hygiene_prompt"] and p < 0.16:
            return STATE2ID["sleep_deprived_slow"]
        return curr_state

    if curr_state == STATE2ID["emotionally_reactive"]:
        if action_id == ACTION2ID["normal_guidance"] and p < 0.28:
            return STATE2ID["high_noise_overloaded"]
        if action_id == ACTION2ID["sleep_hygiene_prompt"] and p < 0.13:
            return STATE2ID["sleep_deprived_slow"]
        return curr_state

    return curr_state


def generate_observation(
    state_id: int,
    hour: int,
    domain_shift: float = 0.0,
    mask_prob: float = 0.18,
) -> np.ndarray:
    """
    obs 구성:
    0  hour_norm
    1  response_time
    2  typing_var
    3  sleep_score
    4  noise_load
    5  affect_intensity
    6  engagement
    7  adherence
    8  fragmentation
    9  latency_instability
    10 focus_drift
    """
    proto = state_proto(state_id)
    hour_norm = hour / 23.0

    circadian_sleep = 0.0
    circadian_cognition = 0.0
    if 0 <= hour <= 5:
        circadian_sleep = 0.12
        circadian_cognition = 0.08
    elif 6 <= hour <= 9:
        circadian_sleep = -0.06
        circadian_cognition = -0.03
    elif 22 <= hour <= 23:
        circadian_sleep = 0.08
        circadian_cognition = 0.05

    # 도메인 쉬프트: val/test에서 평균이 살짝 이동
    shift = domain_shift

    # 노이즈를 꽤 크게 줘서 완전 분리 방지
    response_time = clamp01(np.random.normal(proto["response_time"] + 0.03 * shift, 0.10))
    typing_var = clamp01(np.random.normal(proto["typing_var"] + 0.02 * shift, 0.11))
    sleep_score = clamp01(np.random.normal(proto["sleep_score"] - circadian_sleep - 0.03 * shift, 0.12))
    noise_load = clamp01(np.random.normal(proto["noise_load"] + 0.04 * shift, 0.12))
    affect_intensity = clamp01(np.random.normal(proto["affect_intensity"] + 0.03 * shift, 0.12))
    engagement = clamp01(np.random.normal(proto["engagement"] - 0.02 * shift, 0.10))
    adherence = clamp01(np.random.normal(proto["adherence"] - 0.02 * shift, 0.10))
    fragmentation = clamp01(np.random.normal(proto["fragmentation"] + 0.02 * shift, 0.11))
    latency_instability = clamp01(np.random.normal(proto["latency_instability"] + circadian_cognition, 0.10))
    focus_drift = clamp01(np.random.normal(proto["focus_drift"] + circadian_cognition, 0.11))

    # 상태별 구분 신호를 약하게만 추가
    if state_id == STATE2ID["sleep_deprived_slow"]:
        response_time = clamp01(response_time + 0.03)
        sleep_score = clamp01(sleep_score - 0.04)
        fragmentation = clamp01(fragmentation + 0.03)

    elif state_id == STATE2ID["high_noise_overloaded"]:
        noise_load = clamp01(noise_load + 0.04)
        typing_var = clamp01(typing_var + 0.02)
        focus_drift = clamp01(focus_drift + 0.03)

    elif state_id == STATE2ID["emotionally_reactive"]:
        affect_intensity = clamp01(affect_intensity + 0.04)
        typing_var = clamp01(typing_var + 0.02)

    obs = np.array(
        [
            hour_norm,
            response_time,
            typing_var,
            sleep_score,
            noise_load,
            affect_intensity,
            engagement,
            adherence,
            fragmentation,
            latency_instability,
            focus_drift,
        ],
        dtype=np.float32,
    )

    # partial observability: 일부 핵심 관측 마스킹
    # 3:sleep_score, 4:noise_load, 5:affect_intensity, 8,9,10
    candidate_idx = [3, 4, 5, 8, 9, 10]
    for idx in candidate_idx:
        if np.random.rand() < mask_prob:
            obs[idx] = 0.5  # missing-like neutral token

    return obs


def choose_behavior_action(state_id: int) -> int:
    """
    behavior policy를 완전 정답으로 고정하지 않음.
    """
    r = np.random.rand()
    best = best_action_for_state(state_id)
    secondary = list(secondary_good_actions(state_id))
    all_actions = list(range(len(ACTION_NAMES)))

    if r < 0.58:
        return best
    if r < 0.83 and len(secondary) > 0:
        return random.choice(secondary)
    bads = [a for a in all_actions if a != best and a not in secondary]
    return random.choice(bads)


def generate_sequence(seq_len: int, domain_shift: float = 0.0):
    state = np.random.choice(4, p=[0.28, 0.27, 0.23, 0.22])
    start_hour = np.random.randint(0, 24)

    obs_seq = []
    action_seq = []
    reward_seq = []
    state_seq = []
    next_state_seq = []

    for t in range(seq_len):
        hour = (start_hour + t * np.random.randint(1, 4)) % 24
        obs = generate_observation(
            state_id=state,
            hour=hour,
            domain_shift=domain_shift,
            mask_prob=0.18 if t < seq_len - 1 else 0.10,
        )

        action = choose_behavior_action(state)
        reward = reward_from_state_action(state, action)
        next_state = transition_next_state(state, action)

        obs_seq.append(obs)
        action_seq.append(action)
        reward_seq.append(reward)
        state_seq.append(state)
        next_state_seq.append(next_state)

        state = next_state

    return {
        "obs": np.stack(obs_seq),
        "actions": np.array(action_seq, dtype=np.int64),
        "rewards": np.array(reward_seq, dtype=np.float32),
        "states": np.array(state_seq, dtype=np.int64),
        "next_states": np.array(next_state_seq, dtype=np.int64),
    }


# ============================================================
# Dataset
# ============================================================
class SleepSequenceDataset(Dataset):
    def __init__(self, n_samples: int, seq_len: int, domain_shift: float):
        self.samples = [generate_sequence(seq_len, domain_shift=domain_shift) for _ in range(n_samples)]
        self.labels = [int(x["states"][-1]) for x in self.samples]

        counts = np.bincount(self.labels, minlength=len(STATE_NAMES))
        inv = 1.0 / np.maximum(counts, 1)
        self.sample_weights = [inv[y] for y in self.labels]

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):
        x = self.samples[idx]
        return {
            "obs": torch.tensor(x["obs"], dtype=torch.float32),
            "actions": torch.tensor(x["actions"], dtype=torch.long),
            "rewards": torch.tensor(x["rewards"], dtype=torch.float32),
            "states": torch.tensor(x["states"], dtype=torch.long),
            "next_states": torch.tensor(x["next_states"], dtype=torch.long),
        }


def build_loader(dataset: SleepSequenceDataset, batch_size: int, balanced: bool):
    if balanced:
        sampler = WeightedRandomSampler(
            weights=dataset.sample_weights,
            num_samples=len(dataset),
            replacement=True,
        )
        return DataLoader(dataset, batch_size=batch_size, sampler=sampler, drop_last=False)
    return DataLoader(dataset, batch_size=batch_size, shuffle=False, drop_last=False)


# ============================================================
# Model
# ============================================================
class SleepBeliefPolicyModel(nn.Module):
    def __init__(self, cfg: Config):
        super().__init__()

        self.obs_encoder = nn.Sequential(
            nn.Linear(cfg.obs_dim, cfg.hidden_dim),
            nn.LayerNorm(cfg.hidden_dim),
            nn.GELU(),
            nn.Dropout(cfg.dropout),
            nn.Linear(cfg.hidden_dim, cfg.hidden_dim),
            nn.LayerNorm(cfg.hidden_dim),
            nn.GELU(),
        )

        self.action_emb = nn.Embedding(len(ACTION_NAMES), cfg.hidden_dim // 4)
        self.reward_proj = nn.Linear(1, cfg.hidden_dim // 8)

        rnn_in_dim = cfg.hidden_dim + cfg.hidden_dim // 4 + cfg.hidden_dim // 8

        self.gru = nn.GRU(
            input_size=rnn_in_dim,
            hidden_size=cfg.hidden_dim,
            num_layers=cfg.num_layers,
            batch_first=True,
            dropout=cfg.dropout if cfg.num_layers > 1 else 0.0,
        )

        self.shared = nn.Sequential(
            nn.Linear(cfg.hidden_dim, cfg.hidden_dim),
            nn.LayerNorm(cfg.hidden_dim),
            nn.GELU(),
            nn.Dropout(cfg.dropout),
        )

        self.state_head = nn.Linear(cfg.hidden_dim, len(STATE_NAMES))
        self.action_head = nn.Linear(cfg.hidden_dim, len(ACTION_NAMES))
        self.reward_head = nn.Linear(cfg.hidden_dim, 1)
        self.next_state_head = nn.Linear(cfg.hidden_dim, len(STATE_NAMES))

    def forward(self, obs, prev_actions, prev_rewards):
        obs_feat = self.obs_encoder(obs)
        act_feat = self.action_emb(prev_actions)
        rew_feat = self.reward_proj(prev_rewards.unsqueeze(-1))

        x = torch.cat([obs_feat, act_feat, rew_feat], dim=-1)
        h, _ = self.gru(x)
        z = self.shared(h)

        return {
            "state_logits": self.state_head(z),
            "action_logits": self.action_head(z),
            "reward_pred": self.reward_head(z).squeeze(-1),
            "next_state_logits": self.next_state_head(z),
        }


# ============================================================
# Metrics
# ============================================================
def compute_class_weights(dataset, device):
    counts = np.bincount(dataset.labels, minlength=len(STATE_NAMES)).astype(np.float32)
    w = counts.sum() / np.maximum(counts, 1.0)
    w = w / w.mean()
    return torch.tensor(w, dtype=torch.float32, device=device)


def confusion_matrix(y_true, y_pred, n_cls):
    mat = np.zeros((n_cls, n_cls), dtype=np.int64)
    for t, p in zip(y_true, y_pred):
        mat[t, p] += 1
    return mat


def print_confusion_matrix(cm):
    print("\n[Confusion Matrix] rows=true cols=pred")
    print(" " * 24 + " ".join([f"{s[:10]:>12}" for s in STATE_NAMES]))
    for i, row in enumerate(cm):
        print(f"{STATE_NAMES[i]:>24} " + " ".join([f"{int(v):12d}" for v in row]))


# ============================================================
# Train / Eval
# ============================================================
def train_one_epoch(model, loader, optimizer, class_weights, device, cfg):
    model.train()
    total_loss = 0.0

    for batch in loader:
        obs = batch["obs"].to(device)
        actions = batch["actions"].to(device)
        rewards = batch["rewards"].to(device)
        states = batch["states"].to(device)
        next_states = batch["next_states"].to(device)

        prev_actions = torch.roll(actions, shifts=1, dims=1)
        prev_rewards = torch.roll(rewards, shifts=1, dims=1)
        prev_actions[:, 0] = 0
        prev_rewards[:, 0] = 0.0

        out = model(obs, prev_actions, prev_rewards)

        B, T, S = out["state_logits"].shape
        A = out["action_logits"].shape[-1]

        state_logits = out["state_logits"].reshape(B * T, S)
        action_logits = out["action_logits"].reshape(B * T, A)
        next_state_logits = out["next_state_logits"].reshape(B * T, S)
        reward_pred = out["reward_pred"].reshape(B * T)

        state_targets = states.reshape(B * T)
        action_targets = actions.reshape(B * T)
        next_state_targets = next_states.reshape(B * T)
        reward_targets = rewards.reshape(B * T)

        smoothed = make_smoothed_targets(state_targets, S, cfg.label_smoothing)
        next_smoothed = make_smoothed_targets(next_state_targets, S, cfg.label_smoothing)

        # class weight를 soft CE에 직접 곱하기 위해 sample-wise 가중 적용
        log_probs_state = F.log_softmax(state_logits, dim=-1)
        state_loss_vec = -(smoothed * log_probs_state).sum(dim=-1) * class_weights[state_targets]
        state_loss = state_loss_vec.mean()

        log_probs_next = F.log_softmax(next_state_logits, dim=-1)
        transition_loss_vec = -(next_smoothed * log_probs_next).sum(dim=-1) * class_weights[next_state_targets]
        transition_loss = transition_loss_vec.mean()

        action_loss = F.cross_entropy(action_logits, action_targets)
        reward_loss = F.smooth_l1_loss(reward_pred, reward_targets)
        ent = entropy_from_logits(state_logits)

        loss = (
            state_loss
            + cfg.action_coef * action_loss
            + cfg.reward_coef * reward_loss
            + cfg.transition_coef * transition_loss
            - cfg.entropy_coef * ent
        )

        optimizer.zero_grad(set_to_none=True)
        loss.backward()
        nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        optimizer.step()

        total_loss += float(loss.item())

    return total_loss / max(1, len(loader))


@torch.no_grad()
def evaluate(model, loader, class_weights, device, cfg):
    model.eval()

    total_loss = 0.0
    y_true = []
    y_pred = []
    a_true = []
    a_pred = []

    for batch in loader:
        obs = batch["obs"].to(device)
        actions = batch["actions"].to(device)
        rewards = batch["rewards"].to(device)
        states = batch["states"].to(device)
        next_states = batch["next_states"].to(device)

        prev_actions = torch.roll(actions, shifts=1, dims=1)
        prev_rewards = torch.roll(rewards, shifts=1, dims=1)
        prev_actions[:, 0] = 0
        prev_rewards[:, 0] = 0.0

        out = model(obs, prev_actions, prev_rewards)

        B, T, S = out["state_logits"].shape
        A = out["action_logits"].shape[-1]

        state_logits = out["state_logits"].reshape(B * T, S)
        action_logits = out["action_logits"].reshape(B * T, A)
        next_state_logits = out["next_state_logits"].reshape(B * T, S)
        reward_pred = out["reward_pred"].reshape(B * T)

        state_targets = states.reshape(B * T)
        action_targets = actions.reshape(B * T)
        next_state_targets = next_states.reshape(B * T)
        reward_targets = rewards.reshape(B * T)

        smoothed = make_smoothed_targets(state_targets, S, cfg.label_smoothing)
        next_smoothed = make_smoothed_targets(next_state_targets, S, cfg.label_smoothing)

        log_probs_state = F.log_softmax(state_logits, dim=-1)
        state_loss_vec = -(smoothed * log_probs_state).sum(dim=-1) * class_weights[state_targets]
        state_loss = state_loss_vec.mean()

        log_probs_next = F.log_softmax(next_state_logits, dim=-1)
        transition_loss_vec = -(next_smoothed * log_probs_next).sum(dim=-1) * class_weights[next_state_targets]
        transition_loss = transition_loss_vec.mean()

        action_loss = F.cross_entropy(action_logits, action_targets)
        reward_loss = F.smooth_l1_loss(reward_pred, reward_targets)
        ent = entropy_from_logits(state_logits)

        loss = (
            state_loss
            + cfg.action_coef * action_loss
            + cfg.reward_coef * reward_loss
            + cfg.transition_coef * transition_loss
            - cfg.entropy_coef * ent
        )
        total_loss += float(loss.item())

        last_state_pred = out["state_logits"][:, -1].argmax(dim=-1).cpu().numpy()
        last_state_true = states[:, -1].cpu().numpy()

        last_action_pred = out["action_logits"][:, -1].argmax(dim=-1).cpu().numpy()
        last_action_true = actions[:, -1].cpu().numpy()

        y_true.append(last_state_true)
        y_pred.append(last_state_pred)
        a_true.append(last_action_true)
        a_pred.append(last_action_pred)

    y_true = np.concatenate(y_true)
    y_pred = np.concatenate(y_pred)
    a_true = np.concatenate(a_true)
    a_pred = np.concatenate(a_pred)

    state_acc = float((y_true == y_pred).mean())
    action_acc = float((a_true == a_pred).mean())

    recalls = []
    for c in range(len(STATE_NAMES)):
        idx = (y_true == c)
        recalls.append(float((y_pred[idx] == c).mean()) if idx.sum() > 0 else 0.0)
    macro_recall = float(np.mean(recalls))

    cm = confusion_matrix(y_true, y_pred, len(STATE_NAMES))

    return {
        "loss": total_loss / max(1, len(loader)),
        "state_acc": state_acc,
        "action_acc": action_acc,
        "macro_recall": macro_recall,
        "cm": cm,
    }


# ============================================================
# Profile evaluation
# ============================================================
@torch.no_grad()
def profile_eval(model, device):
    model.eval()

    print("\nPolicy evaluation on multiple user profiles")
    print("-" * 90)

    profiles = [
        ("profile_1", STATE2ID["rested_focused"]),
        ("profile_2", STATE2ID["sleep_deprived_slow"]),
        ("profile_3", STATE2ID["emotionally_reactive"]),
        ("profile_4", STATE2ID["high_noise_overloaded"]),
    ]

    for profile_name, init_state in profiles:
        state = init_state
        start_hour = np.random.randint(0, 24)

        obs_seq = []
        action_history = []
        reward_history = []
        reward_list = []
        inferred = None

        for t in range(12):
            hour = (start_hour + t) % 24
            obs = generate_observation(
                state_id=state,
                hour=hour,
                domain_shift=0.04,
                mask_prob=0.15,
            )
            obs_seq.append(obs)

            # 길이를 반드시 obs_seq와 동일하게 유지
            prev_actions = [0] + action_history
            prev_rewards = [0.0] + reward_history

            prev_actions = prev_actions[: len(obs_seq)]
            prev_rewards = prev_rewards[: len(obs_seq)]

            while len(prev_actions) < len(obs_seq):
                prev_actions.append(0)
            while len(prev_rewards) < len(obs_seq):
                prev_rewards.append(0.0)

            obs_t = torch.tensor(np.array(obs_seq), dtype=torch.float32).unsqueeze(0).to(device)
            act_t = torch.tensor(np.array(prev_actions), dtype=torch.long).unsqueeze(0).to(device)
            rew_t = torch.tensor(np.array(prev_rewards), dtype=torch.float32).unsqueeze(0).to(device)

            out = model(obs_t, act_t, rew_t)
            inferred = out["state_logits"][:, -1].argmax(dim=-1).item()
            action = out["action_logits"][:, -1].argmax(dim=-1).item()

            reward = reward_from_state_action(state, action)
            reward_list.append(reward)

            action_history.append(action)
            reward_history.append(reward)

            state = transition_next_state(state, action)

        print(
            f"{profile_name}: true={STATE_NAMES[init_state]:<22} | "
            f"avg_reward={np.mean(reward_list):.3f} | "
            f"last_inferred={STATE_NAMES[inferred]}"
        )


# ============================================================
# Main
# ============================================================
def main():
    args = parse_args()
    set_seed(args.seed)

    device = torch.device(args.device if torch.cuda.is_available() else "cpu")

    cfg = Config(
        epochs=args.epochs,
        batch_size=args.batch_size,
        lr=args.lr,
        train_samples=args.train_samples,
        val_samples=args.val_samples,
        test_samples=args.test_samples,
        device=str(device),
        seed=args.seed,
    )

    print("=" * 80)
    print("Hard Sleep Research Training")
    print("=" * 80)
    print(f"device       = {device}")
    print(f"epochs       = {cfg.epochs}")
    print(f"batch_size   = {cfg.batch_size}")
    print(f"lr           = {cfg.lr}")
    print(f"train/val/test = {cfg.train_samples}/{cfg.val_samples}/{cfg.test_samples}")
    print("=" * 80)

    train_ds = SleepSequenceDataset(cfg.train_samples, cfg.seq_len, domain_shift=cfg.train_shift)
    val_ds = SleepSequenceDataset(cfg.val_samples, cfg.seq_len, domain_shift=cfg.val_shift)
    test_ds = SleepSequenceDataset(cfg.test_samples, cfg.seq_len, domain_shift=cfg.test_shift)

    train_loader = build_loader(train_ds, cfg.batch_size, balanced=True)
    val_loader = build_loader(val_ds, cfg.batch_size, balanced=False)
    test_loader = build_loader(test_ds, cfg.batch_size, balanced=False)

    class_weights = compute_class_weights(train_ds, device)

    model = SleepBeliefPolicyModel(cfg).to(device)
    optimizer = torch.optim.AdamW(model.parameters(), lr=cfg.lr, weight_decay=cfg.weight_decay)
    scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=cfg.epochs)

    best_score = -1.0

    for epoch in range(1, cfg.epochs + 1):
        train_loss = train_one_epoch(model, train_loader, optimizer, class_weights, device, cfg)
        val_stats = evaluate(model, val_loader, class_weights, device, cfg)
        scheduler.step()

        score = 0.7 * val_stats["macro_recall"] + 0.3 * val_stats["state_acc"]
        if score > best_score:
            best_score = score
            torch.save(
                {
                    "model_state_dict": model.state_dict(),
                    "best_score": best_score,
                    "config": cfg.__dict__,
                },
                args.save_path,
            )

        print(
            f"[epoch {epoch:02d}] "
            f"train_loss={train_loss:.4f} | "
            f"val_loss={val_stats['loss']:.4f} | "
            f"state_acc={val_stats['state_acc']:.4f} | "
            f"macro_recall={val_stats['macro_recall']:.4f} | "
            f"action_acc={val_stats['action_acc']:.4f}"
        )

    print("\nLoading best model...")
    ckpt = torch.load(args.save_path, map_location=device)
    model.load_state_dict(ckpt["model_state_dict"])

    test_stats = evaluate(model, test_loader, class_weights, device, cfg)

    print("\n" + "=" * 80)
    print("FINAL TEST RESULT")
    print("=" * 80)
    print(f"state_acc    : {test_stats['state_acc']:.4f}")
    print(f"macro_recall : {test_stats['macro_recall']:.4f}")
    print(f"action_acc   : {test_stats['action_acc']:.4f}")
    print(f"test_loss    : {test_stats['loss']:.4f}")
    print("=" * 80)

    print_confusion_matrix(test_stats["cm"])
    profile_eval(model, device)


if __name__ == "__main__":
    main()
```

```python
# ============================================================
# sleep_research_improved_closed_loop.py
# ============================================================
from __future__ import annotations

import argparse
import random
from dataclasses import dataclass

import numpy as np
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader, WeightedRandomSampler


# ============================================================
# Jupyter-safe argparse
# ============================================================
def parse_args():
    parser = argparse.ArgumentParser()
    parser.add_argument("--epochs", type=int, default=32)
    parser.add_argument("--batch_size", type=int, default=128)
    parser.add_argument("--lr", type=float, default=3e-4)
    parser.add_argument("--device", type=str, default="cuda")
    parser.add_argument("--seed", type=int, default=42)
    parser.add_argument("--save_path", type=str, default="./best_sleep_research_improved.pt")
    parser.add_argument("--train_samples", type=int, default=18000)
    parser.add_argument("--val_samples", type=int, default=3000)
    parser.add_argument("--test_samples", type=int, default=3000)
    args, _ = parser.parse_known_args()
    return args


# ============================================================
# Reproducibility
# ============================================================
def set_seed(seed: int = 42):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)


# ============================================================
# Definitions
# ============================================================
STATE_NAMES = [
    "rested_focused",
    "sleep_deprived_slow",
    "high_noise_overloaded",
    "emotionally_reactive",
]
STATE2ID = {s: i for i, s in enumerate(STATE_NAMES)}

ACTION_NAMES = [
    "normal_guidance",
    "low_complexity_checkin",
    "step_by_step_support",
    "emotional_validation",
    "noise_reduction_prompt",
    "sleep_hygiene_prompt",
]
ACTION2ID = {a: i for i, a in enumerate(ACTION_NAMES)}


@dataclass
class Config:
    seq_len: int = 14
    obs_dim: int = 12   # + missingness ratio
    hidden_dim: int = 192
    num_layers: int = 1
    dropout: float = 0.15

    train_samples: int = 18000
    val_samples: int = 3000
    test_samples: int = 3000

    epochs: int = 32
    batch_size: int = 128
    lr: float = 3e-4
    weight_decay: float = 1e-4

    state_coef: float = 1.0
    action_coef: float = 0.55
    reward_coef: float = 0.18
    transition_coef: float = 0.35
    entropy_coef: float = 0.0015

    label_smoothing: float = 0.05
    train_shift: float = 0.00
    val_shift: float = 0.04
    test_shift: float = 0.07

    mask_prob_train: float = 0.24
    mask_prob_eval: float = 0.20

    device: str = "cuda"
    seed: int = 42


# ============================================================
# Utilities
# ============================================================
def clamp01(x: float) -> float:
    return max(0.0, min(1.0, x))


def entropy_from_logits(logits: torch.Tensor) -> torch.Tensor:
    p = F.softmax(logits, dim=-1)
    logp = F.log_softmax(logits, dim=-1)
    return -(p * logp).sum(dim=-1).mean()


def smoothed_onehot(targets: torch.Tensor, num_classes: int, smoothing: float) -> torch.Tensor:
    with torch.no_grad():
        dist = torch.full(
            (targets.shape[0], num_classes),
            fill_value=smoothing / (num_classes - 1),
            device=targets.device,
            dtype=torch.float32,
        )
        dist.scatter_(1, targets.unsqueeze(1), 1.0 - smoothing)
    return dist


def confusion_matrix(y_true, y_pred, n_cls):
    mat = np.zeros((n_cls, n_cls), dtype=np.int64)
    for t, p in zip(y_true, y_pred):
        mat[t, p] += 1
    return mat


def print_confusion_matrix(cm):
    print("\n[Confusion Matrix] rows=true cols=pred")
    print(" " * 24 + " ".join([f"{s[:10]:>12}" for s in STATE_NAMES]))
    for i, row in enumerate(cm):
        print(f"{STATE_NAMES[i]:>24} " + " ".join([f"{int(v):12d}" for v in row]))


# ============================================================
# Environment design
# ============================================================
def state_proto(state_id: int):
    # 분리를 너무 강하게 하지 않음
    if state_id == STATE2ID["rested_focused"]:
        return dict(
            response_time=0.38,
            typing_var=0.36,
            sleep_score=0.72,
            noise_load=0.33,
            affect_intensity=0.28,
            engagement=0.79,
            adherence=0.81,
            fragmentation=0.29,
            latency_instability=0.31,
            focus_drift=0.30,
        )
    if state_id == STATE2ID["sleep_deprived_slow"]:
        return dict(
            response_time=0.62,
            typing_var=0.44,
            sleep_score=0.38,
            noise_load=0.41,
            affect_intensity=0.36,
            engagement=0.66,
            adherence=0.68,
            fragmentation=0.57,
            latency_instability=0.61,
            focus_drift=0.58,
        )
    if state_id == STATE2ID["high_noise_overloaded"]:
        return dict(
            response_time=0.57,
            typing_var=0.57,
            sleep_score=0.44,
            noise_load=0.69,
            affect_intensity=0.42,
            engagement=0.61,
            adherence=0.60,
            fragmentation=0.50,
            latency_instability=0.55,
            focus_drift=0.63,
        )
    if state_id == STATE2ID["emotionally_reactive"]:
        return dict(
            response_time=0.54,
            typing_var=0.56,
            sleep_score=0.46,
            noise_load=0.39,
            affect_intensity=0.69,
            engagement=0.57,
            adherence=0.56,
            fragmentation=0.48,
            latency_instability=0.52,
            focus_drift=0.57,
        )
    raise ValueError("invalid state")


def best_action_for_state(state_id: int) -> int:
    if state_id == STATE2ID["rested_focused"]:
        return ACTION2ID["normal_guidance"]
    if state_id == STATE2ID["sleep_deprived_slow"]:
        return ACTION2ID["low_complexity_checkin"]
    if state_id == STATE2ID["high_noise_overloaded"]:
        return ACTION2ID["noise_reduction_prompt"]
    if state_id == STATE2ID["emotionally_reactive"]:
        return ACTION2ID["emotional_validation"]
    raise ValueError("invalid state")


def secondary_good_actions(state_id: int):
    if state_id == STATE2ID["rested_focused"]:
        return {ACTION2ID["step_by_step_support"]}
    if state_id == STATE2ID["sleep_deprived_slow"]:
        return {ACTION2ID["step_by_step_support"], ACTION2ID["sleep_hygiene_prompt"]}
    if state_id == STATE2ID["high_noise_overloaded"]:
        return {ACTION2ID["low_complexity_checkin"], ACTION2ID["step_by_step_support"]}
    if state_id == STATE2ID["emotionally_reactive"]:
        return {ACTION2ID["low_complexity_checkin"], ACTION2ID["step_by_step_support"]}
    return set()


def target_action_distribution_for_state(state_id: int, device=None) -> torch.Tensor:
    dist = torch.full((len(ACTION_NAMES),), 0.02, dtype=torch.float32, device=device)
    best = best_action_for_state(state_id)
    dist[best] = 0.76
    for a in secondary_good_actions(state_id):
        dist[a] += 0.10
    dist = dist / dist.sum()
    return dist


def reward_from_state_action(state_id: int, action_id: int) -> float:
    best = best_action_for_state(state_id)
    if action_id == best:
        return np.random.uniform(0.82, 0.94)
    if action_id in secondary_good_actions(state_id):
        return np.random.uniform(0.60, 0.80)
    return np.random.uniform(0.28, 0.58)


def transition_next_state(curr_state: int, action_id: int) -> int:
    # attractor를 약화: correct action이어도 무조건 rested로 가지 않음
    p = np.random.rand()
    best = best_action_for_state(curr_state)

    if curr_state == STATE2ID["rested_focused"]:
        if action_id == best:
            return curr_state
        if action_id == ACTION2ID["sleep_hygiene_prompt"] and p < 0.08:
            return STATE2ID["sleep_deprived_slow"]
        if action_id == ACTION2ID["noise_reduction_prompt"] and p < 0.07:
            return STATE2ID["high_noise_overloaded"]
        if action_id == ACTION2ID["emotional_validation"] and p < 0.05:
            return STATE2ID["emotionally_reactive"]
        return curr_state

    if curr_state == STATE2ID["sleep_deprived_slow"]:
        if action_id == ACTION2ID["low_complexity_checkin"]:
            if p < 0.10:
                return STATE2ID["rested_focused"]
            if p < 0.24:
                return STATE2ID["sleep_deprived_slow"]
            return STATE2ID["high_noise_overloaded"]
        if action_id == ACTION2ID["sleep_hygiene_prompt"]:
            if p < 0.12:
                return STATE2ID["rested_focused"]
            return STATE2ID["sleep_deprived_slow"]
        if action_id == ACTION2ID["normal_guidance"] and p < 0.30:
            return STATE2ID["high_noise_overloaded"]
        if action_id == ACTION2ID["emotional_validation"] and p < 0.15:
            return STATE2ID["emotionally_reactive"]
        return curr_state

    if curr_state == STATE2ID["high_noise_overloaded"]:
        if action_id == ACTION2ID["noise_reduction_prompt"]:
            if p < 0.10:
                return STATE2ID["rested_focused"]
            if p < 0.28:
                return STATE2ID["high_noise_overloaded"]
            return STATE2ID["sleep_deprived_slow"]
        if action_id == ACTION2ID["normal_guidance"] and p < 0.28:
            return STATE2ID["emotionally_reactive"]
        if action_id == ACTION2ID["sleep_hygiene_prompt"] and p < 0.16:
            return STATE2ID["sleep_deprived_slow"]
        return curr_state

    if curr_state == STATE2ID["emotionally_reactive"]:
        if action_id == ACTION2ID["emotional_validation"]:
            if p < 0.10:
                return STATE2ID["rested_focused"]
            if p < 0.27:
                return STATE2ID["emotionally_reactive"]
            return STATE2ID["high_noise_overloaded"]
        if action_id == ACTION2ID["normal_guidance"] and p < 0.28:
            return STATE2ID["high_noise_overloaded"]
        if action_id == ACTION2ID["sleep_hygiene_prompt"] and p < 0.14:
            return STATE2ID["sleep_deprived_slow"]
        return curr_state

    return curr_state


def choose_behavior_action(state_id: int) -> int:
    # behavior policy는 noisy 하지만 너무 엉망은 아니게
    r = np.random.rand()
    best = best_action_for_state(state_id)
    secondary = list(secondary_good_actions(state_id))
    all_actions = list(range(len(ACTION_NAMES)))

    if r < 0.48:
        return best
    if r < 0.78 and len(secondary) > 0:
        return random.choice(secondary)
    bads = [a for a in all_actions if a != best and a not in secondary]
    return random.choice(bads)


def generate_observation(state_id: int, hour: int, domain_shift: float = 0.0, mask_prob: float = 0.24) -> np.ndarray:
    proto = state_proto(state_id)
    hour_norm = hour / 23.0

    circadian_sleep = 0.0
    circadian_focus = 0.0
    if 0 <= hour <= 5:
        circadian_sleep = 0.14
        circadian_focus = 0.10
    elif 6 <= hour <= 9:
        circadian_sleep = -0.05
        circadian_focus = -0.02
    elif 22 <= hour <= 23:
        circadian_sleep = 0.08
        circadian_focus = 0.06

    shift = domain_shift
    sigma = 0.13

    response_time = clamp01(np.random.normal(proto["response_time"] + 0.03 * shift, sigma))
    typing_var = clamp01(np.random.normal(proto["typing_var"] + 0.02 * shift, sigma))
    sleep_score = clamp01(np.random.normal(proto["sleep_score"] - circadian_sleep - 0.03 * shift, sigma))
    noise_load = clamp01(np.random.normal(proto["noise_load"] + 0.04 * shift, sigma))
    affect_intensity = clamp01(np.random.normal(proto["affect_intensity"] + 0.03 * shift, sigma))
    engagement = clamp01(np.random.normal(proto["engagement"] - 0.03 * shift, sigma))
    adherence = clamp01(np.random.normal(proto["adherence"] - 0.02 * shift, sigma))
    fragmentation = clamp01(np.random.normal(proto["fragmentation"] + 0.03 * shift, sigma))
    latency_instability = clamp01(np.random.normal(proto["latency_instability"] + circadian_focus, sigma))
    focus_drift = clamp01(np.random.normal(proto["focus_drift"] + circadian_focus, sigma))

    # state-specific cue는 약하게만
    if state_id == STATE2ID["sleep_deprived_slow"]:
        response_time = clamp01(response_time + 0.02)
        sleep_score = clamp01(sleep_score - 0.03)
    elif state_id == STATE2ID["high_noise_overloaded"]:
        noise_load = clamp01(noise_load + 0.03)
        focus_drift = clamp01(focus_drift + 0.02)
    elif state_id == STATE2ID["emotionally_reactive"]:
        affect_intensity = clamp01(affect_intensity + 0.03)
        typing_var = clamp01(typing_var + 0.02)

    obs = np.array([
        hour_norm,
        response_time,
        typing_var,
        sleep_score,
        noise_load,
        affect_intensity,
        engagement,
        adherence,
        fragmentation,
        latency_instability,
        focus_drift,
    ], dtype=np.float32)

    candidate_idx = [3, 4, 5, 8, 9, 10]
    missing_count = 0
    for idx in candidate_idx:
        if np.random.rand() < mask_prob:
            obs[idx] = 0.5
            missing_count += 1

    missing_ratio = np.array([missing_count / len(candidate_idx)], dtype=np.float32)
    obs = np.concatenate([obs, missing_ratio], axis=0)
    return obs


def generate_sequence(seq_len: int, domain_shift: float = 0.0, mask_prob: float = 0.24):
    state = np.random.choice(4, p=[0.28, 0.27, 0.23, 0.22])
    start_hour = np.random.randint(0, 24)

    obs_seq, action_seq, reward_seq = [], [], []
    state_seq, next_state_seq = [], []
    optimal_action_seq = []

    for t in range(seq_len):
        hour = (start_hour + t * np.random.randint(1, 4)) % 24
        obs = generate_observation(state, hour, domain_shift=domain_shift, mask_prob=mask_prob)

        behavior_action = choose_behavior_action(state)
        optimal_action = best_action_for_state(state)
        reward = reward_from_state_action(state, behavior_action)
        next_state = transition_next_state(state, behavior_action)

        obs_seq.append(obs)
        action_seq.append(behavior_action)
        optimal_action_seq.append(optimal_action)
        reward_seq.append(reward)
        state_seq.append(state)
        next_state_seq.append(next_state)

        state = next_state

    return {
        "obs": np.stack(obs_seq),
        "actions": np.array(action_seq, dtype=np.int64),
        "optimal_actions": np.array(optimal_action_seq, dtype=np.int64),
        "rewards": np.array(reward_seq, dtype=np.float32),
        "states": np.array(state_seq, dtype=np.int64),
        "next_states": np.array(next_state_seq, dtype=np.int64),
    }


# ============================================================
# Dataset
# ============================================================
class SleepSequenceDataset(Dataset):
    def __init__(self, n_samples: int, seq_len: int, domain_shift: float, mask_prob: float):
        self.samples = [generate_sequence(seq_len, domain_shift=domain_shift, mask_prob=mask_prob) for _ in range(n_samples)]
        self.labels = [int(x["states"][-1]) for x in self.samples]
        counts = np.bincount(self.labels, minlength=len(STATE_NAMES))
        inv = 1.0 / np.maximum(counts, 1)
        self.sample_weights = [inv[y] for y in self.labels]

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):
        x = self.samples[idx]
        return {
            "obs": torch.tensor(x["obs"], dtype=torch.float32),
            "actions": torch.tensor(x["actions"], dtype=torch.long),
            "optimal_actions": torch.tensor(x["optimal_actions"], dtype=torch.long),
            "rewards": torch.tensor(x["rewards"], dtype=torch.float32),
            "states": torch.tensor(x["states"], dtype=torch.long),
            "next_states": torch.tensor(x["next_states"], dtype=torch.long),
        }


def build_loader(dataset: SleepSequenceDataset, batch_size: int, balanced: bool):
    if balanced:
        sampler = WeightedRandomSampler(
            weights=dataset.sample_weights,
            num_samples=len(dataset),
            replacement=True,
        )
        return DataLoader(dataset, batch_size=batch_size, sampler=sampler, drop_last=False)
    return DataLoader(dataset, batch_size=batch_size, shuffle=False, drop_last=False)


# ============================================================
# Model
# ============================================================
class SleepBeliefPolicyModel(nn.Module):
    def __init__(self, cfg: Config):
        super().__init__()

        self.obs_encoder = nn.Sequential(
            nn.Linear(cfg.obs_dim, cfg.hidden_dim),
            nn.LayerNorm(cfg.hidden_dim),
            nn.GELU(),
            nn.Dropout(cfg.dropout),
            nn.Linear(cfg.hidden_dim, cfg.hidden_dim),
            nn.LayerNorm(cfg.hidden_dim),
            nn.GELU(),
        )

        self.action_emb = nn.Embedding(len(ACTION_NAMES), cfg.hidden_dim // 4)
        self.reward_proj = nn.Linear(1, cfg.hidden_dim // 8)
        rnn_in_dim = cfg.hidden_dim + cfg.hidden_dim // 4 + cfg.hidden_dim // 8

        self.gru = nn.GRU(
            input_size=rnn_in_dim,
            hidden_size=cfg.hidden_dim,
            num_layers=cfg.num_layers,
            batch_first=True,
            dropout=cfg.dropout if cfg.num_layers > 1 else 0.0,
        )

        self.shared = nn.Sequential(
            nn.Linear(cfg.hidden_dim, cfg.hidden_dim),
            nn.LayerNorm(cfg.hidden_dim),
            nn.GELU(),
            nn.Dropout(cfg.dropout),
        )

        self.state_head = nn.Linear(cfg.hidden_dim, len(STATE_NAMES))
        self.action_head = nn.Linear(cfg.hidden_dim, len(ACTION_NAMES))
        self.reward_head = nn.Linear(cfg.hidden_dim, 1)
        self.next_state_head = nn.Linear(cfg.hidden_dim, len(STATE_NAMES))

    def forward(self, obs, prev_actions, prev_rewards):
        obs_feat = self.obs_encoder(obs)
        act_feat = self.action_emb(prev_actions)
        rew_feat = self.reward_proj(prev_rewards.unsqueeze(-1))
        x = torch.cat([obs_feat, act_feat, rew_feat], dim=-1)
        h, _ = self.gru(x)
        z = self.shared(h)
        return {
            "state_logits": self.state_head(z),
            "action_logits": self.action_head(z),
            "reward_pred": self.reward_head(z).squeeze(-1),
            "next_state_logits": self.next_state_head(z),
        }


# ============================================================
# Loss helpers
# ============================================================
def compute_class_weights(dataset, device):
    counts = np.bincount(dataset.labels, minlength=len(STATE_NAMES)).astype(np.float32)
    w = counts.sum() / np.maximum(counts, 1.0)
    w = w / w.mean()
    return torch.tensor(w, dtype=torch.float32, device=device)


def make_state_targets_with_overlap(targets: torch.Tensor, smoothing: float) -> torch.Tensor:
    # 혼동이 자연스러운 쌍에 soft overlap 추가
    dist = smoothed_onehot(targets, len(STATE_NAMES), smoothing)

    for i, t in enumerate(targets.tolist()):
        if t == STATE2ID["sleep_deprived_slow"]:
            dist[i, STATE2ID["high_noise_overloaded"]] += 0.06
            dist[i, STATE2ID["emotionally_reactive"]] += 0.02
            dist[i, t] -= 0.08
        elif t == STATE2ID["high_noise_overloaded"]:
            dist[i, STATE2ID["emotionally_reactive"]] += 0.05
            dist[i, STATE2ID["sleep_deprived_slow"]] += 0.03
            dist[i, t] -= 0.08
        elif t == STATE2ID["emotionally_reactive"]:
            dist[i, STATE2ID["high_noise_overloaded"]] += 0.05
            dist[i, STATE2ID["sleep_deprived_slow"]] += 0.02
            dist[i, t] -= 0.07
    dist = dist / dist.sum(dim=-1, keepdim=True)
    return dist


def action_target_distribution(optimal_actions: torch.Tensor, state_targets: torch.Tensor, device) -> torch.Tensor:
    B = optimal_actions.shape[0]
    out = torch.zeros(B, len(ACTION_NAMES), dtype=torch.float32, device=device)
    for i in range(B):
        s = int(state_targets[i].item())
        out[i] = target_action_distribution_for_state(s, device=device)
    return out


# ============================================================
# Train / Eval
# ============================================================
def train_one_epoch(model, loader, optimizer, class_weights, device, cfg):
    model.train()
    total_loss = 0.0

    for batch in loader:
        obs = batch["obs"].to(device)
        actions = batch["actions"].to(device)
        optimal_actions = batch["optimal_actions"].to(device)
        rewards = batch["rewards"].to(device)
        states = batch["states"].to(device)
        next_states = batch["next_states"].to(device)

        prev_actions = torch.roll(actions, shifts=1, dims=1)
        prev_rewards = torch.roll(rewards, shifts=1, dims=1)
        prev_actions[:, 0] = 0
        prev_rewards[:, 0] = 0.0

        out = model(obs, prev_actions, prev_rewards)

        B, T, S = out["state_logits"].shape
        A = out["action_logits"].shape[-1]

        state_logits = out["state_logits"].reshape(B * T, S)
        action_logits = out["action_logits"].reshape(B * T, A)
        next_state_logits = out["next_state_logits"].reshape(B * T, S)
        reward_pred = out["reward_pred"].reshape(B * T)

        state_targets = states.reshape(B * T)
        next_state_targets = next_states.reshape(B * T)
        optimal_action_targets = optimal_actions.reshape(B * T)
        reward_targets = rewards.reshape(B * T)

        state_soft = make_state_targets_with_overlap(state_targets, cfg.label_smoothing)
        next_state_soft = make_state_targets_with_overlap(next_state_targets, cfg.label_smoothing)
        action_soft = action_target_distribution(optimal_action_targets, state_targets, device)

        log_probs_state = F.log_softmax(state_logits, dim=-1)
        state_loss = (-(state_soft * log_probs_state).sum(dim=-1) * class_weights[state_targets]).mean()

        log_probs_next = F.log_softmax(next_state_logits, dim=-1)
        transition_loss = (-(next_state_soft * log_probs_next).sum(dim=-1) * class_weights[next_state_targets]).mean()

        log_probs_action = F.log_softmax(action_logits, dim=-1)
        action_loss = -(action_soft * log_probs_action).sum(dim=-1).mean()

        reward_loss = F.smooth_l1_loss(reward_pred, reward_targets)
        ent = entropy_from_logits(state_logits)

        loss = (
            cfg.state_coef * state_loss
            + cfg.action_coef * action_loss
            + cfg.reward_coef * reward_loss
            + cfg.transition_coef * transition_loss
            - cfg.entropy_coef * ent
        )

        optimizer.zero_grad(set_to_none=True)
        loss.backward()
        nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        optimizer.step()

        total_loss += float(loss.item())

    return total_loss / max(1, len(loader))


@torch.no_grad()
def evaluate(model, loader, class_weights, device, cfg):
    model.eval()

    total_loss = 0.0
    y_true, y_pred = [], []
    a_true, a_pred = [], []

    for batch in loader:
        obs = batch["obs"].to(device)
        actions = batch["actions"].to(device)
        optimal_actions = batch["optimal_actions"].to(device)
        rewards = batch["rewards"].to(device)
        states = batch["states"].to(device)
        next_states = batch["next_states"].to(device)

        prev_actions = torch.roll(actions, shifts=1, dims=1)
        prev_rewards = torch.roll(rewards, shifts=1, dims=1)
        prev_actions[:, 0] = 0
        prev_rewards[:, 0] = 0.0

        out = model(obs, prev_actions, prev_rewards)

        B, T, S = out["state_logits"].shape
        A = out["action_logits"].shape[-1]

        state_logits = out["state_logits"].reshape(B * T, S)
        action_logits = out["action_logits"].reshape(B * T, A)
        next_state_logits = out["next_state_logits"].reshape(B * T, S)
        reward_pred = out["reward_pred"].reshape(B * T)

        state_targets = states.reshape(B * T)
        next_state_targets = next_states.reshape(B * T)
        optimal_action_targets = optimal_actions.reshape(B * T)
        reward_targets = rewards.reshape(B * T)

        state_soft = make_state_targets_with_overlap(state_targets, cfg.label_smoothing)
        next_state_soft = make_state_targets_with_overlap(next_state_targets, cfg.label_smoothing)
        action_soft = action_target_distribution(optimal_action_targets, state_targets, device)

        log_probs_state = F.log_softmax(state_logits, dim=-1)
        state_loss = (-(state_soft * log_probs_state).sum(dim=-1) * class_weights[state_targets]).mean()

        log_probs_next = F.log_softmax(next_state_logits, dim=-1)
        transition_loss = (-(next_state_soft * log_probs_next).sum(dim=-1) * class_weights[next_state_targets]).mean()

        log_probs_action = F.log_softmax(action_logits, dim=-1)
        action_loss = -(action_soft * log_probs_action).sum(dim=-1).mean()

        reward_loss = F.smooth_l1_loss(reward_pred, reward_targets)
        ent = entropy_from_logits(state_logits)

        loss = (
            cfg.state_coef * state_loss
            + cfg.action_coef * action_loss
            + cfg.reward_coef * reward_loss
            + cfg.transition_coef * transition_loss
            - cfg.entropy_coef * ent
        )
        total_loss += float(loss.item())

        last_state_pred = out["state_logits"][:, -1].argmax(dim=-1).cpu().numpy()
        last_state_true = states[:, -1].cpu().numpy()

        last_action_pred = out["action_logits"][:, -1].argmax(dim=-1).cpu().numpy()
        last_action_true = optimal_actions[:, -1].cpu().numpy()

        y_true.append(last_state_true)
        y_pred.append(last_state_pred)
        a_true.append(last_action_true)
        a_pred.append(last_action_pred)

    y_true = np.concatenate(y_true)
    y_pred = np.concatenate(y_pred)
    a_true = np.concatenate(a_true)
    a_pred = np.concatenate(a_pred)

    state_acc = float((y_true == y_pred).mean())
    action_acc = float((a_true == a_pred).mean())

    recalls = []
    for c in range(len(STATE_NAMES)):
        idx = (y_true == c)
        recalls.append(float((y_pred[idx] == c).mean()) if idx.sum() > 0 else 0.0)
    macro_recall = float(np.mean(recalls))

    cm = confusion_matrix(y_true, y_pred, len(STATE_NAMES))
    return {
        "loss": total_loss / max(1, len(loader)),
        "state_acc": state_acc,
        "action_acc": action_acc,
        "macro_recall": macro_recall,
        "cm": cm,
    }


# ============================================================
# Closed-loop profile evaluation
# ============================================================
@torch.no_grad()
def profile_eval(model, device, steps=12):
    model.eval()

    print("\nPolicy evaluation on multiple user profiles")
    print("-" * 108)

    profiles = [
        ("profile_1", STATE2ID["rested_focused"]),
        ("profile_2", STATE2ID["sleep_deprived_slow"]),
        ("profile_3", STATE2ID["emotionally_reactive"]),
        ("profile_4", STATE2ID["high_noise_overloaded"]),
    ]

    success_count = 0

    for profile_name, init_state in profiles:
        state = init_state
        start_hour = np.random.randint(0, 24)

        obs_seq = []
        action_hist = []
        reward_hist = []
        inferred_hist = []
        true_hist = []
        action_name_hist = []
        reward_list = []

        for t in range(steps):
            hour = (start_hour + t) % 24
            obs = generate_observation(
                state_id=state,
                hour=hour,
                domain_shift=0.06,
                mask_prob=0.18,
            )
            obs_seq.append(obs)

            prev_actions = [0] + action_hist
            prev_rewards = [0.0] + reward_hist
            prev_actions = prev_actions[:len(obs_seq)]
            prev_rewards = prev_rewards[:len(obs_seq)]
            while len(prev_actions) < len(obs_seq):
                prev_actions.append(0)
            while len(prev_rewards) < len(obs_seq):
                prev_rewards.append(0.0)

            obs_t = torch.tensor(np.array(obs_seq), dtype=torch.float32).unsqueeze(0).to(device)
            act_t = torch.tensor(np.array(prev_actions), dtype=torch.long).unsqueeze(0).to(device)
            rew_t = torch.tensor(np.array(prev_rewards), dtype=torch.float32).unsqueeze(0).to(device)

            out = model(obs_t, act_t, rew_t)
            inferred = out["state_logits"][:, -1].argmax(dim=-1).item()
            action = out["action_logits"][:, -1].argmax(dim=-1).item()
            reward = reward_from_state_action(state, action)

            true_hist.append(STATE_NAMES[state])
            inferred_hist.append(STATE_NAMES[inferred])
            action_name_hist.append(ACTION_NAMES[action])
            reward_list.append(reward)

            action_hist.append(action)
            reward_hist.append(reward)

            state = transition_next_state(state, action)

        final_true = true_hist[-1]
        final_inferred = inferred_hist[-1]
        recovered = (init_state != STATE2ID["rested_focused"] and final_true == "rested_focused")
        if recovered:
            success_count += 1

        print(f"{profile_name}: init={STATE_NAMES[init_state]:<22} | final_true={final_true:<22} | final_inferred={final_inferred:<22} | avg_reward={np.mean(reward_list):.3f}")
        print(f"  true_traj     : {' -> '.join(true_hist)}")
        print(f"  inferred_traj : {' -> '.join(inferred_hist)}")
        print(f"  action_traj   : {' -> '.join(action_name_hist)}")

    print(f"\n[closed-loop summary] non-rest profiles recovered_to_rested = {success_count}/3")


# ============================================================
# Main
# ============================================================
def main():
    args = parse_args()
    set_seed(args.seed)

    device = torch.device(args.device if torch.cuda.is_available() else "cpu")
    cfg = Config(
        epochs=args.epochs,
        batch_size=args.batch_size,
        lr=args.lr,
        train_samples=args.train_samples,
        val_samples=args.val_samples,
        test_samples=args.test_samples,
        device=str(device),
        seed=args.seed,
    )

    print("=" * 84)
    print("Improved Sleep Research Training")
    print("=" * 84)
    print(f"device         = {device}")
    print(f"epochs         = {cfg.epochs}")
    print(f"batch_size     = {cfg.batch_size}")
    print(f"lr             = {cfg.lr}")
    print(f"train/val/test = {cfg.train_samples}/{cfg.val_samples}/{cfg.test_samples}")
    print("=" * 84)

    train_ds = SleepSequenceDataset(cfg.train_samples, cfg.seq_len, cfg.train_shift, cfg.mask_prob_train)
    val_ds = SleepSequenceDataset(cfg.val_samples, cfg.seq_len, cfg.val_shift, cfg.mask_prob_eval)
    test_ds = SleepSequenceDataset(cfg.test_samples, cfg.seq_len, cfg.test_shift, cfg.mask_prob_eval)

    train_loader = build_loader(train_ds, cfg.batch_size, balanced=True)
    val_loader = build_loader(val_ds, cfg.batch_size, balanced=False)
    test_loader = build_loader(test_ds, cfg.batch_size, balanced=False)

    class_weights = compute_class_weights(train_ds, device)

    model = SleepBeliefPolicyModel(cfg).to(device)
    optimizer = torch.optim.AdamW(model.parameters(), lr=cfg.lr, weight_decay=cfg.weight_decay)
    scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=cfg.epochs)

    best_score = -1.0

    for epoch in range(1, cfg.epochs + 1):
        train_loss = train_one_epoch(model, train_loader, optimizer, class_weights, device, cfg)
        val_stats = evaluate(model, val_loader, class_weights, device, cfg)
        scheduler.step()

        score = 0.65 * val_stats["macro_recall"] + 0.20 * val_stats["state_acc"] + 0.15 * val_stats["action_acc"]
        if score > best_score:
            best_score = score
            torch.save(
                {
                    "model_state_dict": model.state_dict(),
                    "best_score": best_score,
                    "config": cfg.__dict__,
                },
                args.save_path,
            )

        print(
            f"[epoch {epoch:02d}] "
            f"train_loss={train_loss:.4f} | "
            f"val_loss={val_stats['loss']:.4f} | "
            f"state_acc={val_stats['state_acc']:.4f} | "
            f"macro_recall={val_stats['macro_recall']:.4f} | "
            f"action_acc={val_stats['action_acc']:.4f}"
        )

    print("\nLoading best model...")
    ckpt = torch.load(args.save_path, map_location=device)
    model.load_state_dict(ckpt["model_state_dict"])

    test_stats = evaluate(model, test_loader, class_weights, device, cfg)

    print("\n" + "=" * 84)
    print("FINAL TEST RESULT")
    print("=" * 84)
    print(f"state_acc    : {test_stats['state_acc']:.4f}")
    print(f"macro_recall : {test_stats['macro_recall']:.4f}")
    print(f"action_acc   : {test_stats['action_acc']:.4f}")
    print(f"test_loss    : {test_stats['loss']:.4f}")
    print("=" * 84)

    print_confusion_matrix(test_stats["cm"])
    profile_eval(model, device, steps=12)


if __name__ == "__main__":
    main()
```
