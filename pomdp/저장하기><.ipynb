```python
import zipfile
import os

files_to_zip = [
    "test_6.xlsx",
    "pomdp_results.xlsx",
]

zip_name = "results_bundle.zip"

with zipfile.ZipFile(zip_name, "w", zipfile.ZIP_DEFLATED) as zf:
    for file in files_to_zip:
        if os.path.exists(file):
            zf.write(file, arcname=os.path.basename(file))
        else:
            print(f"파일 없음: {file}")

print(f"압축 완료: {zip_name}")
```
