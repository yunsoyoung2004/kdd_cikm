```python
import os
import pandas as pd
import matplotlib.pyplot as plt

RESULT_DIR = "seoul_station_multimodal/results"

weight_path = f"{RESULT_DIR}/weight_sensitivity_results.csv"
out_path = f"{RESULT_DIR}/fig4_weight_sensitivity_distribution.png"

if not os.path.exists(weight_path):
    raise FileNotFoundError(f"Missing file: {weight_path}")

df = pd.read_csv(weight_path)

print("Loaded:", weight_path)
print("Columns:", list(df.columns))

# 네 기존 builder 기준 컬럼
corr_col = "spearman_with_gap"

if corr_col not in df.columns:
    raise ValueError(
        f"Missing column: {corr_col}\n"
        f"Available columns: {list(df.columns)}"
    )

plt.figure(figsize=(7, 5))

plt.hist(
    df[corr_col].dropna(),
    bins=20,
)

plt.xlabel("Spearman Correlation")
plt.ylabel("Count")
plt.title("Weight Sensitivity Distribution")

plt.tight_layout()
plt.savefig(out_path, dpi=300, bbox_inches="tight")
plt.close()

print("Saved:", out_path)
```

```python
# ============================================================
# COMPACT PAPER FIGURE: MLLM AUDIT MAIN RESULT
# English labels version
# ============================================================

from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import spearmanr

# ============================================================
# PATHS
# ============================================================

OUT_DIR = Path("results_mllm_audit_fast")
FIG_DIR = OUT_DIR / "paper_figures_compact"
FIG_DIR.mkdir(parents=True, exist_ok=True)

RES_PATH = OUT_DIR / "mllm_audit_outputs_flat_fixed_reference.csv"

if not RES_PATH.exists():
    RES_PATH = OUT_DIR / "mllm_audit_outputs_flat.csv"

# ============================================================
# LOAD RESULTS
# ============================================================

res = pd.read_csv(RES_PATH)

TARGET_COL = "mobility_minus_subway_gap"
REF_COL = "hidden_demand_reconstruction_score_norm"
FIXED_REF_COL = "hidden_demand_reconstruction_score_norm_fixed"

res["hidden_demand_score"] = pd.to_numeric(
    res["hidden_demand_score"],
    errors="coerce"
)

res[TARGET_COL] = pd.to_numeric(
    res[TARGET_COL],
    errors="coerce"
)

# reconstruction fixed direction
if FIXED_REF_COL not in res.columns:
    res[FIXED_REF_COL] = (
        1 - pd.to_numeric(res[REF_COL], errors="coerce")
    )
else:
    res[FIXED_REF_COL] = pd.to_numeric(
        res[FIXED_REF_COL],
        errors="coerce"
    )

res = res[res["hidden_demand_score"].notna()].copy()

# ============================================================
# CONDITION SETTINGS
# ============================================================

condition_order = [
    "image_only",
    "tabular_only",
    "naive_multimodal",
    "structured_multimodal",
]

condition_labels = {
    "image_only": "Image",
    "tabular_only": "Tabular",
    "naive_multimodal": "Naive MM",
    "structured_multimodal": "Structured MM",
}

# ============================================================
# REPRESENTATIVE STATION ENGLISH LABELS
# ============================================================

station_english = {
    "까치산": "Kkachisan",
    "단대오거리": "Dandaeogeori",
    "마천": "Macheon",
    "신정네거리": "Sinjeongnegeori",
    "홍제": "Hongje",
    "양천구청": "Yangcheon-gu Office",
    "구파발": "Gupabal",
    "성수": "Seongsu",
}

# ============================================================
# HELPER
# ============================================================

def to_bool(series):
    return (
        series.astype(str)
        .str.lower()
        .isin(["true", "1", "yes", "y"])
    )

# ============================================================
# A. ALIGNMENT SUMMARY
# ============================================================

align_rows = []

for cond in condition_order:

    sub = res[res["_condition"] == cond].copy()

    for ref_name, ref_col in [
        ("Gap", TARGET_COL),
        ("Reconstruction", FIXED_REF_COL),
    ]:

        mask = (
            sub["hidden_demand_score"].notna()
            &
            sub[ref_col].notna()
        )

        if mask.sum() >= 3:

            sp = spearmanr(
                sub.loc[mask, "hidden_demand_score"],
                sub.loc[mask, ref_col]
            ).correlation

        else:
            sp = np.nan

        align_rows.append({
            "condition": cond,
            "condition_label": condition_labels[cond],
            "reference": ref_name,
            "spearman": sp,
        })

align_df = pd.DataFrame(align_rows)

align_df.to_csv(
    FIG_DIR / "main_alignment_summary.csv",
    index=False,
    encoding="utf-8-sig"
)

display(align_df)

# ============================================================
# B/C. REASONING BEHAVIOR
# ============================================================

q50 = res[TARGET_COL].quantile(0.50)

bias_rows = []

for cond in condition_order:

    sub = res[res["_condition"] == cond].copy()

    visual = to_bool(sub["uses_visual_reasoning"])
    mobility = to_bool(sub["uses_mobility_subway_reasoning"])
    temporal = to_bool(sub["uses_temporal_reasoning"])

    unsupported_visual = (
        visual
        &
        (sub[TARGET_COL] < q50)
    )

    bias_rows.append({
        "condition": cond,
        "condition_label": condition_labels[cond],
        "visual_reasoning": visual.mean(),
        "mobility_subway_reasoning": mobility.mean(),
        "temporal_reasoning": temporal.mean(),
        "unsupported_visual_reasoning": unsupported_visual.mean(),
    })

bias_df = pd.DataFrame(bias_rows)

bias_df.to_csv(
    FIG_DIR / "main_reasoning_bias_summary.csv",
    index=False,
    encoding="utf-8-sig"
)

display(bias_df)

# ============================================================
# D. REPRESENTATIVE STATION ROBUSTNESS
# ============================================================

representative_stations = [
    "까치산",
    "단대오거리",
    "마천",
    "신정네거리",
    "홍제",
    "양천구청",
    "구파발",
    "성수",
]

robust = res.pivot_table(
    index=["_station_id", "_station_name"],
    columns="_condition",
    values="hidden_demand_score",
    aggfunc="mean"
).reset_index()

rep = robust[
    robust["_station_name"].isin(representative_stations)
].copy()

rep["station_order"] = rep["_station_name"].map(
    {s: i for i, s in enumerate(representative_stations)}
)

rep = rep.sort_values("station_order")

rep["_station_name_en"] = rep["_station_name"].map(
    lambda x: station_english.get(x, x)
)

rep.to_csv(
    FIG_DIR / "main_representative_station_scores.csv",
    index=False,
    encoding="utf-8-sig"
)

display(rep)

# ============================================================
# MAIN 4-PANEL FIGURE
# ============================================================

fig, axes = plt.subplots(
    2,
    2,
    figsize=(13, 8.5)
)

# ------------------------------------------------------------
# (a) Alignment
# ------------------------------------------------------------

ax = axes[0, 0]

plot_a = align_df.pivot(
    index="condition_label",
    columns="reference",
    values="spearman"
)

plot_a = plot_a.reindex(
    [condition_labels[c] for c in condition_order]
)

plot_a.plot(
    kind="bar",
    ax=ax,
    rot=0
)

ax.set_title("(a) Alignment with hidden-demand references")
ax.set_ylabel("Spearman correlation")
ax.set_xlabel("")
ax.axhline(0, linewidth=1)
ax.legend(title="")

# ------------------------------------------------------------
# (b) Unsupported visual reasoning
# ------------------------------------------------------------

ax = axes[0, 1]

plot_b = bias_df.set_index(
    "condition_label"
).reindex(
    [condition_labels[c] for c in condition_order]
)

plot_b["unsupported_visual_reasoning"].plot(
    kind="bar",
    ax=ax,
    rot=0
)

ax.set_title("(b) Unsupported visual reasoning")
ax.set_ylabel("Rate")
ax.set_xlabel("")
ax.set_ylim(0, 1)

# ------------------------------------------------------------
# (c) Evidence usage
# ------------------------------------------------------------

ax = axes[1, 0]

plot_c = plot_b[
    [
        "visual_reasoning",
        "mobility_subway_reasoning",
        "temporal_reasoning"
    ]
].rename(columns={
    "visual_reasoning": "Visual",
    "mobility_subway_reasoning": "Mobility/subway",
    "temporal_reasoning": "Temporal",
})

plot_c.plot(
    kind="bar",
    ax=ax,
    rot=0
)

ax.set_title("(c) Evidence usage by condition")
ax.set_ylabel("Rate")
ax.set_xlabel("")
ax.set_ylim(0, 1)
ax.legend(title="")

# ------------------------------------------------------------
# (d) Representative stations
# ------------------------------------------------------------

ax = axes[1, 1]

for c in condition_order:
    if c not in rep.columns:
        rep[c] = np.nan

plot_d = rep.set_index(
    "_station_name_en"
)[condition_order]

plot_d = plot_d.rename(
    columns=condition_labels
)

plot_d.plot(
    kind="bar",
    ax=ax,
    rot=30
)

ax.set_title("(d) Representative hidden-demand stations")
ax.set_ylabel("MLLM hidden-demand score")
ax.set_xlabel("")
ax.set_ylim(0, 5)
ax.legend(
    title="",
    fontsize=8
)

# ============================================================
# FINALIZE
# ============================================================

plt.tight_layout()

plt.savefig(
    FIG_DIR / "main_mllm_audit_4panel.png",
    dpi=300,
    bbox_inches="tight"
)

plt.savefig(
    FIG_DIR / "main_mllm_audit_4panel.pdf",
    bbox_inches="tight"
)

plt.show()

# ============================================================
# SAVE LATEX TABLES
# ============================================================

align_df.to_latex(
    FIG_DIR / "table_alignment_summary.tex",
    index=False,
    float_format="%.3f",
    escape=True
)

bias_df.to_latex(
    FIG_DIR / "table_reasoning_summary.tex",
    index=False,
    float_format="%.3f",
    escape=True
)

print("\nSaved files:")

for p in sorted(FIG_DIR.glob("*")):
    print("-", p.name)
```
