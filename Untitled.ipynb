```python
import os
import pandas as pd

OUT = "./final_rq_pour"
RQ = f"{OUT}/rq_outputs"

test = pd.read_csv(f"{OUT}/test_result.csv")
hist = pd.read_csv(f"{OUT}/train_history.csv")
bias = pd.read_csv(f"{RQ}/rq1_station_observation_bias.csv")
unc = pd.read_csv(f"{RQ}/test_station_uncertainty.csv")
merged = pd.read_csv(f"{RQ}/rq4_hidden_demand_uncertainty_merged.csv")

print("===== FINAL TEST RESULT =====")
display(test)

print("\n===== BEST VALIDATION EPOCH =====")
best_ep = hist.sort_values("val_mae").iloc[0]
display(best_ep.to_frame().T)

print("\n===== TRAINING TREND =====")
print("first val_mae:", hist["val_mae"].iloc[0])
print("best val_mae :", hist["val_mae"].min())
print("last val_mae :", hist["val_mae"].iloc[-1])
print("test mae     :", test["mae"].iloc[0])
print("test rmse    :", test["rmse"].iloc[0])
print("test corr    :", test["uncertainty_error_corr"].iloc[0])

print("\n===== RQ1: OBSERVATION BIAS SUMMARY =====")
rq1_cols = ["coverage", "sparsity", "temporal_entropy", "hidden_demand_score", "observation_bias_score"]
display(bias[rq1_cols].describe())

print("\nTop hidden-demand stations")
display(
    bias.sort_values("hidden_demand_score", ascending=False)
    .head(20)
)

print("\nTop observation-bias stations")
display(
    bias.sort_values("observation_bias_score", ascending=False)
    .head(20)
)

print("\n===== RQ3/RQ4: UNCERTAINTY SUMMARY =====")
display(unc[["mean_abs_error", "mean_uncertainty"]].describe())

print("\nTop high-uncertainty stations")
display(
    unc.sort_values("mean_uncertainty", ascending=False)
    .head(20)
)

print("\nTop high-error stations")
display(
    unc.sort_values("mean_abs_error", ascending=False)
    .head(20)
)

print("\n===== RQ4: HIDDEN DEMAND × UNCERTAINTY =====")
corr_hidden_unc = merged["hidden_demand_score"].corr(merged["mean_uncertainty"])
corr_bias_unc = merged["observation_bias_score"].corr(merged["mean_uncertainty"])
corr_sparsity_unc = merged["sparsity"].corr(merged["mean_uncertainty"])
corr_error_unc = merged["mean_abs_error"].corr(merged["mean_uncertainty"])

print("corr(hidden_demand, uncertainty):", corr_hidden_unc)
print("corr(observation_bias, uncertainty):", corr_bias_unc)
print("corr(sparsity, uncertainty):", corr_sparsity_unc)
print("corr(error, uncertainty):", corr_error_unc)

display(
    merged.sort_values("mean_uncertainty", ascending=False)
    .head(30)
)
```
